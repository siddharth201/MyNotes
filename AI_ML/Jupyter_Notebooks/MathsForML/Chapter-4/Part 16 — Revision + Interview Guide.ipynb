## 1. The Starting Point — A Linear Decision Boundary

A linear classifier uses a decision boundary of the form:

$$
\boxed{
\vec{w}^{\,T}\vec{x}+b=0
}
$$

where:

- $\vec{x}$ is the input feature vector.
- $\vec{w}$ is the weight vector.
- $b$ is the bias.
- $\vec{w}^{\,T}\vec{x}+b$ is the linear score.

The decision boundary consists of all points for which:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

## 2. The Score

For a particular point $\vec{x}$, we calculate:

$$
\boxed{
s(\vec{x})=\vec{w}^{\,T}\vec{x}+b
}
$$

This is the **linear decision score**.

The score is not itself the perpendicular distance from the boundary.

Its most important role at this stage is that its **sign identifies the side of the decision boundary**.

If:

$$
\vec{w}^{\,T}\vec{x}+b>0
$$

the point lies on one side.

If:

$$
\vec{w}^{\,T}\vec{x}+b<0
$$

the point lies on the opposite side.

If:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

the point lies exactly on the decision boundary.

## 3. Why the Sign Matters

The sign gives information that the magnitude alone cannot provide.

For example:

$$
+5
$$

and:

$$
-5
$$

have the same magnitude:

$$
|+5|=|-5|=5
$$

but they represent opposite directions.

Therefore:

$$
\boxed{
\text{Magnitude tells us how much.}
}
$$

while:

$$
\boxed{
\text{Sign tells us the side/direction.}
}
$$

In classification, this distinction is important because the decision boundary divides the space into two sides.

## 4. Absolute Value

Absolute value preserves magnitude but removes the sign.

For example:

$$
|+4|=4
$$

and:

$$
|-4|=4
$$

Therefore:

$$
\boxed{
|+4|=|-4|
}
$$

The two values become indistinguishable if we only retain their absolute values.

This is why absolute value must be used carefully when the sign contains meaningful classification information.

## 5. Score vs Distance

One of the most important distinctions from this chapter is:

$$
\boxed{
\vec{w}^{\,T}\vec{x}+b
}
$$

is a **signed linear score**.

It is not the actual perpendicular distance.

The signed perpendicular distance is:

$$
\boxed{
d(\vec{x})
=
\frac{
\vec{w}^{\,T}\vec{x}+b
}{
\|\vec{w}\|
}
}
$$

The denominator $\|\vec{w}\|$ is necessary to convert the linear score into geometric distance.

Therefore, never casually say:

> "$\vec{w}^{\,T}\vec{x}+b$ is the distance."

The mathematically precise statement is:

> "$\vec{w}^{\,T}\vec{x}+b$ is the signed linear score, while dividing by $\|\vec{w}\|$ gives the signed perpendicular distance."

## 6. Signed vs Absolute Distance

The signed perpendicular distance is:

$$
\boxed{
d(\vec{x})
=
\frac{
\vec{w}^{\,T}\vec{x}+b
}{
\|\vec{w}\|
}
}
$$

It contains two pieces of information:

- magnitude → how far from the boundary
- sign → which side of the boundary

If we take its absolute value:

$$
\boxed{
|d(\vec{x})|
=
\left|
\frac{
\vec{w}^{\,T}\vec{x}+b
}{
\|\vec{w}\|
}
\right|
}
$$

we obtain the ordinary geometric distance.

Therefore:

$$
\boxed{
\text{Signed distance}
=
\text{magnitude + side information}
}
$$

while:

$$
\boxed{
\text{Absolute distance}
=
\text{magnitude only}
}
$$

## 7. Introducing the Class Label $y_i$

In binary classification, we use the class-label convention:

$$
\boxed{
y_i\in\{-1,+1\}
}
$$

Each training example consists of:

$$
\boxed{
(\vec{x}_i,y_i)
}
$$

where:

- $\vec{x}_i$ is the input vector.
- $y_i$ is the true class label.

The labels $+1$ and $-1$ are particularly convenient because they allow the true class to interact directly with the classifier score.

## 8. The Unified Classification Expression

The classifier score for example $i$ is:

$$
\vec{w}^{\,T}\vec{x}_i+b
$$

The true class label is:

$$
y_i\in\{-1,+1\}
$$

Multiplying them gives:

$$
\boxed{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}
$$

This expression is extremely useful because it puts the true label and the classifier's score into a single quantity.

For a correctly classified point, this quantity is positive.

For a misclassified point, this quantity is negative.

A larger positive value indicates that the point is not merely on the correct side, but has a larger positive score relative to its label.

## 9. Functional Margin

The quantity:

$$
\boxed{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}
$$

is the **functional margin** of example $i$.

Consider the three cases.

### Correct classification

If:

$$
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)>0
$$

the example is correctly classified.

### Decision boundary

If:

$$
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)=0
$$

the example lies on the decision boundary.

### Incorrect classification

If:

$$
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)<0
$$

the example is misclassified.

## 10. Why Do We Divide by $\|\vec{w}\|$?

The functional margin depends on the scale of $\vec{w}$ and $b$.

The geometric distance should represent an actual distance in the feature space.

Therefore, we normalize by the magnitude of the weight vector:

$$
\|\vec{w}\|
$$

This gives the geometric margin:

$$
\boxed{
\gamma_i
=
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
}
$$

This quantity incorporates the classification direction and expresses the result in geometric distance units.

## 12. The Three Quantities You Must Never Confuse

| Quantity | Expression | Meaning |
|---|---|---|
| Decision score | $\vec{w}^{\,T}\vec{x}+b$ | Signed linear score |
| Signed perpendicular distance | $\dfrac{\vec{w}^{\,T}\vec{x}+b}{\|\vec{w}\|}$ | Geometric distance with side information |
| Absolute perpendicular distance | $\left|\dfrac{\vec{w}^{\,T}\vec{x}+b}{\|\vec{w}\|}\right|$ | Geometric distance without side information |

And with the class label:

| Quantity | Expression | Meaning |
|---|---|---|
| Functional margin | $y_i(\vec{w}^{\,T}\vec{x}_i+b)$ | Class-aware signed score |
| Geometric margin | $\dfrac{y_i(\vec{w}^{\,T}\vec{x}_i+b)}{\|\vec{w}\|}$ | Class-aware geometric quantity |

## 15. 🎯 Interview Answers

### Answer 1 — What is a linear decision boundary?

A linear decision boundary is the set of points satisfying:

$$
\boxed{
\vec{w}^{\,T}\vec{x}+b=0
}
$$

It separates the feature space into two regions.

---

### Answer 2 — What does $\vec{w}^{\,T}\vec{x}+b$ represent?

It is the **linear decision score**.

Its sign indicates which side of the decision boundary the point lies on.

---

### Answer 3 — Is it the perpendicular distance?

No.

The signed perpendicular distance is:

$$
\boxed{
\frac{
\vec{w}^{\,T}\vec{x}+b
}{
\|\vec{w}\|
}
}
$$

---

### Answer 4 — Why divide by $\|\vec{w}\|$?

Because the raw score depends on the scale of the weight vector.

Dividing by $\|\vec{w}\|$ converts the signed score into geometric perpendicular distance.

---

### Answer 5 — What does the sign tell us?

The sign tells us which side of the decision boundary the point lies on.

---

### Answer 6 — Why can absolute value be problematic?

Because:

$$
|+a|=|-a|
$$

Absolute value preserves magnitude but removes the sign.

If the sign represents which side of the decision boundary a point occupies, taking the absolute value loses that information.

---

### Answer 7 — Why use $y_i\in\{-1,+1\}$?

This encoding allows the true class label to interact naturally with the classifier score:

$$
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
$$

This makes it easy to determine whether the classifier's side agrees with the true class.

---

### Answer 8 — What does the unified expression represent?

$$
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
$$

is the **functional margin**.

It combines the true class label with the classifier's score.

---

### Answer 9 — What does a positive functional margin mean?

$$
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)>0
$$

means the example is correctly classified.

---

### Answer 10 — What does a negative functional margin mean?

$$
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)<0
$$

means the example is misclassified.

---

### Answer 11 — What happens when the score is zero?

If:

$$
\vec{w}^{\,T}\vec{x}_i+b=0
$$

the point lies exactly on the decision boundary.

---

### Answer 12 — Functional vs geometric margin?

The functional margin is:

$$
\boxed{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}
$$

The geometric margin is:

$$
\boxed{
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
}
$$

The geometric version accounts for the scale of $\vec{w}$ and expresses the quantity in geometric distance terms.

## 16. ⚠️ Common Interview Traps

### Trap 1

> "$\vec{w}^{\,T}\vec{x}+b$ is the distance."

❌ Not precisely correct.

It is the **signed linear score**.

The signed perpendicular distance is:

$$
\frac{
\vec{w}^{\,T}\vec{x}+b
}{
\|\vec{w}\|
}
$$

---

### Trap 2

> "Absolute distance tells us everything about the point."

❌ Incorrect.

It tells us how far the point is from the boundary, but not which side it lies on.

---

### Trap 3

> "Absolute value is bad for Machine Learning."

❌ Incorrect.

Absolute value is useful whenever only magnitude matters.

The issue occurs when the sign contains information that we need.

---

### Trap 4

> "$+4$ and $-4$ are the same."

❌ Incorrect.

They have the same magnitude:

$$
|+4|=|-4|
$$

but opposite signs and therefore different positions relative to a one-dimensional boundary at zero.

---

### Trap 5

> "A positive score always means class $+1$."

⚠️ This depends on the label convention and classifier interpretation.

With the convention:

$$
y_i\in\{-1,+1\}
$$

the score sign is compared with the class label.

The class-aware expression is:

$$
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
$$
