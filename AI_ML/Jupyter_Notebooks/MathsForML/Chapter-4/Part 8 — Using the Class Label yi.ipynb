# Using the Class Label $y_i$

> **Core Question:**  
> If absolute value removes the sign, can we use the class label $y_i$ to preserve the information about which side of the decision boundary is correct?

In the previous part, we discovered an important problem:

$$
|+4|=|-4|=4
$$

Absolute value gives us the distance from the decision boundary, but it removes the sign.

For classification, however, the sign is important because it tells us which side of the decision boundary a point lies on.

In this part, we will use the **class label $y_i$** to handle this problem.

The key idea is:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

The class label tells us which sign should be considered the **correct direction** for that particular training example.

## 1. First, What Is $y_i$?

In supervised classification, each training example consists of two important pieces of information:

- an input vector, represented by $\vec{x}_i$
- a corresponding class label, represented by $y_i$

We can represent one training example as:

$$
(\vec{x}_i, y_i)
$$

Here:

- $\vec{x}_i$ represents the input features of the $i$-th example.
- $y_i$ represents the true class of that example.

For binary classification, we will use the convention:

$$
y_i \in \{-1,+1\}
$$

So there are two possible classes:

| Class | Label |
|---|---:|
| Class 1 | $+1$ |
| Class 2 | $-1$ |

The $+1/-1$ representation is particularly useful when working with linear classifiers, decision boundaries, margins, and support vector machines.

## 2. Why Use \(-1\) and \(+1\)?

At first, using \(-1\) and \(+1\) instead of \(0\) and \(1\) may seem arbitrary.

But there is a beautiful mathematical reason.

Suppose the decision boundary is:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

The classifier can use the sign of the score:

$$
\vec{w}^{\,T}\vec{x}+b>0
$$

for one side, and:

$$
\vec{w}^{\,T}\vec{x}+b<0
$$

for the other side.

Therefore, the natural class labels are:

$$
+1
\qquad\text{and}\qquad
-1
$$

The label can then be used to tell us which sign is the **correct** sign for a particular example.

## 3. The Problem We Had in Part 7

Suppose two training examples produce these signed scores:

$$
+4
$$

and:

$$
-4
$$

If we take absolute values:

$$
|+4|=|-4|=4
$$

we lose the information about which sign was correct.

This is the problem we want to solve.

We do not merely want to know:

> "How large is the score?"

We also want to know:

> "Is the score pointing in the correct direction for this class?"

## 4. The Class Label Gives Us the Missing Information

Suppose the training example has label:

$$
y_i=+1
$$

Then we want the classifier score to be positive:

$$
\vec{w}^{\,T}\vec{x}_i+b>0
$$

Now suppose the training example has label:

$$
y_i=-1
$$

Then we want the classifier score to be negative:

$$
\vec{w}^{\,T}\vec{x}_i+b<0
$$

So the class label tells us what the **correct sign** should be.

This gives us the idea:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

Instead of throwing away the sign using absolute value, we use the label to **align the score with the correct class direction**.

## 5. Let's See the Magic

Consider a correctly classified positive example.

Its label is:

$$
y_i=+1
$$

and suppose its score is:

$$
\vec{w}^{\,T}\vec{x}_i+b=4
$$

Then:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
=
(+1)(4)
=
4
$$

Now consider a correctly classified negative example.

Its label is:

$$
y_i=-1
$$

and suppose its score is:

$$
\vec{w}^{\,T}\vec{x}_i+b=-4
$$

Then:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
=
(-1)(-4)
=
4
$$

Look at what happened:

| Example | $y_i$ | Score | $y_i(\vec{w}^{\,T}\vec{x}_i+b)$ |
|---|---:|---:|---:|
| Positive | \(+1\) | \(+4\) | \(+4\) |
| Negative | \(-1\) | \(-4\) | \(+4\) |

Both correctly classified examples now produce a **positive value**.

This is exactly what we wanted.

## 6. Why Is This Better Than Absolute Value?

Compare the two approaches.

### Absolute value

$$
\left|
\vec{w}^{\,T}\vec{x}_i+b
\right|
$$

This always produces a non-negative value.

So it cannot tell us whether the point was correctly or incorrectly classified.

---

### Label-adjusted score

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

This behaves differently.

For a correctly classified point:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
$$

For a misclassified point:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)<0
$$

Therefore:

$$
\boxed{
\text{The class label preserves information that absolute value destroys.}
}
$$

\vec{w}^{\,T}\vec{x}_i+b=+5

$$

+5

y_i=-1

$$

Therefore:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
=
(-1)(-5)
=
+5
$$

---

### Case 3 — Positive example, incorrectly classified

$$
y_i=+1
$$

but:

## 8. The Important Pattern

The expression:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

has a very useful property.

### Positive value

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
$$

means the point is classified correctly.

### Negative value

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)<0
$$

means the point is classified incorrectly.

### Zero

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)=0
$$

means the point lies exactly on the decision boundary.

Therefore:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
\text{ tells us whether the classifier's score agrees with the true label.}
}
$$

## 9. From Absolute Value to Class-Adjusted Score

In Part 7, we had:

$$
\left|
\vec{w}^{\,T}\vec{x}_i+b
\right|
$$

The problem was that absolute value removes the sign.

Now we introduce the class label:

$$
y_i\in\{-1,+1\}
$$

and use:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

The conceptual progression is:

| Step | Expression | What it tells us |
|---|---|---|
| Decision boundary | $\vec{w}^{\,T}\vec{x}+b=0$ | Separates the two sides |
| Score | $\vec{w}^{\,T}\vec{x}_i+b$ | Which side the point lies on |
| Absolute score | $\lvert\vec{w}^{\,T}\vec{x}_i+b\rvert$ | Magnitude only |
| Class-adjusted score | $y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)$ | Correct direction + magnitude |

The key transition is:

$$
\boxed{
\left|\vec{w}^{\,T}\vec{x}_i+b\right|
\quad\longrightarrow\quad
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

The class label lets us preserve the information about whether the score has the **correct sign**.

## 10. A Very Important Mathematical Nuance

There is an important distinction between:

$$
\left|
\vec{w}^{\,T}\vec{x}_i+b
\right|
$$

and:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

They are equal **only when the example is correctly classified**.

For a correctly classified example:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
$$

and therefore:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
=
\left|
\vec{w}^{\,T}\vec{x}_i+b
\right|
$$

But for a misclassified example:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)<0
$$

while:

$$
\left|
\vec{w}^{\,T}\vec{x}_i+b
\right|>0
$$

This is actually an advantage.

The class-adjusted score does **not hide a mistake**.

It becomes negative and tells us:

> The classifier's prediction is pointing in the wrong direction.
This distinction is especially important for understanding margins and SVMs. The standard geometric-margin expression uses $y_i$ precisely to account for the correct class direction.

## 11. From Functional Margin to Geometric Margin

We now have:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

This quantity is called the **functional margin** of the training example.

It tells us how strongly the example agrees with the classifier's correct class direction.

However, it is not yet the actual perpendicular distance from the decision boundary.

To obtain the geometric margin, we divide by the magnitude of the normal vector:

$$
\|\vec{w}\|
$$

Therefore:

$$
\boxed{
\gamma_i
=
\frac{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}{
\|\vec{w}\|
}
}
$$

This is the **geometric margin** of the example.

It combines:

- the correct class direction through \(y_i\)
- the classifier score
- the actual geometric scaling through \(\|\vec{w}\|\)

## 12. Why Do We Divide by $\|\vec{w}\|$?

Recall from the previous part:

$$
\text{Signed perpendicular distance}
=
\frac{
\vec{w}^{\,T}\vec{x}_i+b
}{
\|\vec{w}\|
}
$$

The reason for dividing by $\|\vec{w}\|$ is that the raw score depends on the scale of $\vec{w}$.

For example, the decision boundary:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

and:

$$
2\vec{w}^{\,T}\vec{x}+2b=0
$$

represent the **same geometric boundary**.

But the raw scores are different by a factor of 2.

Dividing by \(\|\vec{w}\|\) removes this scaling effect and gives us a genuine geometric distance.

Therefore:

$$
\boxed{
\text{Functional margin}
\neq
\text{Geometric margin}
}
$$

The geometric margin is:

$$
\boxed{
\gamma_i
=
\frac{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}{
\|\vec{w}\|
}
}
$$

## 13. Python Verification

```python
import numpy as np

# Two training examples
labels = np.array([+1, -1])

# Scores produced by the classifier
scores = np.array([+6, -6])

# Norm of the weight vector
w_norm = 2

# Functional margin
functional_margin = labels * scores

# Geometric margin
geometric_margin = functional_margin / w_norm

print("Labels:", labels)
print("Scores:", scores)
print("Functional margins:", functional_margin)
print("Geometric margins:", geometric_margin)
```

Notice what happened.
The raw scores were:
$$[+6,-6]$$
The class labels were:
$$[+1,-1]$$
Multiplying them gives:
$$[+6,+6]$$
Both examples are correctly classified, so both functional margins are positive.
After dividing by:
$$|\vec{w}|=2$$
we obtain:
$$[3,3]$$

Both examples are therefore 3 units from the decision boundary in the **class-oriented direction**.

## 14. What Happens When a Point Is Misclassified?

```python
import numpy as np

# First example is correctly classified.
# Second example is incorrectly classified.
labels = np.array([+1, -1])
scores = np.array([+6, +6])

functional_margin = labels * scores

print("Labels:", labels)
print("Scores:", scores)
print("Functional margins:", functional_margin)
```

The first example:
$$(+1)(+6)=+6$$
is correctly classified.
The second example:
$$(-1)(+6)=-6$$
is incorrectly classified.
This is something absolute value could not tell us:
$$|+6|=6$$
The absolute value would hide the mistake.
The class-adjusted score exposes it.

## 16. One Critical Caveat — $y_i$ Must Be \(-1/+1\)

The expression:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

works in this form because:

$$
y_i\in\{-1,+1\}
$$

Suppose instead that your dataset uses:

$$
y_i\in\{0,1\}
$$

Then multiplying by \(y_i\) does not provide the desired behavior.

For example:

$$
0\times
\left(\vec{w}^{\,T}\vec{x}_i+b\right)
=
0
$$

which destroys the score entirely.

Therefore, for this mathematical formulation we use:

$$
\boxed{
y_i\in\{-1,+1\}
}
$$

If a binary dataset is originally encoded as \(0/1\), we can convert it using:

$$
\boxed{
y_i^{(\pm1)}=2y_i^{(0/1)}-1
}
$$

So:

$$
0\rightarrow -1
$$

and:

$$
1\rightarrow +1
$$

## 17. Final Mental Model

Think of the classifier score as an arrow pointing in one of two directions.

The true class label tells us which direction is **correct**.

If the score and label agree:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
$$

the example is correctly classified.

If they disagree:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)<0
$$

the example is incorrectly classified.

Therefore:

$$
\boxed{
y_i
\text{ aligns the classifier score with the true class.}
}
$$

And after normalization:

$$
\boxed{
\gamma_i
=
\frac{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}{
\|\vec{w}\|
}
}
$$

we obtain the class-oriented geometric distance.
