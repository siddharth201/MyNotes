# Chapter 4 — Part 9

# The Unified Classification Expression

> **Core Question:**  
> Can we represent the classification behavior of both classes using one single mathematical expression?

In the previous parts, we developed several important ideas.

A linear decision boundary is:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

For a point $\vec{x}_i$, the classifier produces the score:

$$
\vec{w}^{\,T}\vec{x}_i+b
$$

The sign of this score tells us which side of the decision boundary the point lies on.

For binary classification, we also introduced the class label:

$$
y_i\in\{-1,+1\}
$$

The class label tells us which side is the correct side for that particular training example.

This allows us to combine the classifier score and the true class into one expression:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

This is the central expression we will develop in this part.

## 1. The Two Classes Create Two Cases

Let's first look at the two classes separately.

For a positive example:

$$
y_i=+1
$$

we want the classifier score to be positive:

$$
\vec{w}^{\,T}\vec{x}_i+b>0
$$

For a negative example:

$$
y_i=-1
$$

we want the classifier score to be negative:

$$
\vec{w}^{\,T}\vec{x}_i+b<0
$$

So the desired behavior is:

| True class | Desired score |
|---|---|
| $y_i=+1$ | Positive |
| $y_i=-1$ | Negative |

The problem is that we currently have two separate cases.

We want to express both cases with one rule.

## 2. The Problem with Writing Two Separate Rules

We could describe correct classification using two separate statements.

For $y_i=+1$:

$$
\vec{w}^{\,T}\vec{x}_i+b>0
$$

For $y_i=-1$:

$$
\vec{w}^{\,T}\vec{x}_i+b<0
$$

These statements are correct, but they require us to keep switching between two cases.

Mathematically, it would be much more convenient if we could write one expression that handles both classes automatically.

The class label $y_i$ gives us exactly the information we need.

## 3. Let the Label Correct the Direction

Consider the expression:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

The important idea is that multiplying by $y_i$ makes the score **class-oriented**.

Let's see what happens.

### Positive class

If:

$$
y_i=+1
$$

then:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
=
\vec{w}^{\,T}\vec{x}_i+b
$$

Nothing changes.

So a positive score remains positive.

---

### Negative class

If:

$$
y_i=-1
$$

then:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
=
-\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

The sign is reversed.

Therefore, a correctly classified negative example, whose original score is negative, becomes positive after multiplication by $-1$.

## 4. The Two Cases Become One Rule

Now consider what happens to a correctly classified example.

### Positive example

$$
y_i=+1
$$

and:

$$
\vec{w}^{\,T}\vec{x}_i+b>0
$$

Therefore:

$$
(+1)\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
$$

---

### Negative example

$$
y_i=-1
$$

and:

$$
\vec{w}^{\,T}\vec{x}_i+b<0
$$

Therefore:

$$
(-1)\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
$$

Notice what happened.

Although the two classes originally require opposite signs, after multiplying by their respective labels, **both correctly classified examples produce a positive value**.

Therefore we can write one unified condition:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
}
$$

This single expression replaces the two separate classification conditions.

## 5. Let's Verify All Four Possibilities

Let's examine both classes and both possible classifier outcomes.

| True class | Classifier score | $y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)$ | Result |
|---|---:|---:|---|
| $+1$ | Positive | Positive | Correct |
| $+1$ | Negative | Negative | Incorrect |
| $-1$ | Negative | Positive | Correct |
| $-1$ | Positive | Negative | Incorrect |

This gives us a very powerful rule:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
\quad\Longrightarrow\quad
\text{correct classification}
}
$$

and:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)<0
\quad\Longrightarrow\quad
\text{incorrect classification}
}
$$

If:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)=0
$$

then the point lies exactly on the decision boundary.

## 6. Why This Expression Is So Useful

The expression:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

combines two pieces of information:

### The classifier score

$$
\vec{w}^{\,T}\vec{x}_i+b
$$

tells us where the point lies relative to the decision boundary.

### The class label

$$
y_i
$$

tells us which direction is correct for that example.

Multiplying them together gives us a **class-oriented score**.

Therefore:

$$
\boxed{
\text{classifier score}
\times
\text{true class label}
=
\text{class-oriented score}
}
$$

This is why the expression works for both classes without requiring separate cases.

## 7. A Numerical Example

Suppose we have two training examples.

### Example 1

The true class is:

$$
y_1=+1
$$

and the classifier produces:

$$
\vec{w}^{\,T}\vec{x}_1+b=5
$$

Therefore:

$$
y_1\left(\vec{w}^{\,T}\vec{x}_1+b\right)
=
(+1)(5)
=
5
$$

The result is positive.

So the example is correctly classified.

---

### Example 2

The true class is:

$$
y_2=-1
$$

and the classifier produces:

$$
\vec{w}^{\,T}\vec{x}_2+b=-5
$$

Therefore:

$$
y_2\left(\vec{w}^{\,T}\vec{x}_2+b\right)
=
(-1)(-5)
=
5
$$

Again, the result is positive.

So this example is also correctly classified.

Notice that the original scores were:

$$
+5
\qquad\text{and}\qquad
-5
$$

but the class-adjusted scores are both:

$$
+5
\qquad\text{and}\qquad
+5
$$

The class label has aligned both examples into the same "correct direction."

## 8. What Happens When the Classifier Is Wrong?

Suppose:

$$
y_i=+1
$$

but the classifier produces:

$$
\vec{w}^{\,T}\vec{x}_i+b=-5
$$

Then:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
=
(+1)(-5)
=
-5
$$

The result is negative.

This tells us that the classifier score points in the wrong direction for the true class.

Now consider:

$$
y_i=-1
$$

but:

$$
\vec{w}^{\,T}\vec{x}_i+b=+5
$$

Then:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
=
(-1)(+5)
=
-5
$$

Again, the result is negative.

Therefore:

$$
\boxed{
\text{Negative class-adjusted score}
\Rightarrow
\text{incorrect classification}
}
$$

## 9. The Unified Classification Condition

We can now express the classification rule for both classes using one condition.

For a correctly classified training example:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
}
$$

For an incorrectly classified training example:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)<0
}
$$

For a point exactly on the decision boundary:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)=0
}
$$

Therefore, the sign of:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

completely captures whether the classifier's prediction agrees with the true class.

## 10. Connection to Part 8 — Functional Margin

In Part 8, we introduced:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

This quantity has a formal name:

$$
\boxed{
\hat{\gamma}_i
=
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

It is called the **functional margin** of the classifier with respect to the training example $(\vec{x}_i,y_i)$.

The functional margin tells us both:

- whether the example is classified correctly
- how strongly the classifier score agrees with the true class

A positive functional margin means correct classification.

A negative functional margin means incorrect classification.

A larger positive functional margin means the classifier score is more strongly aligned with the true class.

This is the standard definition used in the development of support vector machines. :contentReference[oaicite:1]{index=1}

## 11. But Is Functional Margin a True Distance?

No.

This is an important distinction.

The functional margin is:

$$
\boxed{
\hat{\gamma}_i
=
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

It is a **score**, not a geometric distance.

Why?

Because its value depends on the scale of $\vec{w}$ and $b$.

Suppose we multiply both by $2$:

$$
\vec{w}'=2\vec{w}
$$

and:

$$
b'=2b
$$

Then the decision boundary becomes:

$$
\vec{w}'^{\,T}\vec{x}+b'
=
2\vec{w}^{\,T}\vec{x}+2b
=
2\left(\vec{w}^{\,T}\vec{x}+b\right)
$$

The equation:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

and the equation:

$$
2\vec{w}^{\,T}\vec{x}+2b=0
$$

describe the **same decision boundary**.

But the functional margin doubles.

Therefore:

$$
\boxed{
\text{Functional margin is not scale-invariant.}
}
$$

This is why we need geometric margin when we want an actual geometric distance.

## 12. From Functional Margin to Geometric Margin

From the previous part, we know that the signed perpendicular distance to the decision boundary is obtained by dividing by:

$$
\|\vec{w}\|
$$

Therefore, the geometric margin of a training example is:

$$
\boxed{
\gamma_i
=
\frac{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}{
\|\vec{w}\|
}
}
$$

So we now have a complete hierarchy:

| Quantity | Expression | Meaning |
|---|---|---|
| Classifier score | $\vec{w}^{\,T}\vec{x}_i+b$ | Side of the boundary |
| Functional margin | $y_i(\vec{w}^{\,T}\vec{x}_i+b)$ | Class-oriented score |
| Geometric margin | $\dfrac{y_i(\vec{w}^{\,T}\vec{x}_i+b)}{\|\vec{w}\|}$ | Class-oriented perpendicular distance |

The geometric margin is the functional margin normalized by the length of $\vec{w}$.

## 13. The Unified Expression

We can now see why:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

is such an important expression.

It takes the two separate classification cases:

$$
\begin{aligned}
y_i=+1 &\Rightarrow \vec{w}^{\,T}\vec{x}_i+b>0 \\
y_i=-1 &\Rightarrow \vec{w}^{\,T}\vec{x}_i+b<0
\end{aligned}
$$

and converts them into one unified condition:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
}
$$

This is the mathematical reason we use labels $y_i\in\{-1,+1\}$ when developing linear classifiers and support vector machines.

Instead of asking two different questions:

> Is the score positive for the positive class?

and:

> Is the score negative for the negative class?

we ask one question:

> **Does the classifier score have the correct sign for the true class?**

That question is captured by:

$$
\boxed{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
}
$$

## 14. Python Verification

````python
import numpy as np

# True class labels
y = np.array([+1, -1, +1, -1])

# Classifier scores
scores = np.array([+5, -5, -3, +3])

# Unified class-adjusted expression
class_adjusted_scores = y * scores

print("Labels:", y)
print("Scores:", scores)
print("Class-adjusted scores:", class_adjusted_scores)
````

### Expected Output

```text
Labels: [ 1 -1  1 -1]
Scores: [ 5 -5 -3  3]
Class-adjusted scores: [ 5  5 -3 -3]

Now interpret the results:

$[5,5,−3,−3]$

The first two examples have positive class-adjusted scores, so they are correctly classified.

The last two examples have negative class-adjusted scores, so they are incorrectly classified.

Therefore:

$$
\boxed{
sign(y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right))
}
$$
	​


tells us whether the classifier agrees with the true class.
