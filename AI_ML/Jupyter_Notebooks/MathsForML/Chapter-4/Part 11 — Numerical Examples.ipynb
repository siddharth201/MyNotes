# Chapter 4 — Part 11

# Numerical Examples

> **Goal:**  
> Use concrete numerical examples to calculate and interpret the classifier score, functional margin, and geometric margin.

In the previous parts, we developed the following expressions.

### Classifier score

$$
s_i=\vec{w}^{\,T}\vec{x}_i+b
$$

### Functional margin

$$
\hat{\gamma}_i
=
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
$$

### Geometric margin

$$
\gamma_i
=
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
$$

In this part, we will calculate these quantities step by step using numerical examples.

The goal is to move from:

> **"I understand the formula."**

to:

> **"I can calculate it and explain what the result means."**

## 1. Example 1 — A Simple One-Dimensional Classifier

Consider the one-dimensional decision boundary:

$$
x=0
$$

We can represent this boundary using:

$$
wx+b=0
$$

Choose:

$$
w=1
$$

and:

$$
b=0
$$

Therefore:

$$
x=0
$$

is our decision boundary.

Suppose we have a training example:

$$
x_1=4
$$

with class label:

$$
y_1=+1
$$

### Step 1 — Calculate the Classifier Score

The classifier score is:

$$
s_1=wx_1+b
$$

Substituting the values:

$$
s_1=(1)(4)+0
$$

Therefore:

$$
\boxed{
s_1=4
}
$$

The score is positive.

Since the true class is:

$$
y_1=+1
$$

the positive score agrees with the true class.

Therefore, the example is correctly classified.

### Step 2 — Calculate the Functional Margin

The functional margin is:

$$
\hat{\gamma}_1
=
y_1(wx_1+b)
$$

Substituting:

$$
\hat{\gamma}_1
=
(+1)(4)
$$

Therefore:

$$
\boxed{
\hat{\gamma}_1=4
}
$$

The functional margin is positive, so the example is correctly classified.

### Step 3 — Calculate the Geometric Margin

The geometric margin is:

$$
\gamma_1
=
\frac{
y_1(wx_1+b)
}{
\|w\|
}
$$

Since this is a one-dimensional example:

$$
\|w\|=|w|=1
$$

Therefore:

$$
\gamma_1
=
\frac{4}{1}
$$

So:

$$
\boxed{
\gamma_1=4
}
$$

This means the point is 4 units away from the decision boundary in the class-oriented direction.

## 2. Example 2 — A Negative Class Example

Let's now consider a point on the other side of the boundary.

The decision boundary is still:

$$
x=0
$$

and:

$$
w=1,\qquad b=0
$$

Now choose:

$$
x_2=-4
$$

with class label:

$$
y_2=-1
$$

### Step 1 — Calculate the Classifier Score

The score is:

$$
s_2=wx_2+b
$$

Substituting:

$$
s_2=(1)(-4)+0
$$

Therefore:

$$
\boxed{
s_2=-4
}
$$

The score is negative.

Since the true class is:

$$
y_2=-1
$$

the score has the correct sign.

Therefore, the example is correctly classified.

### Step 2 — Calculate the Functional Margin

The functional margin is:

$$
\hat{\gamma}_2
=
y_2(wx_2+b)
$$

Substituting:

$$
\hat{\gamma}_2
=
(-1)(-4)
$$

Therefore:

$$
\boxed{
\hat{\gamma}_2=4
}
$$

Notice something important.

The two examples had different raw scores:

$$
+4
\qquad\text{and}\qquad
-4
$$

But their functional margins are both:

$$
+4
$$

The class label has aligned both examples with the correct class direction.

### Step 3 — Calculate the Geometric Margin

The geometric margin is:

$$
\gamma_2
=
\frac{
y_2(wx_2+b)
}{
\|w\|
}
$$

Since:

$$
\|w\|=1
$$

we get:

$$
\gamma_2
=
\frac{4}{1}
$$

Therefore:

$$
\boxed{
\gamma_2=4
}
$$

So both examples are 4 units from the decision boundary in their respective correct directions.

## 3. Compare the Two Examples

Let's put the two examples together.

| Example | $x_i$ | $y_i$ | Score | Functional margin | Geometric margin |
|---|---:|---:|---:|---:|---:|
| 1 | $+4$ | $+1$ | $+4$ | $+4$ | $+4$ |
| 2 | $-4$ | $-1$ | $-4$ | $+4$ | $+4$ |

The raw classifier scores are different:

$$
+4
\qquad\text{and}\qquad
-4
$$

because the points are on opposite sides of the decision boundary.

But after incorporating the true class label:

$$
y_i(wx_i+b)
$$

both examples produce:

$$
+4
$$

This is exactly why the class label is so useful.

It lets us express the two class-specific cases using one common quantity.

## 4. Example 3 — An Incorrectly Classified Point

Let's now deliberately choose a point whose class does not agree with the classifier.

Keep:

$$
w=1
$$

and:

$$
b=0
$$

Suppose:

$$
x_3=4
$$

but its true class is:

$$
y_3=-1
$$

The point is on the positive side of the decision boundary, but its true class is negative.

### Step 1 — Calculate the Classifier Score

$$
s_3=wx_3+b
$$

Substituting:

$$
s_3=(1)(4)+0
$$

Therefore:

$$
\boxed{
s_3=4
}
$$

The score is positive.

But the true class is:

$$
y_3=-1
$$

Therefore, the classifier predicts the wrong class.

### Step 2 — Calculate the Functional Margin

$$
\hat{\gamma}_3
=
y_3(wx_3+b)
$$

Substituting:

$$
\hat{\gamma}_3
=
(-1)(4)
$$

Therefore:

$$
\boxed{
\hat{\gamma}_3=-4
}
$$

The functional margin is negative.

This immediately tells us:

$$
\boxed{
\text{The example is incorrectly classified.}
}
$$

### Step 3 — Calculate the Geometric Margin

$$
\gamma_3
=
\frac{
y_3(wx_3+b)
}{
\|w\|
}
$$

Therefore:

$$
\gamma_3
=
\frac{-4}{1}
$$

So:

$$
\boxed{
\gamma_3=-4
}
$$

The negative sign indicates that the point lies on the wrong side relative to its true class.

The magnitude tells us that it is 4 units from the decision boundary.

## 5. What Does the Sign Tell Us?

From the examples above, we can see:

$$
\gamma_i>0
$$

means the point is correctly classified.

$$
\gamma_i<0
$$

means the point is incorrectly classified.

And:

$$
\gamma_i=0
$$

means the point lies on the decision boundary.

Therefore:

$$
\boxed{
\operatorname{sign}(\gamma_i)
\rightarrow
\text{classification correctness}
}
$$

while:

$$
\boxed{
|\gamma_i|
\rightarrow
\text{distance magnitude}
}
$$

The signed geometric margin therefore contains both pieces of information.
