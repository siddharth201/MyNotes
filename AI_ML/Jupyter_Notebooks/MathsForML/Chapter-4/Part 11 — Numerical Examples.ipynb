# Chapter 4 — Part 11

# Numerical Examples

> **Goal:**  
> Use concrete numerical examples to calculate and interpret the classifier score, functional margin, and geometric margin.

In the previous parts, we developed the following expressions.

### Classifier score

$$
s_i=\vec{w}^{\,T}\vec{x}_i+b
$$

### Functional margin

$$
\hat{\gamma}_i
=
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
$$

### Geometric margin

$$
\gamma_i
=
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
$$

In this part, we will calculate these quantities step by step using numerical examples.

The goal is to move from:

> **"I understand the formula."**

to:

> **"I can calculate it and explain what the result means."**

## 1. Example 1 — A Simple One-Dimensional Classifier

Consider the one-dimensional decision boundary:

$$
x=0
$$

We can represent this boundary using:

$$
wx+b=0
$$

Choose:

$$
w=1
$$

and:

$$
b=0
$$

Therefore:

$$
x=0
$$

is our decision boundary.

Suppose we have a training example:

$$
x_1=4
$$

with class label:

$$
y_1=+1
$$

### Step 1 — Calculate the Classifier Score

The classifier score is:

$$
s_1=wx_1+b
$$

Substituting the values:

$$
s_1=(1)(4)+0
$$

Therefore:

$$
\boxed{
s_1=4
}
$$

The score is positive.

Since the true class is:

$$
y_1=+1
$$

the positive score agrees with the true class.

Therefore, the example is correctly classified.

### Step 2 — Calculate the Functional Margin

The functional margin is:

$$
\hat{\gamma}_1
=
y_1(wx_1+b)
$$

Substituting:

$$
\hat{\gamma}_1
=
(+1)(4)
$$

Therefore:

$$
\boxed{
\hat{\gamma}_1=4
}
$$

The functional margin is positive, so the example is correctly classified.

### Step 3 — Calculate the Geometric Margin

The geometric margin is:

$$
\gamma_1
=
\frac{
y_1(wx_1+b)
}{
|w|
}
$$

Since this is a one-dimensional example:

$$
\|w\|=|w|=1
$$

Therefore:

$$
\gamma_1
=
\frac{4}{1}
$$

So:

$$
\boxed{
\gamma_1=4
}
$$

This means the point is 4 units away from the decision boundary in the class-oriented direction.
