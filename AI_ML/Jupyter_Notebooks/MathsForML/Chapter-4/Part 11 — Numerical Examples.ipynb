# Chapter 4 — Part 11

# Numerical Examples

> **Goal:**  
> Use concrete numerical examples to calculate and interpret the classifier score, functional margin, and geometric margin.

In the previous parts, we developed the following expressions.

### Classifier score

$$
s_i=\vec{w}^{\,T}\vec{x}_i+b
$$

### Functional margin

$$
\hat{\gamma}_i
=
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
$$

### Geometric margin

$$
\gamma_i
=
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
$$

In this part, we will calculate these quantities step by step using numerical examples.

The goal is to move from:

> **"I understand the formula."**

to:

> **"I can calculate it and explain what the result means."**

## 1. Example 1 — A Simple One-Dimensional Classifier

Consider the one-dimensional decision boundary:

$$
x=0
$$

We can represent this boundary using:

$$
wx+b=0
$$

Choose:

$$
w=1
$$

and:

$$
b=0
$$

Therefore:

$$
x=0
$$

is our decision boundary.

Suppose we have a training example:

$$
x_1=4
$$

with class label:

$$
y_1=+1
$$

### Step 1 — Calculate the Classifier Score

The classifier score is:

$$
s_1=wx_1+b
$$

Substituting the values:

$$
s_1=(1)(4)+0
$$

Therefore:

$$
\boxed{
s_1=4
}
$$

The score is positive.

Since the true class is:

$$
y_1=+1
$$

the positive score agrees with the true class.

Therefore, the example is correctly classified.

### Step 2 — Calculate the Functional Margin

The functional margin is:

$$
\hat{\gamma}_1
=
y_1(wx_1+b)
$$

Substituting:

$$
\hat{\gamma}_1
=
(+1)(4)
$$

Therefore:

$$
\boxed{
\hat{\gamma}_1=4
}
$$

The functional margin is positive, so the example is correctly classified.

### Step 3 — Calculate the Geometric Margin

The geometric margin is:

$$
\gamma_1
=
\frac{
y_1(wx_1+b)
}{
\|w\|
}
$$

Since this is a one-dimensional example:

$$
\|w\|=|w|=1
$$

Therefore:

$$
\gamma_1
=
\frac{4}{1}
$$

So:

$$
\boxed{
\gamma_1=4
}
$$

This means the point is 4 units away from the decision boundary in the class-oriented direction.

## 2. Example 2 — A Negative Class Example

Let's now consider a point on the other side of the boundary.

The decision boundary is still:

$$
x=0
$$

and:

$$
w=1,\qquad b=0
$$

Now choose:

$$
x_2=-4
$$

with class label:

$$
y_2=-1
$$

### Step 1 — Calculate the Classifier Score

The score is:

$$
s_2=wx_2+b
$$

Substituting:

$$
s_2=(1)(-4)+0
$$

Therefore:

$$
\boxed{
s_2=-4
}
$$

The score is negative.

Since the true class is:

$$
y_2=-1
$$

the score has the correct sign.

Therefore, the example is correctly classified.

### Step 2 — Calculate the Functional Margin

The functional margin is:

$$
\hat{\gamma}_2
=
y_2(wx_2+b)
$$

Substituting:

$$
\hat{\gamma}_2
=
(-1)(-4)
$$

Therefore:

$$
\boxed{
\hat{\gamma}_2=4
}
$$

Notice something important.

The two examples had different raw scores:

$$
+4
\qquad\text{and}\qquad
-4
$$

But their functional margins are both:

$$
+4
$$

The class label has aligned both examples with the correct class direction.

### Step 3 — Calculate the Geometric Margin

The geometric margin is:

$$
\gamma_2
=
\frac{
y_2(wx_2+b)
}{
\|w\|
}
$$

Since:

$$
\|w\|=1
$$

we get:

$$
\gamma_2
=
\frac{4}{1}
$$

Therefore:

$$
\boxed{
\gamma_2=4
}
$$

So both examples are 4 units from the decision boundary in their respective correct directions.

## 3. Compare the Two Examples

Let's put the two examples together.

| Example | $x_i$ | $y_i$ | Score | Functional margin | Geometric margin |
|---|---:|---:|---:|---:|---:|
| 1 | $+4$ | $+1$ | $+4$ | $+4$ | $+4$ |
| 2 | $-4$ | $-1$ | $-4$ | $+4$ | $+4$ |

The raw classifier scores are different:

$$
+4
\qquad\text{and}\qquad
-4
$$

because the points are on opposite sides of the decision boundary.

But after incorporating the true class label:

$$
y_i(wx_i+b)
$$

both examples produce:

$$
+4
$$

This is exactly why the class label is so useful.

It lets us express the two class-specific cases using one common quantity.

## 4. Example 3 — An Incorrectly Classified Point

Let's now deliberately choose a point whose class does not agree with the classifier.

Keep:

$$
w=1
$$

and:

$$
b=0
$$

Suppose:

$$
x_3=4
$$

but its true class is:

$$
y_3=-1
$$

The point is on the positive side of the decision boundary, but its true class is negative.

### Step 1 — Calculate the Classifier Score

$$
s_3=wx_3+b
$$

Substituting:

$$
s_3=(1)(4)+0
$$

Therefore:

$$
\boxed{
s_3=4
}
$$

The score is positive.

But the true class is:

$$
y_3=-1
$$

Therefore, the classifier predicts the wrong class.

### Step 2 — Calculate the Functional Margin

$$
\hat{\gamma}_3
=
y_3(wx_3+b)
$$

Substituting:

$$
\hat{\gamma}_3
=
(-1)(4)
$$

Therefore:

$$
\boxed{
\hat{\gamma}_3=-4
}
$$

The functional margin is negative.

This immediately tells us:

$$
\boxed{
\text{The example is incorrectly classified.}
}
$$

### Step 3 — Calculate the Geometric Margin

$$
\gamma_3
=
\frac{
y_3(wx_3+b)
}{
\|w\|
}
$$

Therefore:

$$
\gamma_3
=
\frac{-4}{1}
$$

So:

$$
\boxed{
\gamma_3=-4
}
$$

The negative sign indicates that the point lies on the wrong side relative to its true class.

The magnitude tells us that it is 4 units from the decision boundary.

## 5. What Does the Sign Tell Us?

From the examples above, we can see:

$$
\gamma_i>0
$$

means the point is correctly classified.

$$
\gamma_i<0
$$

means the point is incorrectly classified.

And:

$$
\gamma_i=0
$$

means the point lies on the decision boundary.

Therefore:

$$
\boxed{
\operatorname{sign}(\gamma_i)
\rightarrow
\text{classification correctness}
}
$$

while:

$$
\boxed{
|\gamma_i|
\rightarrow
\text{distance magnitude}
}
$$

The signed geometric margin therefore contains both pieces of information.

## 6. Example 4 — A Larger Margin

Consider two correctly classified positive examples.

For the first point:

$$
y_1=+1
$$

and:

$$
\vec{w}^{\,T}\vec{x}_1+b=2
$$

Therefore:

$$
\hat{\gamma}_1=(+1)(2)=2
$$

For the second point:

$$
y_2=+1
$$

and:

$$
\vec{w}^{\,T}\vec{x}_2+b=8
$$

Therefore:

$$
\hat{\gamma}_2=(+1)(8)=8
$$

Both examples are correctly classified.

But:

$$
8>2
$$

So the second example has a larger functional margin.

Intuitively, the classifier has a stronger positive score for the second example.

## 7. But Remember the Scaling Problem

The previous example might make us think:

> "Then we should simply maximize the functional margin."

But Part 10 showed why this is not enough.

Suppose:

$$
\vec{w}^{\,T}\vec{x}_i+b=2
$$

and:

$$
y_i=+1
$$

Then:

$$
\hat{\gamma}_i=2
$$

Now multiply both $\vec{w}$ and $b$ by $10$.

The decision boundary does not change.

But the new score becomes:

$$
20
$$

and the new functional margin becomes:

$$
20
$$

So:

$$
2\rightarrow20
$$

even though the geometric boundary is exactly the same.

Therefore:

$$
\boxed{
\text{Functional margin depends on parameter scale.}
}
$$

This is why we need the geometric margin.

## 8. Example 5 — Functional Margin vs Geometric Margin

Suppose:

$$
y_i=+1
$$

and:

$$
\vec{w}^{\,T}\vec{x}_i+b=6
$$

Also suppose:

$$
\|\vec{w}\|=2
$$

### Functional margin

$$
\hat{\gamma}_i
=
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
$$

Therefore:

$$
\hat{\gamma}_i
=
(+1)(6)
=
6
$$

So:

$$
\boxed{
\hat{\gamma}_i=6
}
$$

### Geometric margin

$$
\gamma_i
=
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
$$

Therefore:

$$
\gamma_i
=
\frac{6}{2}
=
3
$$

So:

$$
\boxed{
\gamma_i=3
}
$$

The functional margin is $6$, while the actual class-oriented perpendicular distance is $3$.

## 9. Example 6 — Scaling the Same Boundary

Let's make the scaling issue completely numerical.

Suppose the original parameters are:

$$
w=2
$$

and:

$$
b=0
$$

The decision boundary is:

$$
2x=0
$$

which gives:

$$
x=0
$$

Now scale both parameters by $3$:

$$
w'=6
$$

and:

$$
b'=0
$$

The new boundary is:

$$
6x=0
$$

which is still:

$$
x=0
$$

Therefore, the geometric decision boundary has not changed.

Now consider:

$$
x=2
$$

The original score is:

$$
(2)(2)=4
$$

The scaled score is:

$$
(6)(2)=12
$$

The functional score has tripled:

$$
4\rightarrow12
$$

But the point has not moved and the decision boundary has not moved.

Therefore, the raw score and functional margin cannot by themselves represent geometric distance.

## 10. Normalize the Functional Margin

The geometric margin fixes the scaling problem by dividing by:

$$
\|\vec{w}\|
$$

Recall:

$$
\gamma_i
=
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
$$

Consider the previous example.

### Original parameters

$$
w=2
$$

The score for $x=2$ was:

$$
4
$$

Therefore:

$$
\gamma
=
\frac{4}{2}
=
2
$$

### Scaled parameters

$$
w'=6
$$

The new score was:

$$
12
$$

Therefore:

$$
\gamma'
=
\frac{12}{6}
=
2
$$

So:

$$
\boxed{
\gamma=\gamma'=2
}
$$

The geometric margin remains unchanged.

This is exactly what we want from a geometric quantity.

## 11. One Example — All Three Quantities

Let's summarize one example.

Suppose:

$$
\vec{w}
=
\begin{bmatrix}
2
\end{bmatrix}
$$

$$
b=0
$$

$$
\vec{x}_i
=
\begin{bmatrix}
3
\end{bmatrix}
$$

and:

$$
y_i=+1
$$

### Classifier score

$$
\vec{w}^{\,T}\vec{x}_i+b
=
(2)(3)+0
=
6
$$

Therefore:

$$
\boxed{
\text{Score}=6
}
$$

### Functional margin

$$
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
=
(+1)(6)
=
6
$$

Therefore:

$$
\boxed{
\text{Functional margin}=6
}
$$

### Norm of $\vec{w}$

$$
\|\vec{w}\|=2
$$

### Geometric margin

$$
\gamma_i
=
\frac{6}{2}
=
3
$$

Therefore:

$$
\boxed{
\text{Geometric margin}=3
}
$$

So for this example:

$$
\boxed{
\text{Score}=6
}
$$

$$
\boxed{
\text{Functional margin}=6
}
$$

$$
\boxed{
\text{Geometric margin}=3
}
$$
