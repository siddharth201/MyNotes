# Chapter 4 — Part 13

# Machine Learning Perspective

> **Core Question:**  
> We have developed decision boundaries, classifier scores, class-adjusted scores, and margins mathematically. How do all of these ideas fit into the actual Machine Learning process?

So far, we have mostly looked at the problem from a mathematical and geometric perspective.

Now let's step back and look at the same problem as a Machine Learning practitioner.

A supervised classification problem can be viewed as:

$$
\boxed{
\text{Data}
\rightarrow
\text{Model}
\rightarrow
\text{Prediction}
\rightarrow
\text{Measure}
\rightarrow
\text{Optimization}
}
$$

Our goal is to understand where the mathematics from the previous parts fits into this pipeline.

## 1. Start With Training Data

In supervised learning, we begin with labeled training examples.

Each training example consists of:

$$
(\vec{x}_i,y_i)
$$

where:

- $\vec{x}_i$ is the input feature vector.
- $y_i$ is the true class label.

For binary classification, we use:

$$
y_i\in\{-1,+1\}
$$

Suppose we have $m$ training examples.

Then our training dataset can be represented as:

$$
\mathcal{D}
=
\left\{
(\vec{x}_1,y_1),
(\vec{x}_2,y_2),
\ldots,
(\vec{x}_m,y_m)
\right\}
$$

The Machine Learning algorithm receives these examples and tries to learn a classifier that can correctly classify them.

## 2. The Model Has Parameters

We use a linear classifier.

Its decision boundary is:

$$
\boxed{
\vec{w}^{\,T}\vec{x}+b=0
}
$$

The model therefore has two sets of parameters:

$$
\boxed{
\vec{w}
}
$$

and:

$$
\boxed{
b
}
$$

Here:

- $\vec{w}$ determines the orientation of the decision boundary.
- $b$ determines its position.

These parameters are not chosen randomly forever.

They are **learned from the training data**.

This is one of the most important ideas in Machine Learning:

> **Learning means finding model parameters that produce useful predictions on the training data and, ideally, on unseen data.**

## 3. The Model Produces a Score

Once the model has parameters $\vec{w}$ and $b$, we can give it a training example $\vec{x}_i$.

The model calculates:

$$
\boxed{
s_i
=
\vec{w}^{\,T}\vec{x}_i+b
}
$$

This is the classifier's raw score.

The sign of this score tells us which side of the decision boundary the point lies on.

$$
s_i>0
\quad\Rightarrow\quad
\text{positive side}
$$

$$
s_i<0
\quad\Rightarrow\quad
\text{negative side}
$$

and:

$$
s_i=0
\quad\Rightarrow\quad
\text{on the decision boundary}
$$

So the model first produces a numerical score.

The classification decision is then based on that score.

## 4. From Score to Prediction

For binary classification, we can convert the score into a predicted class.

A simple prediction rule is:

$$
\hat{y}_i=
\begin{cases}
+1, & \text{if } \vec{w}^{\,T}\vec{x}_i+b>0\\
-1, & \text{if } \vec{w}^{\,T}\vec{x}_i+b<0
\end{cases}
$$

Here:

$$
\hat{y}_i
$$

represents the model's predicted class.

Notice the difference between:

$$
y_i
$$

and:

$$
\hat{y}_i
$$

### $y_i$

The true class supplied by the training data.

### $\hat{y}_i$

The class predicted by the model.

Machine Learning is fundamentally concerned with making these agree:

$$
\boxed{
\hat{y}_i=y_i
}
$$

for as many relevant examples as possible.

## 5. Prediction vs True Label

Let's consider a simple example.

Suppose:

$$
y_i=+1
$$

The true class is positive.

If the model produces:

$$
\vec{w}^{\,T}\vec{x}_i+b=5
$$

then:

$$
\hat{y}_i=+1
$$

Therefore:

$$
\hat{y}_i=y_i
$$

The prediction is correct.

Now suppose the model produces:

$$
\vec{w}^{\,T}\vec{x}_i+b=-5
$$

Then:

$$
\hat{y}_i=-1
$$

while:

$$
y_i=+1
$$

Therefore:

$$
\hat{y}_i\neq y_i
$$

The prediction is incorrect.

So we can think of classification as a comparison between:

$$
\boxed{
\text{Predicted class}
}
$$

and:

$$
\boxed{
\text{True class}
}
$$

## 6. Why the Class-Adjusted Score Is Useful

Instead of separately comparing $\hat{y}_i$ and $y_i$, we introduced the class-adjusted score:

$$
\boxed{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}
$$

This expression automatically accounts for the true class.

For a positive example:

$$
y_i=+1
$$

the expression keeps the original score.

For a negative example:

$$
y_i=-1
$$

the expression reverses the sign of the score.

Therefore, both classes can be evaluated using the same rule:

$$
\boxed{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)>0
}
$$

means the classifier is correct.

This is extremely useful because Machine Learning algorithms need mathematical quantities that can be evaluated systematically across the entire dataset.

## 7. From Individual Examples to the Dataset

A Machine Learning model is not trained using one example at a time conceptually.

We care about its behavior across the training dataset:

$$
\mathcal{D}
=
\left\{
(\vec{x}_1,y_1),
\ldots,
(\vec{x}_m,y_m)
\right\}
$$

For every training example, we can calculate:

$$
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
$$

So we obtain a collection of class-adjusted scores:

$$
y_1
\left(
\vec{w}^{\,T}\vec{x}_1+b
\right),
\ldots,
y_m
\left(
\vec{w}^{\,T}\vec{x}_m+b
\right)
$$

The Machine Learning problem now becomes:

> **How should we evaluate all of these values together?**

And even more importantly:

> **How should we choose $\vec{w}$ and $b$ so that the resulting classifier is good?**

## 8. This Is Where an Objective Comes In

Machine Learning algorithms need a way to measure the quality of a model.

We therefore define an objective.

Conceptually:

$$
\boxed{
\text{Model parameters}
\rightarrow
\text{Objective value}
}
$$

For our classifier, the parameters are:

$$
(\vec{w},b)
$$

and the objective depends on how these parameters behave on the training examples.

We can therefore think of:

$$
J(\vec{w},b)
$$

as a function that evaluates the quality of a particular choice of model parameters.

The learning algorithm then tries to find parameters that optimize this objective.

This is the fundamental Machine Learning pattern:

$$
\boxed{
\text{Define what "good" means}
\rightarrow
\text{Optimize the model to achieve it}
}
$$

## 9. What Does "Good" Mean for Our Classifier?

At the simplest level, we want correct classifications.

That means:

$$
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)>0
$$

for the training examples.

But Part 10 taught us something deeper.

We do not merely want a classifier that happens to classify the examples correctly.

We want the examples to be separated from the decision boundary by a meaningful amount.

This leads us toward the idea of **margin**.

Recall the geometric margin:

$$
\boxed{
\gamma_i
=
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
}
$$

So from a Machine Learning perspective, we can say:

> **A good linear classifier should not merely classify the training examples correctly; it should provide a useful separation between the classes.**

## 10. The Learning Problem

We can now describe our Machine Learning problem more formally.

We have:

### Data

$$
(\vec{x}_i,y_i)
$$

### Model

$$
\vec{w}^{\,T}\vec{x}+b
$$

### Parameters

$$
\vec{w},b
$$

### Prediction

$$
\hat{y}_i
$$

### Evaluation

Compare the model's behavior with:

$$
y_i
$$

### Objective

Choose parameters that produce a desirable classifier.

So the overall learning problem can be visualized as:

$$
\boxed{
\begin{array}{c}
\text{Training Data}\\
(\vec{x}_i,y_i)
\end{array}
}
\rightarrow
\boxed{
\begin{array}{c}
\text{Learn}\\
\vec{w},b
\end{array}
}
\rightarrow
\boxed{
\begin{array}{c}
\text{Classifier}\\
\vec{w}^{\,T}\vec{x}+b
\end{array}
}
$$

The algorithm's job is to find parameters that satisfy the desired learning objective.

## 11. Training Is Parameter Search

This gives us another useful way to think about Machine Learning.

The parameters:

$$
\vec{w},b
$$

can take many possible values.

Each choice produces a different decision boundary.

Therefore, training can be viewed conceptually as a search:

$$
(\vec{w},b)
\rightarrow
\text{decision boundary}
\rightarrow
\text{model behavior}
$$

The learning algorithm searches for parameters that give desirable behavior according to the chosen objective.

So:

> **Training a model means finding suitable parameter values.**

The mathematics of optimization gives us tools for performing this search efficiently.

## 12. Why We Cannot Just Memorize the Training Data

There is another important Machine Learning perspective.

Suppose a model performs perfectly on the training examples.

That is useful, but it does not automatically mean that the model is good.

Why?

Because the real purpose of Machine Learning is to make useful predictions on **unseen examples**.

We therefore distinguish between:

### Training data

Examples used to learn the model.

### Unseen data

Examples the model did not use during training.

The real goal is:

$$
\boxed{
\text{Good performance on unseen data}
}
$$

This ability is called **generalization**.

Therefore, the objective we choose during training should encourage a model that captures useful structure in the data rather than merely memorizing the training examples.

## 13. Why Margin Becomes Interesting

Now our geometric discussion becomes directly relevant to Machine Learning.

Imagine two classifiers that both correctly classify all training examples.

### Classifier A

Some training examples lie very close to the decision boundary.

### Classifier B

The training examples are separated from the decision boundary by a larger margin.

Both may have zero classification errors on the training data.

But Classifier B has a more robust separation between the classes.

This motivates the maximum-margin idea:

$$
\boxed{
\text{Prefer a classifier with a larger geometric margin}
}
$$

This is the central intuition behind Support Vector Machines.

The margin is not merely a geometric curiosity.

It becomes part of the learning principle.

## 14. From Geometry to an ML Objective

We can now connect the mathematical quantities to the Machine Learning objective.

For each training example:

$$
\gamma_i
=
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
$$

represents its class-oriented geometric margin.

A classifier with a larger minimum margin provides greater separation from the decision boundary.

Therefore, conceptually, we want:

$$
\boxed{
\text{maximize the minimum geometric margin}
}
$$

In other words:

$$
\boxed{
\max_{\vec{w},b}
\;
\min_i
\;
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
}
$$

This expression captures the maximum-margin perspective.

We will later see how this formulation is transformed into the standard optimization problem used by SVMs.

## 15. Why "Minimum" Margin?

At first, the phrase:

> **maximize the minimum margin**

may sound unusual.

Why not maximize the average margin?

The reason is that classification safety is determined by the examples that are **closest to the decision boundary**.

Suppose three training examples have geometric margins:

$$
2,\qquad 5,\qquad 10
$$

The minimum margin is:

$$
\min(2,5,10)=2
$$

The closest example has only 2 units of separation.

Now imagine another classifier with margins:

$$
4,\qquad 5,\qquad 8
$$

Its minimum margin is:

$$
\min(4,5,8)=4
$$

The second classifier provides a larger guaranteed separation.

Therefore:

$$
\boxed{
\text{Maximum-margin learning focuses on the closest training examples.}
}
$$

These closest examples will become extremely important when we study Support Vector Machines.

## 17. What the Algorithm Is Really Learning

It is tempting to say:

> "The algorithm learns the decision boundary."

That is true, but there is a more precise way to think about it.

The algorithm actually learns the parameters:

$$
\boxed{
\vec{w},b
}
$$

These parameters determine the decision boundary:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

So the learning process is:

$$
\boxed{
\text{Training data}
\rightarrow
\text{learn }(\vec{w},b)
\rightarrow
\text{decision boundary}
}
$$

The choice of $\vec{w}$ and $b$ is determined by the optimization objective.

This is why defining the objective correctly is one of the most important parts of Machine Learning.

The algorithm then searches for:

## 19. Important Distinction — Model vs Objective

Two concepts are easy to confuse.

### Model

The model tells us how predictions are produced.

For our classifier:

$$
f(\vec{x})
=
\vec{w}^{\,T}\vec{x}+b
$$

### Objective

The objective tells us what kind of model we want.

It evaluates the quality of the chosen parameters according to the learning goal.

Therefore:

$$
\boxed{
\text{Model}
\neq
\text{Objective}
}
$$

The model says:

> **How do I make a prediction?**

The objective says:

> **What makes those predictions good?**

Training connects the two by adjusting the model parameters to optimize the objective.

## 20. The Bigger ML Pattern

The ideas from this chapter are an example of a much broader Machine Learning pattern.

$$
\boxed{
\text{Choose a model}
}
$$

$$↓$$

$$
\boxed{
\text{Define an objective}
}
$$

$$↓$$

$$
\boxed{
\text{Optimize the parameters}
}
$$

$$↓$$

$$
\boxed{
\text{Evaluate generalization}
}
$$

This pattern appears throughout Machine Learning.

Different algorithms may use different models and different objectives, but the underlying idea remains:

> **Define what you want the model to achieve, then learn parameters that move the model toward that goal.**

Our linear classification problem is one concrete example of this general Machine Learning principle.
