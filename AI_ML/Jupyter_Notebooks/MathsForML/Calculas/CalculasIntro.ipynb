

# Calculus — What Is It?

If we start Calculus directly with *functions, limits, and derivatives*, it is easy to learn the mechanics without understanding **why Calculus exists in the first place**.

So let's first answer three questions:

1. **What is Calculus?**
2. **Why was Calculus developed?**
3. **Why does Machine Learning need it?**

---

# 0. The Big Idea

At the highest level:

> **Calculus is the mathematics of change and accumulation.**

Two ideas are at its heart:

### 1. Change

How quickly is something changing?

This leads to:

$$
\boxed{\text{Derivative}}
$$

### 2. Accumulation

How much has something accumulated over a range?

This leads to:

$$
\boxed{\text{Integral}}
$$

So, very roughly:

```text
                         CALCULUS
                            │
              ┌─────────────┴─────────────┐
              │                           │
           CHANGE                    ACCUMULATION
              │                           │
              ▼                           ▼
         Derivatives                  Integrals
              │                           │
              │                           │
       "How fast?"                 "How much total?"
```

But this simple definition becomes much more meaningful when we understand **the problem that led humans to develop Calculus**.

---

# 1. What Problem Was Calculus Trying to Solve?

Imagine you are watching a car.

At 10:00:00 AM, the car is here.

At 10:00:10 AM, the car is somewhere else.

You can calculate its **average speed**:

$$
\text{Average speed}
=
\frac{\text{Distance travelled}}{\text{Time taken}}
$$

For example:

The car travels:

$$
100\text{ m}
$$

in:

$$
10\text{ s}
$$

Therefore:

$$
\text{Average speed}
=
\frac{100}{10}
=
10\text{ m/s}
$$

Easy.

But now ask a more interesting question:

> **What was the speed of the car at exactly 10:00:05 AM?**

That's much harder.

Because at one exact instant, there is seemingly:

* no distance interval
* no time interval

We want:

$$
\text{Speed at an instant}
=
\frac{\text{change in distance}}{\text{change in time}}
$$

but the time interval is approaching zero.

This is one of the fundamental problems that led toward Calculus.

---

# 2. Another Problem: What Is the Slope at One Exact Point?

Consider:

$$
y=x^2
$$

We can easily calculate the slope between two points.

Suppose:

$$
A=(1,1)
$$

and:

$$
B=(3,9)
$$

The slope is:

$$
m=
\frac{9-1}{3-1}
=
4
$$

That's the slope of the **line connecting two points**.

This is called a **secant line**.

But what if I ask:

> What is the slope of the curve at exactly $x=1$?

There isn't another point.

We need to somehow bring the second point closer and closer to the first.

```text
Two points

A ●────────────● B
     secant


Bring B closer

A ●──────● B


Bring B even closer

A ●──● B


Eventually...

A ●
  ↑
  tangent direction
```

The mathematical machinery for handling this "getting closer and closer" idea is the **limit**.

And from the limit comes the **derivative**.

---

# 3. The Deeper Problem: Continuous Change

This is where the historical importance of Calculus becomes clearer.

Humans had mathematics for many things:

* counting
* arithmetic
* geometry
* equations

But many natural phenomena aren't static.

They **change continuously**.

For example:

* position changes with time
* velocity changes with time
* temperature changes with time
* planetary positions change
* populations change
* water levels change
* light intensity changes

The difficult question became:

> **How can mathematics describe something that is continuously changing?**

That is one of the central questions Calculus gives us tools to answer.

---

# 4. Why Was Calculus Developed?

There wasn't one single problem or one single inventor.

Calculus emerged from centuries of work by many mathematicians, with major independent developments by:

* **Isaac Newton**
* **Gottfried Wilhelm Leibniz**

in the 17th century.

Their motivations were somewhat different.

### Newton

Newton was deeply interested in:

* motion
* velocity
* acceleration
* planetary motion
* physics

He needed mathematics capable of describing **continuously changing physical quantities**.

### Leibniz

Leibniz developed a powerful symbolic framework for:

* infinitesimal changes
* derivatives
* integration
* mathematical relationships

His notation is closely related to the notation we still use today.

For example:

$$
\frac{dy}{dx}
$$

is Leibniz's notation.

---

# 5. But Why Couldn't Existing Mathematics Handle This?

Because ordinary algebra is excellent at describing relationships, but it doesn't naturally answer questions like:

> "How quickly is this quantity changing **right now**?"

Suppose:

$$
s(t)
$$

represents the position of an object at time $t$.

We can calculate average velocity:

$$
\frac{s(t_2)-s(t_1)}{t_2-t_1}
$$

But physics often needs:

> velocity at exactly time $t$

Calculus gives us:

$$
v(t)
=
\frac{ds}{dt}
$$

And then something even more interesting:

> How quickly is velocity itself changing?

That gives:

$$
a(t)
=
\frac{dv}{dt}
$$

or:

$$
a(t)
=
\frac{d^2s}{dt^2}
$$

So we get:

```text
Position
   │
   │ derivative
   ▼
Velocity
   │
   │ derivative
   ▼
Acceleration
```

This is a beautiful example of what Calculus allows us to do.

---

# 6. Calculus Is Really About "Zooming In"

Here's an extremely useful intuition that I want you to keep.

Imagine a curved road.

From far away:

```text
        ╭──────╮
     ╭──╯      ╰──╮
─────╯             ╰────
```

It is clearly curved.

But imagine zooming in at one tiny point.

```text
                 /
                /
               /
              ●
             /
            /
```

At a sufficiently small scale, the curve behaves approximately like a **straight line**.

Calculus gives us a mathematical way to reason about this local behavior.

That's why the derivative can tell us:

> **What is happening to the function right here?**

This idea of **local behavior** will become extremely important in Machine Learning.

---

# 7. Calculus Has Two Major Questions

We can now make our earlier definition much more meaningful.

## Question 1 — How is something changing?

For example:

> If I increase $x$ slightly, how much does $y$ change?

This leads to:

$$
\boxed{\text{Derivative}}
$$

---

## Question 2 — How much has something accumulated?

Suppose we know the velocity of a car at every moment.

Can we determine the total distance travelled?

Now we're asking an accumulation question.

This leads to:

$$
\boxed{\text{Integral}}
$$

So:

```text
             CALCULUS
                │
        ┌───────┴────────┐
        │                │
        ▼                ▼
      CHANGE         ACCUMULATION
        │                │
        ▼                ▼
   Derivative         Integral
        │                │
   "How fast?"      "How much?"
```

And these two ideas are deeply connected through the **Fundamental Theorem of Calculus**.

We will eventually study that connection.

---

# 8. Now the Important Part: Why Does ML Need Calculus?

This is where our Calculus journey becomes directly relevant.

Imagine we build a very simple ML model:

$$
\hat y = wx
$$

Here:

* $x$ = input
* $w$ = model parameter
* $\hat y$ = prediction

Suppose the true answer is $y$.

Our model makes some error.

We define a loss:

$$
L(w)
$$

The goal of training is:

$$
\boxed{\text{Find }w\text{ that makes }L(w)\text{ as small as possible}}
$$

This is an **optimization problem**.

---

# 9. The Question ML Asks

Suppose our current parameter is:

$$
w=2
$$

and we are considering changing it slightly.

We want to know:

> If I increase $w$, does the loss increase or decrease?

And more importantly:

> **How sensitive is the loss to a change in $w$?**

That is exactly what a derivative tells us.

$$
\frac{dL}{dw}
$$

The derivative answers, conceptually:

> **If I change $w$ slightly, how does $L$ respond?**

And suddenly Calculus becomes directly useful to ML.

---

# 10. The Simplest ML Example

Suppose our loss is:

$$
L(w)=(w-3)^2
$$

Our goal is:

$$
\min_w L(w)
$$

Let's look at a few values:

| $w$ | $L(w)$ |
| --: | -----: |
|   0 |      9 |
|   1 |      4 |
|   2 |      1 |
|   3 |      0 |
|   4 |      1 |
|   5 |      4 |

We can see that the minimum occurs at:

$$
w=3
$$

But real ML problems aren't usually this easy.

We don't want to manually inspect every possible value of $w$.

We need a systematic way to determine:

> **Which direction should I move $w$ to reduce the loss?**

Calculus gives us that information.

---

# 11. Derivative Gives Us the Direction of Change

For:

$$
L(w)=(w-3)^2
$$

the derivative is:

$$
\frac{dL}{dw}
=
2(w-3)
$$

At:

$$
w=5
$$

we get:

$$
\frac{dL}{dw}
=
2(5-3)
=
4
$$

The derivative is positive.

That tells us the loss is increasing as $w$ increases locally.

Therefore, to reduce the loss, we should move **left**.

Now consider:

$$
w=1
$$

Then:

$$
\frac{dL}{dw}
=
2(1-3)
=
-4
$$

The derivative is negative.

Now the loss decreases if we move toward larger $w$.

So:

```text
                Loss
                 ▲
                 │
             ●   │   ●
           /       \
         /           \
       ●               ●
      /                 \
─────●───────────────────●──────► w
     0   1   2   3   4   5
                 ↑
              minimum
```

The derivative tells us **which way the slope is pointing**.

---

# 12. This Leads Directly to Gradient Descent

For one parameter:

$$
w_{new}
=
w_{old}
-
\eta
\frac{dL}{dw}
$$

where:

* $w$ = model parameter
* $L$ = loss
* $\frac{dL}{dw}$ = derivative of loss with respect to parameter
* $\eta$ = learning rate

The fundamental idea is:

> **Look at how the loss changes, then move the parameter in the direction that decreases the loss.**

That is the heart of gradient-based optimization.

---

# 13. Real ML Has Thousands or Millions of Parameters

Now our simple example had one parameter:

$$
w
$$

But a real model might have:

$$
\vec{w}
=
[w_1,w_2,w_3,\ldots,w_n]
$$

and loss:

$$
L(w_1,w_2,\ldots,w_n)
$$

Now we need to ask:

> How does the loss change when $w_1$ changes?

$$
\frac{\partial L}{\partial w_1}
$$

What about $w_2$?

$$
\frac{\partial L}{\partial w_2}
$$

And so on.

We collect all of them:

$$
\nabla L
=
\begin{bmatrix}
\frac{\partial L}{\partial w_1}\\
\frac{\partial L}{\partial w_2}\\
\vdots\\
\frac{\partial L}{\partial w_n}
\end{bmatrix}
$$

This is the **gradient**.

And now:

$$
\boxed{
\text{Calculus}
\rightarrow
\text{Gradient}
\rightarrow
\text{Gradient Descent}
}
$$

---

# 14. The Most Important ML Connection

We can now see the entire chain:

```text
                    MACHINE LEARNING
                           │
                           ▼
                     Model makes
                     predictions
                           │
                           ▼
                       Calculate
                         Loss
                           │
                           ▼
                  We want to minimize
                         Loss
                           │
                           ▼
                    Optimization
                           │
                           ▼
              How does Loss change
                 when parameters
                     change?
                           │
                           ▼
                       Calculus
                           │
                           ▼
                     Derivatives
                           │
                           ▼
                  Partial Derivatives
                           │
                           ▼
                       Gradient
                           │
                           ▼
                  Gradient Descent
                           │
                           ▼
                 Update Parameters
                           │
                           ▼
                 Better Predictions
```

This is why Calculus is not just another mathematical prerequisite for ML.

**It is part of the mechanism through which many ML models learn.**

---

# 15. And Neural Networks Go Even Deeper

Consider a neural network.

Conceptually:

$$
\vec{x}
\rightarrow
\text{Layer 1}
\rightarrow
\text{Layer 2}
\rightarrow
\text{Layer 3}
\rightarrow
\hat y
$$

We calculate a loss:

$$
L
$$

Now we want to know:

> How much did each parameter contribute to the final loss?

That requires derivatives.

But because the network is a **composition of many functions**, we need the:

$$
\boxed{\text{Chain Rule}}
$$

The chain rule allows derivatives to propagate backward through the network.

That's the mathematical foundation behind:

$$
\boxed{\text{Backpropagation}}
$$

So eventually our journey becomes:

$$
\boxed{
\text{Calculus}
\rightarrow
\text{Derivatives}
\rightarrow
\text{Chain Rule}
\rightarrow
\text{Gradients}
\rightarrow
\text{Backpropagation}
}
$$

---

# 16. Where Does Integration Fit Into ML?

Derivatives will be much more prominent in our ML journey, but integrals also matter.

For example, probability distributions often involve continuous quantities.

If:

$$
p(x)
$$

is a probability density function, then probability over an interval is expressed using an integral:

$$
P(a\leq X\leq b)
=
\int_a^b p(x)\,dx
$$

Integrals also appear in:

* probability
* expectation
* continuous distributions
* statistical learning theory
* Bayesian methods
* probabilistic ML
* physics-informed ML

So our emphasis will be different:

```text
Calculus
│
├── Derivatives ──────────────── VERY IMPORTANT for ML
│      │
│      ├── Gradients
│      ├── Optimization
│      ├── Gradient Descent
│      └── Backpropagation
│
└── Integrals ───────────────── Important foundation
       │
       ├── Probability
       ├── Expectation
       └── Continuous distributions
```

# 18. Calculus Was Not "Invented for Machine Learning"

This distinction is important.

Calculus existed **hundreds of years before modern Machine Learning**.

It was developed primarily to solve mathematical and physical problems involving:

* motion
* changing quantities
* geometry
* areas
* rates
* planetary motion
* physical laws

Machine Learning later found Calculus extremely useful because **model training is fundamentally an optimization problem involving continuously varying parameters**.

So the historical progression is more like:

```text
Physical / Mathematical Problems
             ↓
          Calculus
             ↓
        Optimization
             ↓
      Numerical Methods
             ↓
      Machine Learning
             ↓
      Neural Networks
             ↓
     Modern Deep Learning
```
