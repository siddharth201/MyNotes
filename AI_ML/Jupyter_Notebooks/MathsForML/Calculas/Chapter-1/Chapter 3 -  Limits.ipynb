# Chapter 2 — Limits

Before we learn **derivatives**, we need to solve one fundamental problem:

> **How can we mathematically describe what happens to a function when its input gets arbitrarily close to a particular value?**

That is the idea of a **limit**.

Limits are not just another Calculus topic to memorize. They are the mathematical foundation on which the derivative is built.

The conceptual chain is:

$$
\boxed{
\text{Function}
\rightarrow
\text{Change}
\rightarrow
\text{Limit}
\rightarrow
\text{Derivative}
}
$$

---

# 2.1 Why Do We Need Limits?

Let's begin with a simple function:

$$
f(x)=x^2
$$

Suppose we want to know:

> What happens to $f(x)$ as $x$ gets closer and closer to $2$?

We can approach $2$ from below:

$$
1.9,\ 1.99,\ 1.999,\ 1.9999,\ldots
$$

or from above:

$$
2.1,\ 2.01,\ 2.001,\ 2.0001,\ldots
$$

Let's calculate the corresponding outputs.

|     $x$ | $f(x)=x^2$ |
| ------: | ---------: |
|   $1.9$ |     $3.61$ |
|  $1.99$ |   $3.9601$ |
| $1.999$ | $3.996001$ |
| $2.001$ | $4.004001$ |
|  $2.01$ |   $4.0401$ |
|   $2.1$ |     $4.41$ |

Notice what is happening.

As:

$$
x\rightarrow2
$$

the output gets closer and closer to:

$$
4
$$

So we write:

$$
\boxed{
\lim_{x\rightarrow2}x^2=4
}
$$

Read this as:

> **"The limit of $x^2$ as $x$ approaches $2$ is $4$."**

---

# 2.2 What Does "Approaches" Mean?

This is the most important word in the definition.

When we say:

$$
x\rightarrow2
$$

we are **not initially saying that $x=2$**.

We are saying:

> $x$ is getting closer and closer to $2$.

For example:

$$
1.9\rightarrow2
$$

$$
1.99\rightarrow2
$$

$$
1.999\rightarrow2
$$

and from the other side:

$$
2.1\rightarrow2
$$

$$
2.01\rightarrow2
$$

$$
2.001\rightarrow2
$$

The values don't have to equal $2$.

They are getting **arbitrarily close** to $2$.

---

# 2.3 Approaching Is Not the Same as Reaching

This distinction is extremely important.

Suppose:

$$
x\rightarrow2
$$

We are interested in values such as:

$$
1.9,\ 1.99,\ 1.999,\ldots
$$

and:

$$
2.1,\ 2.01,\ 2.001,\ldots
$$

We are asking:

> What value is the function approaching?

We are **not necessarily asking**:

> What is the actual value of the function at $x=2$?

This distinction becomes crucial when the function is not defined at that point.

---

# 2.4 A Function Can Have a Limit Even When It Is Not Defined at the Point

Consider:

$$
f(x)=\frac{x^2-4}{x-2}
$$

Let's substitute:

$$
x=2
$$

We get:

$$
f(2)
=
\frac{2^2-4}{2-2}
=
\frac{0}{0}
$$

But:

$$
\frac00
$$

is undefined.

So:

> **$f(2)$ does not exist.**

Does that mean we cannot say anything about what happens near $x=2$?

No.

Let's simplify the expression.

Since:

$$
x^2-4=(x-2)(x+2)
$$

we get:

$$
f(x)
=
\frac{(x-2)(x+2)}{x-2}
$$

For:

$$
x\neq2
$$

we can cancel $x-2$:

$$
f(x)=x+2
$$

Therefore, as $x$ gets closer to $2$:

$$
x+2\rightarrow4
$$

So:

$$
\boxed{
\lim_{x\rightarrow2}
\frac{x^2-4}{x-2}
=4
}
$$

even though:

$$
f(2)
$$

is undefined.

This is one of the most important reasons limits are needed.

---

# 2.5 The Hole in the Graph

The previous function behaves like:

$$
y=x+2
$$

everywhere except at:

$$
x=2
$$

There is a missing point at:

$$
(2,4)
$$

We can visualize this directly in Jupyter.



## Jupyter Visualization

````python
import numpy as np
import matplotlib.pyplot as plt

# Values on both sides of x = 2
x_left = np.linspace(-2, 1.95, 300)
x_right = np.linspace(2.05, 6, 300)

y_left = (x_left**2 - 4) / (x_left - 2)
y_right = (x_right**2 - 4) / (x_right - 2)

plt.figure(figsize=(8, 5))

plt.plot(x_left, y_left)
plt.plot(x_right, y_right)

# Mark the missing point
plt.scatter([2], [4], facecolors="white", edgecolors="black", s=80, zorder=5)

plt.axhline(0)
plt.axvline(0)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(r"$f(x)=\frac{x^2-4}{x-2}$")

plt.grid(True)
plt.show()
````



The important observation is:

> The graph approaches the missing point $(2,4)$ from both sides.

Therefore:

$$
\boxed{
\lim_{x\rightarrow2}f(x)=4
}
$$

even though $f(2)$ itself is undefined.

---

# 2.6 The Core Meaning of a Limit

We can now give our first proper intuition.

> **The limit describes the value that a function's output approaches as the input gets arbitrarily close to a particular value.**

In notation:

$$
\boxed{
\lim_{x\rightarrow a}f(x)=L
}
$$

means:

> As $x$ gets arbitrarily close to $a$, $f(x)$ gets arbitrarily close to $L$.

Notice that the limit is fundamentally about **behavior near a point**, not necessarily the value exactly at that point.

---

# 2.7 Limit vs Function Value

This distinction deserves its own section.

Suppose:

$$
\lim_{x\rightarrow a}f(x)=L
$$

There are three possibilities:

### Case 1 — Function is defined and equals the limit

$$
f(a)=L
$$

Everything behaves nicely.

---

### Case 2 — Function is not defined at $a$

$$
f(a)\text{ is undefined}
$$

but:

$$
\lim_{x\rightarrow a}f(x)=L
$$

The graph has a hole.

---

### Case 3 — Function is defined, but has a different value

It is possible that:

$$
f(a)\neq L
$$

while:

$$
\lim_{x\rightarrow a}f(x)=L
$$

The function's assigned value at $a$ can differ from the value the surrounding function approaches.

This distinction will become important when we study **continuity**.

---

# 2.8 Limits From the Left and Right

So far, we've approached $2$ from both sides.

But we can study each side separately.

## Approaching from the left

We write:

$$
\boxed{
\lim_{x\rightarrow a^-}f(x)
}
$$

The superscript $-$ means:

> Approach $a$ from values smaller than $a$.

For example:

$$
1.9,\ 1.99,\ 1.999
$$

when approaching $2$.

---

## Approaching from the right

We write:

$$
\boxed{
\lim_{x\rightarrow a^+}f(x)
}
$$

The superscript $+$ means:

> Approach $a$ from values greater than $a$.

For example:

$$
2.1,\ 2.01,\ 2.001
$$

when approaching $2$.

---

# 2.9 Why Do We Need One-Sided Limits?

Because sometimes the function behaves differently on the two sides.

Consider a function:

$$
f(x)=
\begin{cases}
0, & x<0\\
1, & x>0
\end{cases}
$$

As we approach $0$ from the left:

$$
\lim_{x\rightarrow0^-}f(x)=0
$$

But from the right:

$$
\lim_{x\rightarrow0^+}f(x)=1
$$

Therefore:

$$
\boxed{
\lim_{x\rightarrow0}f(x)
\text{ does not exist}
}
$$

because the two sides approach different values.

---

# 2.10 The Rule for a Two-Sided Limit

A two-sided limit exists only when:

$$
\boxed{
\lim_{x\rightarrow a^-}f(x)
=
\lim_{x\rightarrow a^+}f(x)
}
$$

If both equal $L$, then:

$$
\boxed{
\lim_{x\rightarrow a}f(x)=L
}
$$

If they are different:

$$
\lim_{x\rightarrow a^-}f(x)
\neq
\lim_{x\rightarrow a^+}f(x)
$$

then:

$$
\boxed{
\lim_{x\rightarrow a}f(x)
\text{ does not exist}
}
$$

This is a fundamental test.

---

# 2.11 Graphical Interpretation of a Limit

Suppose we have:

$$
\lim_{x\rightarrow a}f(x)=L
$$

Graphically, we can think:

```text
             y
             ↑
             │
          ●  │  ●
         /   │   \
        /    │    \
───────/─────●─────\──────► x
             │
             a
             │
             L
```

The important question is not necessarily:

> Is there a filled point at $(a,L)$?

Instead:

> **As we move toward $x=a$ from either side, where does the graph go?**

If it approaches the same $y$-value $L$, the limit is $L$.

---

# 2.12 Numerical Interpretation

There are three useful ways to understand a limit.

### 1. Algebraically

Manipulate the expression.

### 2. Numerically

Evaluate the function at values increasingly close to $a$.

### 3. Graphically

Observe where the graph approaches.

These are different perspectives on the same underlying idea.

For example:

$$
\lim_{x\rightarrow2}x^2=4
$$

can be understood through:

```text
Algebra
   ↓
lim x→2 x² = 4

Numerical
   ↓
x = 1.99 → f(x) = 3.9601
x = 2.01 → f(x) = 4.0401

Graph
   ↓
Curve approaches (2, 4)
```

A strong understanding should connect all three.

---

# 2.13 Evaluating Simple Limits by Direct Substitution

For many ordinary functions, we can simply substitute.

Consider:

$$
\lim_{x\rightarrow3}(2x+1)
$$

Substitute:

$$
x=3
$$

Then:

$$
2(3)+1=7
$$

Therefore:

$$
\boxed{
\lim_{x\rightarrow3}(2x+1)=7
}
$$

Similarly:

$$
\lim_{x\rightarrow2}x^2=4
$$

Direct substitution works for many functions that are continuous at the point.

---

# 2.14 When Direct Substitution Fails

Consider:

$$
\lim_{x\rightarrow2}
\frac{x^2-4}{x-2}
$$

Direct substitution gives:

$$
\frac00
$$

This is an **indeterminate form**.

It does **not** mean the limit is zero.

It means:

> Direct substitution has not given us enough information to determine the limit.

We need another method.

Here, factoring solves the problem:

$$
\frac{x^2-4}{x-2}
=
\frac{(x-2)(x+2)}{x-2}
$$

For $x\neq2$:

$$
=x+2
$$

Therefore:

$$
\lim_{x\rightarrow2}(x+2)=4
$$

So:

$$
\boxed{4}
$$

---

# 2.15 Important: $\frac00$ Is Not the Answer

This is a very common beginner mistake.

If:

$$
\lim_{x\rightarrow a}f(x)
$$

produces:

$$
\frac00
$$

after direct substitution, don't conclude:

$$
\text{limit}=0
$$

Instead:

> **$0/0$ is an indeterminate form.**

It tells us that we need to investigate the expression further.

Other indeterminate forms we'll eventually encounter include:

$$
\frac{\infty}{\infty}
$$

and:

$$
\infty-\infty
$$

We don't need advanced techniques for these yet.

---

# 2.16 Important Limit Laws

Once we understand the intuition, we can establish some useful rules.

If:

$$
\lim_{x\rightarrow a}f(x)=L
$$

and:

$$
\lim_{x\rightarrow a}g(x)=M
$$

then, under the usual conditions:

### Sum

$$
\boxed{
\lim_{x\rightarrow a}[f(x)+g(x)]
=
L+M
}
$$

### Difference

$$
\boxed{
\lim_{x\rightarrow a}[f(x)-g(x)]
=
L-M
}
$$

### Constant multiple

$$
\boxed{
\lim_{x\rightarrow a}[cf(x)]
=
cL
}
$$

### Product

$$
\boxed{
\lim_{x\rightarrow a}[f(x)g(x)]
=
LM
}
$$

### Quotient

If:

$$
M\neq0
$$

then:

$$
\boxed{
\lim_{x\rightarrow a}
\frac{f(x)}{g(x)}
=
\frac{L}{M}
}
$$

These rules allow us to break complicated limits into simpler ones.

---

# 2.17 Limits at Infinity

So far, we have studied:

$$
x\rightarrow a
$$

where $a$ is a finite number.

But we can also ask:

> What happens to a function as $x$ becomes extremely large?

For example:

$$
f(x)=\frac{1}{x}
$$

As $x$ becomes larger:

$$
x=10,\ 100,\ 1000,\ 10000,\ldots
$$

we get:

$$
\frac1{10},\frac1{100},\frac1{1000},\frac1{10000},\ldots
$$

These values approach $0$.

Therefore:

$$
\boxed{
\lim_{x\rightarrow\infty}\frac1x=0
}
$$

This does **not** mean that $x$ becomes an actual number called infinity.

Infinity represents an unbounded direction:

> $x$ keeps increasing without bound.

---

# 2.18 Infinite Limits

There is another situation.

Consider:

$$
f(x)=\frac1{x^2}
$$

As $x$ approaches $0$:

$$
x^2\rightarrow0^+
$$

and:

$$
\frac1{x^2}
$$

becomes arbitrarily large.

So we write:

$$
\boxed{
\lim_{x\rightarrow0}\frac1{x^2}
=
+\infty
}
$$

This means:

> The function grows without bound as $x$ approaches $0$.

Technically, $+\infty$ is not a real-number output here.

It describes unbounded behavior.

---

# 2.19 Limit Does Not Always Exist

A limit can fail to exist for several reasons.

### 1. Left and right limits differ

For example:

$$
\lim_{x\rightarrow0^-}f(x)=0
$$

but:

$$
\lim_{x\rightarrow0^+}f(x)=1
$$

Therefore, the two-sided limit doesn't exist.

---

### 2. Function oscillates

Consider:

$$
f(x)=\sin\left(\frac1x\right)
$$

as:

$$
x\rightarrow0
$$

The function oscillates faster and faster and does not settle toward one value.

Therefore:

$$
\boxed{
\lim_{x\rightarrow0}\sin\left(\frac1x\right)
\text{ does not exist}
}
$$

This is an important example because it shows:

> A function being bounded does not guarantee that its limit exists.

---

### 3. Function grows without bound

For example:

$$
\frac1{x^2}
$$

as $x\rightarrow0$ becomes arbitrarily large.

Depending on the formal definition being used, we describe this as an **infinite limit**, rather than a finite real-valued limit.

---

# 2.20 Formal Definition — The Intuitive Version First

So far we've used:

> "Gets closer and closer."

Mathematics eventually needs a precise definition of that phrase.

Suppose:

$$
\lim_{x\rightarrow a}f(x)=L
$$

The formal idea is:

> We can make $f(x)$ as close to $L$ as we want by making $x$ sufficiently close to $a$.

This is captured by the **epsilon-delta definition**.

For every:

$$
\varepsilon>0
$$

there exists:

$$
\delta>0
$$

such that:

$$
0<|x-a|<\delta
$$

implies:

$$
|f(x)-L|<\varepsilon
$$

So formally:

$$
\boxed{
\forall\varepsilon>0,\ \exists\delta>0:
0<|x-a|<\delta
\Rightarrow
|f(x)-L|<\varepsilon
}
$$

---

# 2.21 What Does This Complicated Definition Actually Say?

Don't memorize the symbols yet.

Translate them into English.

### $\varepsilon$

How close we demand the **output** to be to $L$.

### $\delta$

How close we need to make the **input** to $a$ to guarantee that output closeness.

So:

```text
We choose how close the output should be
                 │
                 ▼
              ε (epsilon)
                 │
                 ▼
Mathematics tells us how close x
must be to a
                 │
                 ▼
              δ (delta)
```

This is a profound idea:

> **We control the required accuracy of the output, and the definition tells us how much input precision is sufficient to achieve it.**

We will study epsilon-delta limits more deeply only when they become useful, rather than turning the first encounter into a symbol-memorization exercise.

---

# 2.22 Why Is the Limit So Important for Derivatives?

This is the most important reason we are studying limits.

Recall our earlier question:

> What is the slope of a curve at exactly one point?

Suppose we have two points:

$$
(x,f(x))
$$

and:

$$
(x+h,f(x+h))
$$

The slope between them is:

$$
\frac{f(x+h)-f(x)}{h}
$$

This is the **average rate of change**.

But we want the slope at exactly one point.

So we make the second point approach the first:

$$
h\rightarrow0
$$

Therefore:

$$
\boxed{
f'(x)
=
\lim_{h\rightarrow0}
\frac{f(x+h)-f(x)}{h}
}
$$

And there it is.

The derivative is defined using a **limit**.

---

# 2.23 The Secant-to-Tangent Idea

This is worth understanding visually.

With two separated points:

```text
             ●
           /
         /
       ●
```

we have a **secant line**.

As the second point gets closer:

```text
             ●
           /
         ●
```

Then:

```text
           ●
          /
         ●
```

Eventually, in the limiting sense, the secant slope approaches the **tangent slope**.

That limiting slope is the derivative.

So:

$$
\boxed{
\text{Derivative}
=
\text{Limit of average rates of change}
}
$$

This is one of the most important ideas in Calculus.

---

# 2.24 Why ML Ultimately Cares About This

Let's return to Machine Learning.

Suppose:

$$
L(w)
$$

is our loss function.

We want to understand:

> What happens to the loss when $w$ changes slightly?

We can first consider an ordinary finite change:

$$
\frac{L(w+\Delta w)-L(w)}{\Delta w}
$$

This tells us the **average change in loss per unit change in parameter** over a finite interval.

But ML optimization often needs the **local behavior** at the current parameter value.

So we take:

$$
\Delta w\rightarrow0
$$

giving:

$$
\boxed{
\frac{dL}{dw}
=
\lim_{\Delta w\rightarrow0}
\frac{L(w+\Delta w)-L(w)}
{\Delta w}
}
$$

That is the derivative of the loss with respect to the parameter.

And for many parameters:

$$
L(\vec{w})
$$

we eventually get partial derivatives and the gradient:

$$
\nabla L(\vec{w})
$$

So the conceptual chain is:

```text
Function
   ↓
"How does it change?"
   ↓
Average change
   ↓
Make the change smaller
   ↓
Limit
   ↓
Instantaneous/local change
   ↓
Derivative
   ↓
Partial derivatives
   ↓
Gradient
   ↓
Gradient descent
```

This is why understanding limits properly is worth the effort.

---

# 2.25 Limits and Machine Learning — The Right Perspective

It is important not to overstate the role of limits.

You will not normally sit down to train a neural network by manually calculating epsilon-delta limits.

Instead:

> **Limits provide the mathematical foundation that makes the derivative well-defined.**

And derivatives are what we use heavily in:

* optimization
* gradient descent
* neural networks
* backpropagation

So the practical dependency is:

$$
\boxed{
\text{Limit}
\rightarrow
\text{Derivative}
\rightarrow
\text{Gradient}
\rightarrow
\text{Optimization}
}
$$

# 2.27 A More Useful Way to Read a Limit

Whenever you see:

$$
\lim_{x\rightarrow a}f(x)=L
$$

train yourself to read it as a sentence:

> **"When the input $x$ gets arbitrarily close to $a$, the output $f(x)$ gets arbitrarily close to $L$."**

Don't read it merely as:

> "Plug $a$ into the function."

Sometimes substitution works.

Sometimes it doesn't.

The **meaning of the limit is about behavior near the point**.

That is much more fundamental.

---

# 2.28 Common Mistakes

### Mistake 1 — Thinking $x\rightarrow a$ means $x=a$

No.

It means:

> $x$ approaches $a$.

---

### Mistake 2 — Assuming the limit must equal $f(a)$

Not necessarily.

We can have:

$$
\lim_{x\rightarrow a}f(x)=L
$$

while:

$$
f(a)
$$

is undefined or even different from $L$.

---

### Mistake 3 — Thinking $\frac00=0$

No.

$$
\frac00
$$

is an indeterminate form.

It means further analysis is required.

---

### Mistake 4 — Looking only from one side

A two-sided limit requires both:

$$
\lim_{x\rightarrow a^-}f(x)
$$

and:

$$
\lim_{x\rightarrow a^+}f(x)
$$

to agree.

---

### Mistake 5 — Thinking infinity is an ordinary number

Infinity describes unbounded behavior.

It is not simply another real number that we substitute into a function.

---

### Mistake 6 — Memorizing epsilon-delta without understanding it

The symbols matter, but first understand the idea:

> **Control how close the output is by restricting how close the input is.**
