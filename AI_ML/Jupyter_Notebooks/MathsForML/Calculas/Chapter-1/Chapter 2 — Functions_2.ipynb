# 1.5 Independent and Dependent Variables

This is one of the most important ideas to understand before Calculus.

Consider:

$$
y=2x+1
$$

Here, we choose a value for $x$.

The function then determines the value of $y$.

For example:

$$
x=2
$$

gives:

$$
y=2(2)+1=5
$$

So:

```text
        We choose x
             │
             ▼
             2
             │
             │ function
             ▼
        y = 2x + 1
             │
             ▼
             5
```

Therefore:

### Independent variable

The variable whose value we are free to choose or vary.

Here:

$$
\boxed{x=\text{independent variable}}
$$

### Dependent variable

The variable whose value depends on the independent variable.

Here:

$$
\boxed{y=\text{dependent variable}}
$$

because:

$$
y=f(x)
$$

---

## Why Are They Called Independent and Dependent?

Because the output depends on the input.

If:

$$
y=f(x)
$$

then changing $x$ can cause $y$ to change.

For example:

$$
y=x^2
$$

| $x$ | $y$ |
| --: | --: |
|   1 |   1 |
|   2 |   4 |
|   3 |   9 |
|   4 |  16 |

Here, $y$ depends on $x$.

So we can think:

$$
\boxed{x\rightarrow y}
$$

or:

$$
\boxed{\text{independent variable}\rightarrow\text{dependent variable}}
$$

---

# 1.6 A Subtle but Important Point

"Independent" does **not always mean that the variable has absolutely no relationship with anything else."

It means that **within the function being considered**, it is treated as the input variable.

For example:

$$
y=f(x)
$$

we treat $x$ as the independent variable and $y$ as dependent.

Later, in Machine Learning, we may have:

$$
\hat y=f(\vec{x};\vec{w})
$$

Here:

* $\vec{x}$ represents input features
* $\vec{w}$ represents parameters
* $\hat y$ represents the output

The word **dependent** simply tells us that the output is determined by the function given the relevant inputs and parameters.

This distinction will become increasingly important when we study derivatives with respect to **inputs versus parameters**.

---

# 1.7 Functions as Mappings

There is a more formal way of thinking about a function.

A function is a **mapping** from one set to another.

We write:

$$
f:A\rightarrow B
$$

This means:

> Function $f$ maps elements from set $A$ to elements of set $B$.

For example:

$$
A=\{1,2,3\}
$$

and:

$$
B=\{2,4,6\}
$$

with:

$$
f(x)=2x
$$

Then:

$$
f(1)=2
$$

$$
f(2)=4
$$

$$
f(3)=6
$$

Conceptually:

```text
A                         B

1 ─────────────────────► 2
2 ─────────────────────► 4
3 ─────────────────────► 6
```

The important idea is:

> A function establishes a rule for mapping each valid input to exactly one output.

---

## Why Is the Mapping View Important?

Because it scales naturally to Machine Learning.

A simple function:

$$
f:\mathbb{R}\rightarrow\mathbb{R}
$$

might map one number to another.

For example:

$$
f(x)=x^2
$$

But an ML model might map a vector of features to a prediction:

$$
f:\mathbb{R}^n\rightarrow\mathbb{R}
$$

For example:

$$
\vec{x}\in\mathbb{R}^n
$$

could contain $n$ features, while:

$$
\hat y\in\mathbb{R}
$$

is the predicted value.

So:

$$
\boxed{
\mathbb{R}^n
\rightarrow
\mathbb{R}
}
$$

is already a useful mathematical description of many ML models.

# 1.8 Linear Functions

Now let's distinguish **linear and nonlinear functions**.

Consider:

$$
f(x)=2x+3
$$

This produces a straight line when plotted.

In elementary ML discussions, this is commonly called a **linear function** or linear relationship.

Its general form is:

$$
\boxed{f(x)=mx+b}
$$

where:

* $m$ = slope
* $b$ = intercept

For example:

$$
f(x)=2x+3
$$

has:

$$
m=2
$$

and:

$$
b=3
$$

### Jupyter Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 200)
y = 2 * x + 3

plt.figure(figsize=(8, 5))
plt.plot(x, y)

plt.axhline(0)
plt.axvline(0)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Linear Function: f(x) = 2x + 3")
plt.grid(True)

plt.show()
```

The important visual characteristic is:

> **The graph is a straight line.**

---

# 1.9 Nonlinear Functions

Now consider:

$$
f(x)=x^2
$$

Its graph is curved.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 200)
y = x**2

plt.figure(figsize=(8, 5))
plt.plot(x, y)

plt.axhline(0)
plt.axvline(0)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Nonlinear Function: f(x) = x²")
plt.grid(True)

plt.show()
```

This is a **nonlinear function**.

Other examples include:

$$
f(x)=x^3
$$

$$
f(x)=\sin(x)
$$

$$
f(x)=e^x
$$

$$
f(x)=\log(x)
$$

The key intuition is:

> A nonlinear relationship does not have a constant rate of change across its entire domain.

That statement will become much more meaningful once we study derivatives.

---

# 1.10 Linear vs Nonlinear

| Property         | Linear            | Nonlinear                         |
| ---------------- | ----------------- | --------------------------------- |
| Typical graph    | Straight line     | Curve or other non-straight shape |
| Example          | $f(x)=2x+3$       | $f(x)=x^2$                        |
| Slope            | Constant          | Generally changes                 |
| Example ML model | Linear Regression | Neural network / nonlinear model  |

### Important ML nuance

There is a subtle point here.

In ML, we often call:

$$
\hat y=w_1x_1+w_2x_2+\cdots+w_nx_n+b
$$

a **linear model**.

Technically, as a function of the parameters and features, the presence of $b$ means the function is **affine**, not strictly linear in the mathematical linear-algebra sense.

A strictly linear transformation satisfies:

$$
f(\vec{x}+\vec{z})
=
f(\vec{x})+f(\vec{z})
$$

and:

$$
f(c\vec{x})=cf(\vec{x})
$$

which means:

$$
f(\vec{0})=\vec{0}
$$

The bias term means:

$$
f(\vec{0})=b
$$

rather than necessarily zero.

So:

> In ML, "linear model" is common terminology, while mathematically $w^Tx+b$ is an **affine function**.

We will revisit this distinction when Linear Algebra and ML meet.

# 1.11 Graphical Interpretation

A graph is a visual representation of how a function maps inputs to outputs.

For:

$$
y=f(x)
$$

we normally place:

* $x$ on the horizontal axis
* $y$ on the vertical axis

Every point on the graph represents:

$$
(x,f(x))
$$

For example, if:

$$
f(x)=x^2
$$

then the point corresponding to $x=2$ is:

$$
(2,f(2))
$$

Since:

$$
f(2)=4
$$

the point is:

$$
(2,4)
$$

---

## What Does the Graph Tell Us?

A graph lets us visually inspect the relationship between input and output.

We can see:

* whether the function increases
* whether it decreases
* whether it is constant
* whether it is curved
* where it reaches high or low values
* whether there are discontinuities
* how rapidly its behavior appears to change

For example:

$$
f(x)=x^2
$$

has a minimum at:

$$
x=0
$$

and becomes larger as we move away from zero.

### Jupyter Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 400)
y = x**2

plt.figure(figsize=(8, 5))
plt.plot(x, y)

plt.axhline(0)
plt.axvline(0)

plt.xlabel("Input, x")
plt.ylabel("Output, f(x)")
plt.title("Graphical Interpretation of f(x) = x²")
plt.grid(True)

plt.show()
```

The important thing is not simply seeing the curve.

We should learn to **read the graph mathematically**.

# 1.12 Multivariable Functions

So far we've considered functions with one input variable:

$$
y=f(x)
$$

But real-world problems often depend on **multiple variables**.

For example, the price of a house may depend on:

* area
* number of bedrooms
* age
* location

So instead of:

$$
y=f(x)
$$

we might have:

$$
y=f(x_1,x_2,x_3,x_4)
$$

This is a **multivariable function**.

---

## Simple Example

Consider:

$$
f(x,y)=x^2+y^2
$$

Now the function has **two input variables**:

$$
x
$$

and:

$$
y
$$

and produces one output:

$$
z=f(x,y)
$$

For example:

$$
f(2,3)
=
2^2+3^2
$$

$$
=4+9
$$

$$
=13
$$

So:

$$
(2,3)\rightarrow13
$$

---

# 1.13 Why Multivariable Functions Matter So Much in ML

A machine-learning model usually doesn't receive just one feature.

Suppose we have:

$$
\vec{x}
=
\begin{bmatrix}
x_1\\
x_2\\
x_3
\end{bmatrix}
$$

Then the model could be:

$$
\hat y=f(x_1,x_2,x_3)
$$

For Linear Regression:

$$
\hat y
=
w_1x_1+w_2x_2+w_3x_3+b
$$

So:

```text
x₁ ──┐
     │
x₂ ──┼──► ML Function ──► ŷ
     │
x₃ ──┘
```

Now Calculus has a new problem:

> If there are many inputs or parameters, how do we determine how the output changes with respect to **one particular variable**?

This leads directly to:

$$
\boxed{\text{Partial Derivatives}}
$$

and eventually:

$$
\boxed{\text{Gradient}}
$$

---

# 1.14 Visualizing a Multivariable Function

For:

$$
z=f(x,y)=x^2+y^2
$$

the output depends on **two inputs**.

Therefore, instead of a normal 2D graph, we need a 3D surface to visualize the relationship:

```text
              z
              ↑
             / \
            /   \
           /     \
          /       \
         ───────────► x
        /
       /
      y
```

We will use 3D Matplotlib visualizations later when we study multivariable Calculus and gradients.

For now, the key conceptual point is:

> **One input → one-variable function.**

> **Multiple inputs → multivariable function.**

# 1.15 Function Composition

Now let's look at another extremely important concept.

Suppose we have two functions:

$$
g(x)=x+1
$$

and:

$$
f(x)=x^2
$$

We can take the output of $g$ and use it as the input of $f$.

First:

$$
x
\overset{g}{\longrightarrow}
g(x)
$$

Then:

$$
g(x)
\overset{f}{\longrightarrow}
f(g(x))
$$

This is called **function composition**.

---

## Example

We have:

$$
g(x)=x+1
$$

and:

$$
f(x)=x^2
$$

Therefore:

$$
f(g(x))
=
(g(x))^2
$$

Since:

$$
g(x)=x+1
$$

we get:

$$
\boxed{f(g(x))=(x+1)^2}
$$

---

## Numerical Example

Take:

$$
x=2
$$

First apply $g$:

$$
g(2)=2+1=3
$$

Then apply $f$:

$$
f(3)=3^2=9
$$

Therefore:

$$
f(g(2))=9
$$

Conceptually:

```text
x = 2
  │
  ▼
┌─────────┐
│ g(x)=x+1│
└────┬────┘
     │
     ▼
     3
     │
     ▼
┌─────────┐
│ f(x)=x² │
└────┬────┘
     │
     ▼
     9
```

---

# 1.16 Order Matters in Composition

This is important.

In general:

$$
f(g(x))\neq g(f(x))
$$

Using our example:

$$
f(x)=x^2
$$

$$
g(x)=x+1
$$

We already calculated:

$$
f(g(x))
=
(x+1)^2
$$

But:

$$
g(f(x))
=
x^2+1
$$

These are not generally equal.

For $x=2$:

$$
f(g(2))=9
$$

but:

$$
g(f(2))=5
$$

Therefore:

> **The order in which functions are composed matters.**

---

# 1.17 Why Function Composition Matters in ML

This is one of the most important connections we will make.

A neural network can be viewed as a sequence of functions.

Conceptually:

$$
\vec{x}
\rightarrow
f_1
\rightarrow
f_2
\rightarrow
f_3
\rightarrow
\hat y
$$

Mathematically:

$$
\hat y
=
f_3(f_2(f_1(\vec{x})))
$$

That is simply **function composition**.

And now we can see why the Calculus roadmap is structured the way it is:

```text
Functions
    ↓
Function Composition
    ↓
How does a function change?
    ↓
Derivative
    ↓
Chain Rule
    ↓
Gradient
    ↓
Backpropagation
```

The **chain rule** exists partly to help us differentiate compositions of functions.

And neural networks are built from compositions.

This is a major bridge between Calculus and Deep Learning.

# 1.18 Functions: Putting Everything Together

We can now describe a function from several complementary perspectives.

### As a rule

> A function is a rule that maps an input to exactly one output.

### As a mapping

$$
f:A\rightarrow B
$$

### As an equation

$$
y=f(x)
$$

### As a graph

A visual representation of:

$$
(x,f(x))
$$

### As a transformation

```text
Input → Function → Output
```

### As an ML model

$$
\hat y=f(\vec{x};\vec{w})
$$

### As a composition

$$
F(x)=f_3(f_2(f_1(x)))
$$

These aren't different concepts.

They are **different ways of looking at the same underlying idea**.

# 1.19 Why This Chapter Matters for Calculus

We can now ask the question that motivates the next chapter.

We have:

$$
y=f(x)
$$

We understand:

> Given $x$, the function produces $y$.

But Calculus asks a deeper question:

> **What happens to $y$ when $x$ changes?**

For example:

$$
x\rightarrow x+\Delta x
$$

What happens to:

$$
f(x)\rightarrow f(x+\Delta x)?
$$

We can initially measure the change using:

$$
\frac{\Delta y}{\Delta x}
$$

which means:

> **How much does the output change compared with the change in the input?**

And when we make that change increasingly small, we arrive at the central idea of the next chapter:

$$
\boxed{\text{Limit}}
$$

which eventually gives us:

$$
\boxed{\text{Derivative}}
$$

# 🧠 The Complete Mental Model

Think of a function as a **transformation machine**:

```text
                         FUNCTION
                            │
                            │ rule
                            ▼
                  ┌──────────────────┐
Input ───────────►│                  │──────────► Output
                  │   Transformation │
                  │                  │
                  └──────────────────┘
```

For one variable:

$$
y=f(x)
$$

For multiple variables:

$$
y=f(x_1,x_2,\ldots,x_n)
$$

For Machine Learning:

$$
\hat y=f(\vec{x};\vec{w})
$$

For a neural network:

$$
\hat y=f_3(f_2(f_1(\vec{x})))
$$

And now Calculus enters:

> **What happens when one of these inputs or parameters changes?**

That is the question we will spend much of the rest of this journey answering.
