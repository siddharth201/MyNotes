# Chapter 3 — Derivatives

## Part 9 — Product and Quotient Rules

In the previous parts, we learned how to differentiate:

- Constants
- Powers
- Polynomial terms
- Exponential functions
- Logarithmic functions
- Trigonometric functions

We also learned that sums and differences can be differentiated term-by-term.

But what happens when **two functions are multiplied together**?

For example:

$$
f(x)=x^2\sin x
$$

We cannot simply differentiate each part and multiply the results:

$$
\frac{d}{dx}(x^2\sin x)
\neq
\frac{d}{dx}(x^2)
\frac{d}{dx}(\sin x)
$$

The same problem occurs when one function is divided by another:

$$
f(x)=\frac{x^2}{e^x}
$$

This is why we need two new rules:

1. **Product Rule**
2. **Quotient Rule**

These rules are fundamental because many real-world and Machine Learning expressions contain products and ratios of changing quantities.

# 9.1 Why Do We Need the Product Rule?

Suppose:

$$
y=f(x)g(x)
$$

Both:

$$
f(x)
$$

and:

$$
g(x)
$$

can change when $x$ changes.

Therefore, the product:

$$
f(x)g(x)
$$

changes for **two reasons**:

1. $f(x)$ changes.
2. $g(x)$ changes.

This is the key reason we cannot simply multiply the two derivatives.

### A simple analogy

Imagine the area of a rectangle:

$$
A=LW
$$

where:

- $L$ = length
- $W$ = width

If both length and width change, the area changes because:

- the length changed
- the width changed

So we need to account for **both sources of change**.

The Product Rule does exactly this.

# 9.2 Product Rule

Suppose:

$$
y=f(x)g(x)
$$

Then:

$$
\boxed{
\frac{d}{dx}[f(x)g(x)]
=
f'(x)g(x)+f(x)g'(x)
}
$$

In prime notation:

$$
\boxed{
(fg)'=f'g+fg'
}
$$

### In words

> **Derivative of the first × the second + first × derivative of the second.**

Notice that the rule contains **two terms**.

$$
\underbrace{f'(x)g(x)}_{\text{first changes}}
+
\underbrace{f(x)g'(x)}_{\text{second changes}}
$$

This reflects the two ways the product can change.

## A Very Important Warning

A common incorrect approach is:

$$
(fg)'=f'g'
$$

This is **wrong**.

The correct rule is:

$$
\boxed{
(fg)'=f'g+fg'
}
$$

Why are there two terms?

Because when $x$ changes, **both factors can change**.

### Mental Model

Think:

$$
\boxed{
\text{Product changes}
=
\text{change from first factor}
+
\text{change from second factor}
}
$$

That is the intuition behind the Product Rule.

# 9.3 Product Rule — Visual Intuition

Suppose:

$$
A=f(x)g(x)
$$

Imagine $f(x)$ is the width of a rectangle and $g(x)$ is its height.

Now increase $x$ slightly.

Both dimensions may change:

$$
f(x)\rightarrow f(x)+\Delta f
$$

and:

$$
g(x)\rightarrow g(x)+\Delta g
$$

The new area is:

$$
[f(x)+\Delta f][g(x)+\Delta g]
$$

Expanding:

$$
=
f(x)g(x)
+
g(x)\Delta f
+
f(x)\Delta g
+
\Delta f\Delta g
$$

So the change in area is:

$$
\Delta A
=
g(x)\Delta f
+
f(x)\Delta g
+
\Delta f\Delta g
$$

The first two terms represent the two main sources of change.

The final term:

$$
\Delta f\Delta g
$$

is a very small second-order change.

As the change becomes infinitesimally small, this term becomes negligible in the derivative limit.

This is why the final derivative contains:

$$
\boxed{
f'g+fg'
}
$$
