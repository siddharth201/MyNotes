# Chapter 3 — Derivatives

## Part 9 — Product and Quotient Rules

In the previous parts, we learned how to differentiate:

- Constants
- Powers
- Polynomial terms
- Exponential functions
- Logarithmic functions
- Trigonometric functions

We also learned that sums and differences can be differentiated term-by-term.

But what happens when **two functions are multiplied together**?

For example:

$$
f(x)=x^2\sin x
$$

We cannot simply differentiate each part and multiply the results:

$$
\frac{d}{dx}(x^2\sin x)
\neq
\frac{d}{dx}(x^2)
\frac{d}{dx}(\sin x)
$$

The same problem occurs when one function is divided by another:

$$
f(x)=\frac{x^2}{e^x}
$$

This is why we need two new rules:

1. **Product Rule**
2. **Quotient Rule**

These rules are fundamental because many real-world and Machine Learning expressions contain products and ratios of changing quantities.

# 9.1 Why Do We Need the Product Rule?

Suppose:

$$
y=f(x)g(x)
$$

Both:

$$
f(x)
$$

and:

$$
g(x)
$$

can change when $x$ changes.

Therefore, the product:

$$
f(x)g(x)
$$

changes for **two reasons**:

1. $f(x)$ changes.
2. $g(x)$ changes.

This is the key reason we cannot simply multiply the two derivatives.

### A simple analogy

Imagine the area of a rectangle:

$$
A=LW
$$

where:

- $L$ = length
- $W$ = width

If both length and width change, the area changes because:

- the length changed
- the width changed

So we need to account for **both sources of change**.

The Product Rule does exactly this.

# 9.2 Product Rule

Suppose:

$$
y=f(x)g(x)
$$

Then:

$$
\boxed{
\frac{d}{dx}[f(x)g(x)]
=
f'(x)g(x)+f(x)g'(x)
}
$$

In prime notation:

$$
\boxed{
(fg)'=f'g+fg'
}
$$

### In words

> **Derivative of the first × the second + first × derivative of the second.**

Notice that the rule contains **two terms**.

$$
\underbrace{f'(x)g(x)}_{\text{first changes}}
+
\underbrace{f(x)g'(x)}_{\text{second changes}}
$$

This reflects the two ways the product can change.

## A Very Important Warning

A common incorrect approach is:

$$
(fg)'=f'g'
$$

This is **wrong**.

The correct rule is:

$$
\boxed{
(fg)'=f'g+fg'
}
$$

Why are there two terms?

Because when $x$ changes, **both factors can change**.

### Mental Model

Think:

$$
\boxed{
\text{Product changes}
=
\text{change from first factor}
+
\text{change from second factor}
}
$$

That is the intuition behind the Product Rule.

# 9.3 Product Rule — Visual Intuition

Suppose:

$$
A=f(x)g(x)
$$

Imagine $f(x)$ is the width of a rectangle and $g(x)$ is its height.

Now increase $x$ slightly.

Both dimensions may change:

$$
f(x)\rightarrow f(x)+\Delta f
$$

and:

$$
g(x)\rightarrow g(x)+\Delta g
$$

The new area is:

$$
[f(x)+\Delta f][g(x)+\Delta g]
$$

Expanding:

$$
=
f(x)g(x)
+
g(x)\Delta f
+
f(x)\Delta g
+
\Delta f\Delta g
$$

So the change in area is:

$$
\Delta A
=
g(x)\Delta f
+
f(x)\Delta g
+
\Delta f\Delta g
$$

The first two terms represent the two main sources of change.

The final term:

$$
\Delta f\Delta g
$$

is a very small second-order change.

As the change becomes infinitesimally small, this term becomes negligible in the derivative limit.

This is why the final derivative contains:

$$
\boxed{
f'g+fg'
}
$$

# 9.4 Derivation of the Product Rule

Let's derive the rule from first principles.

Let:

$$
F(x)=f(x)g(x)
$$

The derivative is:

$$
F'(x)
=
\lim_{h\to0}
\frac{F(x+h)-F(x)}{h}
$$

Substitute:

$$
F(x+h)=f(x+h)g(x+h)
$$

Therefore:

$$
F'(x)
=
\lim_{h\to0}
\frac{
f(x+h)g(x+h)-f(x)g(x)
}{h}
$$

The expression does not immediately separate into useful pieces.

So we use a mathematical trick: **add and subtract the same quantity**.

Add:

$$
f(x+h)g(x)
$$

and subtract:

$$
f(x+h)g(x)
$$

Then:

$$
F'(x)
=
\lim_{h\to0}
\frac{
f(x+h)g(x+h)
-
f(x+h)g(x)
+
f(x+h)g(x)
-
f(x)g(x)
}{h}
$$

## Continue the Derivation

Group the terms:

$$
F'(x)
=
\lim_{h\to0}
\left[
\frac{
f(x+h)[g(x+h)-g(x)]
}{h}
+
\frac{
g(x)[f(x+h)-f(x)]
}{h}
\right]
$$

Now separate the limits:

$$
F'(x)
=
\lim_{h\to0}f(x+h)
\frac{g(x+h)-g(x)}{h}
+
g(x)
\lim_{h\to0}
\frac{f(x+h)-f(x)}{h}
$$

As:

$$
h\to0
$$

we have:

$$
f(x+h)\to f(x)
$$

and:

$$
\frac{g(x+h)-g(x)}{h}
\to g'(x)
$$

while:

$$
\frac{f(x+h)-f(x)}{h}
\to f'(x)
$$

Therefore:

$$
F'(x)
=
f(x)g'(x)+g(x)f'(x)
$$

Rearranging the order:

$$
\boxed{
F'(x)=f'(x)g(x)+f(x)g'(x)
}
$$

This is the Product Rule.

# 9.5 Product Rule — Step-by-Step Example

Consider:

$$
f(x)=x^2\sin x
$$

We have two factors:

$$
u=x^2
$$

and:

$$
v=\sin x
$$

Differentiate each:

$$
u'=2x
$$

and:

$$
v'=\cos x
$$

Apply the Product Rule:

$$
(uv)'=u'v+uv'
$$

Therefore:

$$
f'(x)
=
(2x)(\sin x)
+
(x^2)(\cos x)
$$

So:

$$
\boxed{
f'(x)=2x\sin x+x^2\cos x
}
$$

# 9.6 Another Product Rule Example

Consider:

$$
f(x)=x^3e^x
$$

Let:

$$
u=x^3
$$

and:

$$
v=e^x
$$

Then:

$$
u'=3x^2
$$

and:

$$
v'=e^x
$$

Apply the Product Rule:

$$
f'(x)
=
u'v+uv'
$$

Therefore:

$$
f'(x)
=
3x^2e^x+x^3e^x
$$

We can factor:

$$
f'(x)
=
e^x(3x^2+x^3)
$$

Thus:

$$
\boxed{
f'(x)=e^x(x^3+3x^2)
}
$$

# 9.7 Product Rule With More Than Two Factors

Suppose:

$$
f(x)=u(x)v(x)w(x)
$$

The Product Rule can be applied repeatedly.

First group two factors:

$$
f(x)=[u(x)v(x)]w(x)
$$

Differentiate:

$$
f'(x)
=
[uv]'w
+
[uv]w'
$$

Now:

$$
[uv]'=u'v+uv'
$$

Therefore:

$$
\boxed{
f'
=
u'vw
+
uv'w
+
uvw'
}
$$

The pattern is:

> Each factor gets differentiated once, while the other factors remain unchanged.

For three factors, there are three terms.

For more factors, the same principle continues.

# 9.8 Quotient Rule

Now consider division.

Suppose:

$$
y=\frac{f(x)}{g(x)}
$$

where:

$$
g(x)\neq0
$$

The derivative is:

$$
\boxed{
\frac{d}{dx}
\left[
\frac{f(x)}{g(x)}
\right]
=
\frac{
g(x)f'(x)-f(x)g'(x)
}{
[g(x)]^2
}
}
$$

In prime notation:

$$
\boxed{
\left(\frac{f}{g}\right)'
=
\frac{gf'-fg'}{g^2}
}
$$

### Memory pattern

A useful way to remember it is:

> **Bottom × derivative of top − top × derivative of bottom, all over bottom squared.**

That is:

$$
\boxed{
\frac{
\text{bottom}\cdot\text{derivative of top}
-
\text{top}\cdot\text{derivative of bottom}
}{
(\text{bottom})^2
}
}
$$

# 9.9 Why Do We Need the Quotient Rule?

Suppose:

$$
y=\frac{f(x)}{g(x)}
$$

Both the numerator and denominator may change when $x$ changes.

Therefore, the ratio can change because:

- the numerator changes,
- the denominator changes.

This is similar to the Product Rule, except division creates a different mathematical relationship.

The denominator also appears **squared** in the final formula:

$$
\boxed{
\frac{gf'-fg'}{g^2}
}
$$

This is why we cannot simply write:

$$
\frac{f'}{g'}
$$

That is generally incorrect.

### Important

In general:

$$
\boxed{
\frac{d}{dx}
\left(
\frac{f}{g}
\right)
\neq
\frac{f'}{g'}
}
$$

# 9.10 Derivation of the Quotient Rule

Start with:

$$
y=\frac{f(x)}{g(x)}
$$

Rewrite division as multiplication:

$$
y=f(x)[g(x)]^{-1}
$$

Now use the Product Rule:

$$
y'
=
f'(x)[g(x)]^{-1}
+
f(x)\frac{d}{dx}[g(x)^{-1}]
$$

Using the Power Rule together with the Chain Rule idea that will be formally developed later:

$$
\frac{d}{dx}[g(x)^{-1}]
=
-g(x)^{-2}g'(x)
$$

Therefore:

$$
y'
=
\frac{f'(x)}{g(x)}
-
\frac{f(x)g'(x)}{g(x)^2}
$$

Put both terms over the common denominator $g(x)^2$:

$$
y'
=
\frac{
g(x)f'(x)-f(x)g'(x)
}{
g(x)^2
}
$$

Therefore:

$$
\boxed{
\left(\frac{f}{g}\right)'
=
\frac{gf'-fg'}{g^2}
}
$$

This shows that the Quotient Rule is closely related to the Product Rule and the derivative of a reciprocal.

# 9.11 Quotient Rule — Example 1

Consider:

$$
f(x)=\frac{x^2}{x+1}
$$

Identify:

$$
u=x^2
$$

and:

$$
v=x+1
$$

Then:

$$
u'=2x
$$

and:

$$
v'=1
$$

Apply the Quotient Rule:

$$
f'(x)
=
\frac{
v u'-u v'
}{
v^2
}
$$

Substitute:

$$
f'(x)
=
\frac{
(x+1)(2x)-x^2(1)
}{
(x+1)^2
}
$$

Expand the numerator:

$$
=
\frac{
2x^2+2x-x^2
}{
(x+1)^2
}
$$

Therefore:

$$
\boxed{
f'(x)
=
\frac{x^2+2x}{(x+1)^2}
}
$$

We could also write:

$$
\boxed{
f'(x)
=
\frac{x(x+2)}{(x+1)^2}
}
$$

# 9.12 Quotient Rule — Example 2

Consider:

$$
f(x)=\frac{\sin x}{x}
$$

where:

$$
x\neq0
$$

Let:

$$
u=\sin x
$$

and:

$$
v=x
$$

Then:

$$
u'=\cos x
$$

and:

$$
v'=1
$$

Apply the Quotient Rule:

$$
f'(x)
=
\frac{
x\cos x-\sin x(1)
}{
x^2
}
$$

Therefore:

$$
\boxed{
f'(x)
=
\frac{x\cos x-\sin x}{x^2}
}
$$

# 9.13 Product Rule vs Quotient Rule

It is useful to compare them directly.

### Product Rule

For:

$$
y=f(x)g(x)
$$

we have:

$$
\boxed{
y'=f'g+fg'
}
$$

Think:

> **First derivative × second + first × second derivative**

---

### Quotient Rule

For:

$$
y=\frac{f(x)}{g(x)}
$$

we have:

$$
\boxed{
y'
=
\frac{gf'-fg'}{g^2}
}
$$

Think:

> **Bottom × derivative of top − top × derivative of bottom, over bottom squared.**

### Do not confuse them

Product:

$$
\boxed{f'g+fg'}
$$

Quotient:

$$
\boxed{\frac{gf'-fg'}{g^2}}
$$

# 9.14 When You May Not Need the Quotient Rule

Sometimes a quotient can be rewritten using negative powers.

For example:

$$
f(x)=\frac{1}{x^3}
$$

Rewrite:

$$
f(x)=x^{-3}
$$

Then the Power Rule is simpler:

$$
f'(x)
=
-3x^{-4}
$$

Therefore:

$$
\boxed{
f'(x)=-\frac3{x^4}
}
$$

You do not need the Quotient Rule here.

### Practical rule

Before applying the Quotient Rule, ask:

> **Can I rewrite the expression using powers and use a simpler rule?**

For simple denominators such as:

$$
x^n
$$

the answer is often yes.

For something like:

$$
\frac{x^2+1}{e^x+\sin x}
$$

the Quotient Rule is usually the natural choice.

# 9.15 Common Mistakes

## Mistake 1 — Multiplying the derivatives

Wrong:

$$
(fg)'=f'g'
$$

Correct:

$$
\boxed{
(fg)'=f'g+fg'
}
$$

---

## Mistake 2 — Dividing the derivatives

Wrong:

$$
\left(\frac fg\right)'
=
\frac{f'}{g'}
$$

Correct:

$$
\boxed{
\left(\frac fg\right)'
=
\frac{gf'-fg'}{g^2}
}
$$

---

## Mistake 3 — Forgetting the second product-rule term

Wrong:

$$
(fg)'=f'g
$$

Correct:

$$
\boxed{
(fg)'=f'g+fg'
}
$$

Both factors can change.

---

## Mistake 4 — Forgetting the minus sign in the quotient rule

Wrong:

$$
\frac{gf'+fg'}{g^2}
$$

Correct:

$$
\boxed{
\frac{gf'-fg'}{g^2}
}
$$

---

## Mistake 5 — Forgetting to square the denominator

Wrong:

$$
\frac{gf'-fg'}{g}
$$

Correct:

$$
\boxed{
\frac{gf'-fg'}{g^2}
}
$$

# 9.16 Connection to Machine Learning

Product and quotient structures occur frequently in Machine Learning.

Consider a simple model:

$$
y=wx
$$

Here both quantities can vary in a broader optimization setting.

Or consider a loss containing a product:

$$
L(w)=w^2e^w
$$

We need the Product Rule:

$$
\frac{dL}{dw}
=
(2w)e^w+w^2e^w
$$

Therefore:

$$
\boxed{
\frac{dL}{dw}
=
e^w(2w+w^2)
}
$$

Similarly, probabilistic and normalized models often contain ratios.

For example:

$$
p(x)=\frac{f(x)}{g(x)}
$$

The derivative of such an expression requires the Quotient Rule unless we can simplify or rewrite it more conveniently.

### Why this matters

Real ML derivatives are rarely just:

$$
x^n
$$

They are often combinations such as:

$$
\text{polynomial}
\times
\text{exponential}
$$

or:

$$
\frac{\text{one changing quantity}}
{\text{another changing quantity}}
$$

The Product and Quotient Rules allow us to break these expressions into manageable pieces.

# 9.17 A Reliable Workflow

When you see a multiplication:

$$
f(x)g(x)
$$

ask:

> **Is this a product of two functions?**

If yes, use:

$$
\boxed{
f'g+fg'
}
$$

---

When you see a division:

$$
\frac{f(x)}{g(x)}
$$

ask:

> **Is the denominator also a function of $x$?**

If yes, consider the Quotient Rule:

$$
\boxed{
\frac{gf'-fg'}{g^2}
}
$$

---

### Before using the Quotient Rule

Check whether rewriting makes the problem simpler.

For example:

$$
\frac1{x^4}
=
x^{-4}
$$

Power Rule is simpler.

But:

$$
\frac{x^2+1}{x+e^x}
$$

naturally suggests the Quotient Rule.

# 9.19 Key Takeaways

### Product Rule

When two functions are multiplied:

$$
\boxed{
(fg)'=f'g+fg'
}
$$

The product can change because **either factor can change**.

---

### Quotient Rule

When one function is divided by another:

$$
\boxed{
\left(\frac fg\right)'
=
\frac{gf'-fg'}{g^2}
}
$$

The denominator must be nonzero.

---

### The most important conceptual lesson

Do not think of the Product Rule as a formula to memorize blindly.

Think:

> **When two changing quantities are multiplied, the total change has contributions from both quantities.**

Similarly, for a quotient:

> **When two changing quantities form a ratio, both the numerator and denominator affect the rate of change.**

### Rule-selection mental model

$$
\boxed{
\text{Addition/Subtraction}
\rightarrow
\text{Sum/Difference Rule}
}
$$

$$
\boxed{
\text{Power}
\rightarrow
\text{Power Rule}
}
$$

$$
\boxed{
\text{Multiplication}
\rightarrow
\text{Product Rule}
}
$$

$$
\boxed{
\text{Division}
\rightarrow
\text{Quotient Rule}
}
$$

These rules are the next major building blocks toward differentiating the complex expressions that appear in Machine Learning.
