# Chapter 3 — Derivatives

## Part 4 — What the Derivative Tells Us

In Part 3, we learned the geometric meaning of a derivative.

For a function:

$$
y=f(x)
$$

the derivative at a point tells us the **slope of the tangent line** at that point.

We can therefore use the derivative to understand how the function is behaving locally.

The derivative can tell us:

- whether the function is increasing or decreasing,
- whether it is increasing or decreasing at a particular point,
- whether the tangent is horizontal,
- how strongly the output responds to a small input change.

The central idea is:

$$
\boxed{
\text{The sign and magnitude of the derivative describe local behavior.}
}
$$

We will build this carefully rather than treating these as rules to memorize.

## 3.4.1 Increasing and Decreasing Functions

Before connecting derivatives to increasing and decreasing behavior, let's understand what these terms mean.

A function is **increasing** over an interval if larger values of $x$ produce larger values of $f(x)$.

In simple language:

> As we move from left to right, the graph generally goes upward.

For example:

$$
f(x)=x
$$

As $x$ increases:

$$
1\rightarrow2\rightarrow3
$$

the output also increases:

$$
1\rightarrow2\rightarrow3
$$

Therefore, the function is increasing.

---

A function is **decreasing** over an interval if larger values of $x$ produce smaller values of $f(x)$.

In simple language:

> As we move from left to right, the graph generally goes downward.

For example:

$$
f(x)=-x
$$

As $x$ increases:

$$
1\rightarrow2\rightarrow3
$$

the output decreases:

$$
-1\rightarrow-2\rightarrow-3
$$

Therefore, the function is decreasing.

## 3.4.2 Increasing vs Decreasing Is a Local Idea

A function does not have to be increasing or decreasing everywhere.

Consider:

$$
f(x)=x^2
$$

Its graph is a parabola.

For negative values of $x$, as we move from left to right, the function goes downward.

For positive values of $x$, as we move from left to right, the function goes upward.

So:

$$
f(x)=x^2
$$

is:

- decreasing for $x<0$,
- increasing for $x>0$.

At:

$$
x=0
$$

the graph reaches its lowest point.

This shows why we need a way to describe the behavior **at a particular location**.

The derivative provides exactly this information.

For:

$$
f(x)=x^2
$$

the derivative is:

$$
f'(x)=2x
$$

Therefore:

- if $x<0$, then $f'(x)<0$,
- if $x>0$, then $f'(x)>0$,
- if $x=0$, then $f'(x)=0$.

The derivative captures the change in behavior of the function.

````python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2

x = np.linspace(-3, 3, 400)
y = f(x)

plt.figure(figsize=(8, 5))

plt.plot(x, y, label=r"$f(x)=x^2$")
plt.axvline(0, linestyle="--", label="x = 0")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Increasing and Decreasing Behavior of f(x) = x²")
plt.grid(True)
plt.legend()

plt.show()
````

## 3.4.3 Positive Derivative

Suppose at a particular point $x=a$:

$$
f'(a)>0
$$

Since the derivative represents the slope of the tangent line, this means:

> **The tangent line slopes upward from left to right.**

Therefore, locally, the function is increasing.

We can visualize this as:

```text
          /
         /
        /
       /
------●----------------
      a

The tangent rises as we move to the right.

Therefore:

$$ \boxed{ f'(a)>0 \Rightarrow \text{function is locally increasing} } $$
Example

Consider:

$$ f(x)=x^2 $$

At:

$$ x=2 $$

we have:

$$ f'(x)=2x $$

so:

$$ f'(2)=4 $$

Since:

$$ 4>0 $$

the function is increasing at $x=2$.

The tangent line has positive slope.

Important wording

A positive derivative tells us about the function's local behavior.

It does not mean that the function must be increasing everywhere.


## 3.4.4 Negative Derivative

Suppose at a particular point $x=a$:

$$
f'(a)<0
$$

Since the derivative represents the tangent slope, this means:

> **The tangent line slopes downward from left to right.**

Therefore, locally, the function is decreasing.

Conceptually:

```text
------●----------------
       \
        \
         \
          \
           \
          a

Therefore:

$$ \boxed{ f'(a)<0 \Rightarrow \text{function is locally decreasing} } $$
Example

Again consider:

$$ f(x)=x^2 $$

Its derivative is:

$$ f'(x)=2x $$

At:

$$ x=-2 $$

we get:

$$ f'(-2)=-4 $$

Since:

$$ -4<0 $$

the function is decreasing at $x=-2$.

The tangent line has a negative slope.


## 3.4.5 Zero Derivative

Now consider:

$$
f'(a)=0
$$

Since the derivative represents the tangent slope:

$$
\boxed{
f'(a)=0
\Rightarrow
\text{tangent line is horizontal}
}
$$

A horizontal tangent looks like:

```text
----------------●----------------
                a

This means that, at that point, the function has zero instantaneous rate of change.

For example:

$$ f(x)=x^2 $$

has:

$$ f'(x)=2x $$

At:

$$ x=0 $$

we get:

$$ f'(0)=0 $$

The tangent to the parabola at its lowest point is horizontal.

Important warning

A zero derivative does not automatically mean that the point is a maximum or minimum.

For example:

$$ f(x)=x^3 $$

has:

$$ f'(x)=3x^2 $$

and:

$$ f'(0)=0 $$

But $x=0$ is not a maximum or minimum.

The function continues increasing through the point.

Therefore:

$$ \boxed{ f'(a)=0 \text{ means horizontal tangent} } $$

but it does not by itself prove:

$$ \boxed{ \text{maximum or minimum} } $$

We will study critical points and extrema more carefully later.

````python

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 400)

# Two functions with zero derivative at x = 0
y1 = x**2
y2 = x**3

plt.figure(figsize=(8, 5))

plt.plot(x, y1, label=r"$f(x)=x^2$")
plt.plot(x, y2, label=r"$f(x)=x^3$")
plt.scatter([0], [0], zorder=5, label="x = 0")

plt.axhline(0, linestyle="--", label="Horizontal tangent at x = 0")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Zero Derivative Does Not Always Mean Maximum or Minimum")
plt.grid(True)
plt.legend()

plt.show()
````

## 3.4.6 The Sign of the Derivative

We can summarize the basic relationship:

$$
\boxed{
f'(x)>0
\Rightarrow
\text{locally increasing}
}
$$

$$
\boxed{
f'(x)<0
\Rightarrow
\text{locally decreasing}
}
$$

$$
\boxed{
f'(x)=0
\Rightarrow
\text{horizontal tangent}
}
$$

Think geometrically:

| Derivative | Tangent | Local behavior |
|---|---|---|
| $f'(x)>0$ | Upward slope | Increasing |
| $f'(x)<0$ | Downward slope | Decreasing |
| $f'(x)=0$ | Horizontal | Zero instantaneous rate |

The derivative therefore gives us a compact mathematical way to describe the local direction of a function.

## 3.4.7 The Magnitude of the Derivative

The derivative tells us more than just whether a function is increasing or decreasing.

Its **magnitude** tells us how steeply the function is changing.

Consider:

$$
f'(a)=2
$$

and:

$$
g'(a)=10
$$

Both derivatives are positive.

Therefore, both functions are increasing locally.

But:

$$
|10|>|2|
$$

So the second function has a much steeper local slope.

Similarly:

$$
f'(a)=-2
$$

and:

$$
g'(a)=-10
$$

are both negative, so both functions are decreasing locally.

But the second function is decreasing more steeply.

Therefore:

$$
\boxed{
|f'(a)|
=
\text{magnitude of the local slope}
}
$$

The sign tells us the **direction** of change.

The magnitude tells us the **strength or steepness** of that change.

## 3.4.8 Derivative as Sensitivity

This leads to another very useful interpretation of the derivative:

> **A derivative measures how sensitive the output is to a small change in the input.**

Suppose:

$$
y=f(x)
$$

and we make a small change in $x$:

$$
\Delta x
$$

The corresponding change in the output is approximately:

$$
\boxed{
\Delta y
\approx
f'(x)\Delta x
}
$$

This equation is extremely important.

It says:

> For a sufficiently small change in the input, the derivative tells us approximately how much the output will change.

### Example

Suppose:

$$
f'(x)=5
$$

and we make a small change:

$$
\Delta x=0.1
$$

Then:

$$
\Delta y
\approx
5(0.1)
=
0.5
$$

So the output will change by approximately:

$$
0.5
$$

This is why we can think of the derivative as **local sensitivity**.

A larger absolute derivative means that a small input change produces a larger output change.

A smaller absolute derivative means that the output is less sensitive to small input changes.

## 3.4.9 Why Does $\Delta y\approx f'(x)\Delta x$ Work?

Recall the tangent-line idea from Part 3.

Near a point $x=a$:

$$
f(x)
\approx
f(a)+f'(a)(x-a)
$$

Let:

$$
\Delta x=x-a
$$

Then:

$$
x=a+\Delta x
$$

Substituting:

$$
f(a+\Delta x)
\approx
f(a)+f'(a)\Delta x
$$

Therefore:

$$
f(a+\Delta x)-f(a)
\approx
f'(a)\Delta x
$$

So:

$$
\boxed{
\Delta y
\approx
f'(a)\Delta x
}
$$

This is not saying that the approximation is exact for arbitrary changes.

It becomes especially useful when $\Delta x$ is small.

The reason is that a smooth curve looks increasingly like its tangent line when we zoom in close enough.

## 3.4.10 Sensitivity: A Simple Real-World Analogy

Imagine two volume controls.

For Control A, a tiny movement of the knob produces a huge change in volume.

For Control B, the same tiny movement produces only a small change.

We would say:

> Control A is more sensitive.

The derivative expresses a similar idea mathematically.

If:

$$
|f'(x)|
$$

is large, a small change in $x$ can produce a relatively large change in $f(x)$.

If:

$$
|f'(x)|
$$

is small, the output changes less for the same small input change.

Therefore:

$$
\boxed{
\text{Large }|f'(x)|
\Rightarrow
\text{high local sensitivity}
}
$$

and:

$$
\boxed{
\text{Small }|f'(x)|
\Rightarrow
\text{low local sensitivity}
}
$$

This interpretation becomes particularly valuable in Machine Learning, where we frequently ask:

> **How sensitive is the model's loss or prediction to a small change in an input or parameter?**

## 3.4.11 Sensitivity in Machine Learning

Suppose a model has one parameter:

$$
w
$$

and its loss is:

$$
L(w)
$$

The derivative:

$$
\frac{dL}{dw}
$$

measures how sensitive the loss is to a small change in $w$.

For a small parameter change:

$$
\Delta w
$$

we have approximately:

$$
\boxed{
\Delta L
\approx
\frac{dL}{dw}\Delta w
}
$$

This gives us a powerful interpretation.

### If:

$$
\frac{dL}{dw}=10
$$

then a small increase in $w$ tends to produce a relatively strong increase in the loss.

### If:

$$
\frac{dL}{dw}=0.1
$$

the loss is much less sensitive to the same small change in $w$.

### If:

$$
\frac{dL}{dw}<0
$$

then increasing $w$ locally decreases the loss.

Therefore, the derivative provides both:

1. **Direction** — is the loss increasing or decreasing?
2. **Sensitivity** — how strongly is the loss changing?

This is one of the key reasons derivatives are so useful for Machine Learning optimization.

## 3.4.12 A Simple Machine Learning Example

Suppose the loss function is:

$$
L(w)=(w-3)^2
$$

The derivative is:

$$
\frac{dL}{dw}=2(w-3)
$$

Consider:

$$
w=5
$$

Then:

$$
\frac{dL}{dw}
=
2(5-3)
=
4
$$

So:

$$
\boxed{
\frac{dL}{dw}=4
}
$$

The positive sign tells us:

> Locally, increasing $w$ increases the loss.

The magnitude:

$$
|4|=4
$$

tells us the local slope is relatively steep compared with a point where the derivative is close to zero.

Now consider:

$$
w=3
$$

Then:

$$
\frac{dL}{dw}
=
2(3-3)
=
0
$$

The loss curve has a horizontal tangent there.

In this particular example, $w=3$ is also the minimum of the loss.

But remember:

> A zero derivative alone does not guarantee a minimum.

That conclusion depends on the surrounding behavior of the function.

## 3.4.13 A Very Important Distinction: Derivative at a Point vs Over an Interval

We need to keep two ideas separate.

### Derivative at a point

$$
f'(a)
$$

describes the local rate of change at one specific point.

### Average rate over an interval

$$
\frac{f(b)-f(a)}
{b-a}
$$

describes the overall rate of change between two points.

These are not generally the same.

For example, for:

$$
f(x)=x^2
$$

between:

$$
x=1
$$

and:

$$
x=3
$$

the average rate is:

$$
4
$$

But:

$$
f'(1)=2
$$

and:

$$
f'(3)=6
$$

So the function's instantaneous rate changes throughout the interval.

This is why the derivative gives us information about **local behavior**, while the average rate gives us information about **interval behavior**.

## 3.4.14 Reading a Graph Using the Derivative

Suppose you are given a graph but no equation.

You can still reason about the derivative.

Move from left to right and observe the local slope.

### Graph going upward

The tangent slopes upward:

$$
f'(x)>0
$$

The function is locally increasing.

### Graph going downward

The tangent slopes downward:

$$
f'(x)<0
$$

The function is locally decreasing.

### Graph is momentarily flat

The tangent is horizontal:

$$
f'(x)=0
$$

The instantaneous rate of change is zero.

### Graph is very steep

The magnitude of the derivative is large:

$$
|f'(x)|\text{ is large}
$$

The function is highly sensitive to small changes in $x$.

### Graph is almost flat

The magnitude of the derivative is small:

$$
|f'(x)|\text{ is small}
$$

The function is less sensitive to small changes in $x$.

Therefore, even before calculating anything, the graph itself can give us a qualitative understanding of the derivative.

## 3.4.15 Derivative Sign Changes and Turning Points

The derivative becomes especially interesting when its sign changes.

Suppose:

$$
f'(x)<0
$$

before a point, and:

$$
f'(x)>0
$$

after that point.

The function changes from:

$$
\text{decreasing}
\rightarrow
\text{increasing}
$$

This often indicates a **local minimum**.

Conversely, suppose:

$$
f'(x)>0
$$

before a point, and:

$$
f'(x)<0
$$

after that point.

The function changes from:

$$
\text{increasing}
\rightarrow
\text{decreasing}
$$

This often indicates a **local maximum**.

So the derivative does not merely tell us whether a function is moving up or down.

By examining how its sign changes, we can also identify important features of the function.

We will study maxima, minima, and critical points in greater detail later.
