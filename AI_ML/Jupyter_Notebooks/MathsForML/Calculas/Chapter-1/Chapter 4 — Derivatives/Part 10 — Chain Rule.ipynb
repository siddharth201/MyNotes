# Part 10 — Chain Rule

The **Chain Rule** is one of the most important differentiation rules in calculus and one of the most important mathematical ideas in Machine Learning.

It tells us how to differentiate a function when **one function is inside another function**.

For example:

\[
y=(3x^2+5)^4
\]

Here, \(x\) first goes through one function:

\[
x \rightarrow 3x^2+5
\]

and then the result goes through another function:

\[
3x^2+5 \rightarrow (3x^2+5)^4
\]

So the complete process is:

\[
x
\rightarrow
3x^2+5
\rightarrow
(3x^2+5)^4
\]

The Chain Rule tells us how a change in \(x\) travels through these stages and eventually changes \(y\).

The central idea is:

\[
\boxed{
\text{Overall rate of change}
=
\text{Rate at outer stage}
\times
\text{Rate at inner stage}
}
\]

Mathematically, if

\[
x\rightarrow u\rightarrow y,
\]

then:

\[
\boxed{
\frac{dy}{dx}
=
\frac{dy}{du}
\frac{du}{dx}
}
\]

This simple-looking rule becomes extremely powerful when functions contain many layers.

It is also the mathematical foundation behind **backpropagation in neural networks**.
