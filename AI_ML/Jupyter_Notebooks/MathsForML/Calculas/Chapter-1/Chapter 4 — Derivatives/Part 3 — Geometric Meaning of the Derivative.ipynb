# Chapter 3 — Derivatives

## Part 3 — Geometric Meaning of the Derivative

In Part 2, we built the derivative from first principles:

$$
f'(a)
=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
$$

We understood this algebraically as the limit of the average rate of change.

Now we will understand the same idea **geometrically**.

The central question is:

> **What does a derivative actually look like on a graph?**

The answer connects four ideas:

$$
\boxed{
\text{Secant Line}
\rightarrow
\text{Secant Slope}
\rightarrow
\text{Tangent Line}
\rightarrow
\text{Derivative}
}
$$

By the end of this part, we should be able to look at a graph and intuitively understand what a derivative means.

## 3.1 A Function as a Curve

Consider a function:

$$
y=f(x)
$$

When we plot the function, each value of $x$ gives us a corresponding output:

$$
y=f(x)
$$

Therefore, a point on the graph can be written as:

$$
P=(a,f(a))
$$

Here:

- $a$ is the x-coordinate.
- $f(a)$ is the y-coordinate.
- $P$ is one point on the graph.

Now choose another point:

$$
Q=(b,f(b))
$$

We now have two different points:

$$
P
\quad\text{and}\quad
Q
$$

These two points allow us to calculate a slope.

For a curved function, this slope will describe the **average behavior of the function between those two points**.

## 3.2 Secant Line

A **secant line** is a straight line that passes through two points on a curve.

Suppose:

$$
P=(a,f(a))
$$

and:

$$
Q=(b,f(b))
$$

are two points on the graph of:

$$
y=f(x)
$$

The line passing through $P$ and $Q$ is the secant line.

Its slope is:

$$
\boxed{
m_{\text{secant}}
=
\frac{f(b)-f(a)}
{b-a}
}
$$

This is exactly the average rate of change of the function between $x=a$ and $x=b$.

Therefore:

$$
\boxed{
\text{Secant Slope}
=
\text{Average Rate of Change}
}
$$

The secant line does not tell us the exact behavior at one particular point.

It summarizes the change between two points.

## 3.3 Understanding the Secant Slope Visually

Imagine walking along a curved road.

You choose two locations:

$$
P
$$

and:

$$
Q
$$

The secant line is like drawing a straight line connecting those two locations.

It ignores the detailed shape of the road between them.

Instead, it gives us the overall slope from $P$ to $Q$.

That is why the secant slope represents an **average** rate of change.

Mathematically:

$$
\boxed{
m_{\text{secant}}
=
\frac{\text{Change in }y}
{\text{Change in }x}
}
$$

or:

$$
\boxed{
m_{\text{secant}}
=
\frac{f(b)-f(a)}
{b-a}
}
$$

If the secant slope is positive, the function has increased overall between the two points.

If it is negative, the function has decreased overall.

If it is zero, the two points have the same output value.

But this still does not tell us the exact slope at either point.

````python
import numpy as np
import matplotlib.pyplot as plt

# Function
def f(x):
    return x**2

# Two points
a = 1
b = 3

# Curve
x = np.linspace(0, 4, 400)
y = f(x)

# Coordinates of the two points
ya = f(a)
yb = f(b)

# Secant slope
m_secant = (yb - ya) / (b - a)

# Equation of secant line:
# y = ya + m(x - a)
secant_y = ya + m_secant * (x - a)

plt.figure(figsize=(8, 5))

plt.plot(x, y, label=r"$f(x)=x^2$")
plt.plot(
    x,
    secant_y,
    linestyle="--",
    label="Secant line"
)

plt.scatter(
    [a, b],
    [ya, yb],
    zorder=5,
    label="Points P and Q"
)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Secant Line Through Two Points")
plt.grid(True)
plt.legend()

plt.show()
````

## 3.4 From Two Points to One Point

The secant line requires two points:

$$
P=(a,f(a))
$$

and:

$$
Q=(b,f(b))
$$

But suppose we want to know the slope **exactly at $P$**.

The secant line cannot directly answer this because it uses another point $Q$.

So we make an interesting move:

> Keep $P$ fixed and move $Q$ closer and closer to $P$.

To make this easier to express mathematically, instead of writing the second x-coordinate as $b$, we write it as:

$$
a+h
$$

Therefore:

$$
P=(a,f(a))
$$

and:

$$
Q=(a+h,f(a+h))
$$

The secant slope becomes:

$$
\boxed{
m_{\text{secant}}
=
\frac{f(a+h)-f(a)}
{h}
}
$$

This is the same difference quotient we studied in Part 2.

Now we can give the geometric meaning of $h$:

> **$h$ is the horizontal distance between the two points.**

As $h$ becomes smaller, the two points move closer together.

## 3.5 What Happens as $Q$ Gets Closer to $P$?

Let's imagine that:

$$
P=(a,f(a))
$$

is fixed.

We gradually reduce $h$.

For example:

$$
h=1
$$

Then:

$$
Q=(a+1,f(a+1))
$$

Next:

$$
h=0.5
$$

Then:

$$
Q=(a+0.5,f(a+0.5))
$$

Next:

$$
h=0.1
$$

Then:

$$
Q=(a+0.1,f(a+0.1))
$$

And so on.

The second point keeps getting closer to $P$:

$$
Q\rightarrow P
$$

because:

$$
h\rightarrow0
$$

At every stage, we can calculate a secant slope:

$$
\frac{f(a+h)-f(a)}
{h}
$$

The important question is:

> **What does the secant line approach as $Q$ gets arbitrarily close to $P$?**

## 3.6 The Tangent Line

As the second point $Q$ gets closer and closer to $P$, the secant line approaches a special limiting line.

This line is called the **tangent line**.

The tangent line represents the direction of the curve at that particular point.

Informally, we can think of it as:

> **The straight line that best represents the curve's direction right at that point.**

More precisely, when the derivative exists, the tangent line at $P$ is the line whose slope equals the derivative at that point.

Therefore:

$$
\boxed{
\text{Tangent Slope at }x=a
=
f'(a)
}
$$

This gives us the geometric meaning of the derivative.

The derivative is not merely an algebraic expression.

It is the **slope of the tangent line to the function at that point**.

## 3.7 Secant → Tangent

We can now see the entire process.

Start with two points:

$$
P=(a,f(a))
$$

and:

$$
Q=(a+h,f(a+h))
$$

The line through them is the secant line.

Its slope is:

$$
\frac{f(a+h)-f(a)}
{h}
$$

Now reduce $h$:

$$
h=1
\rightarrow
0.5
\rightarrow
0.1
\rightarrow
0.01
\rightarrow
\cdots
$$

Point $Q$ moves toward $P$.

The secant line changes as $Q$ moves.

Eventually, we ask for the limiting position of this secant line as:

$$
h\rightarrow0
$$

That limiting line is the tangent line.

Therefore:

$$
\boxed{
\text{Secant Line}
\xrightarrow{h\rightarrow0}
\text{Tangent Line}
}
$$

And for the slopes:

$$
\boxed{
\text{Secant Slope}
\xrightarrow{h\rightarrow0}
\text{Tangent Slope}
}
$$

Since the tangent slope is the derivative:

$$
\boxed{
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
=
f'(a)
}
$$

````python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2

a = 2

x = np.linspace(0.5, 3.5, 500)
y = f(x)

h_values = [1.0, 0.5, 0.2, 0.1, 0.05]

for h in h_values:
    x_q = a + h

    y_p = f(a)
    y_q = f(x_q)

    secant_slope = (y_q - y_p) / (x_q - a)

    # Secant line passing through P
    secant_y = y_p + secant_slope * (x - a)

    plt.figure(figsize=(8, 5))

    plt.plot(
        x,
        y,
        label=r"$f(x)=x^2$"
    )

    plt.plot(
        x,
        secant_y,
        linestyle="--",
        label=f"Secant line, h = {h}"
    )

    plt.scatter(
        [a, x_q],
        [y_p, y_q],
        zorder=5
    )

    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.title("Secant Line Approaching the Tangent")
    plt.grid(True)
    plt.legend()

    plt.show()
````

## 3.8 The Tangent Line Is a Local Approximation

There is another important way to understand the tangent line.

A curved function can be complicated.

But if we zoom in sufficiently close to a smooth point on the curve, the curve begins to look almost like a straight line.

The tangent line captures this local straight-line behavior.

For example, consider:

$$
f(x)=x^2
$$

At:

$$
x=2
$$

the tangent line describes the local direction of the parabola around that point.

This leads to a powerful idea:

> **A differentiable function behaves approximately like a straight line when viewed sufficiently close to a point.**

The derivative gives us the slope of that local straight-line approximation.

This idea will become very important later when we study:

- Linear approximation
- Taylor expansion
- Optimization
- Gradient descent
- Local behavior of loss functions

## 3.9 Derivative as Slope

We can now connect the algebraic and geometric definitions.

From first principles:

$$
f'(a)
=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
$$

Geometrically:

$$
f'(a)
=
\text{Slope of the tangent line at }x=a
$$

Therefore:

$$
\boxed{
f'(a)
=
\text{Tangent Slope at }x=a
}
$$

This is one of the most important interpretations of the derivative.

For example, if:

$$
f'(a)=3
$$

then the tangent line at $x=a$ has slope:

$$
3
$$

This means that, locally, an increase of approximately $1$ unit in $x$ corresponds to an increase of approximately $3$ units in $y$.

If:

$$
f'(a)=-2
$$

the tangent line has slope $-2$.

Locally, increasing $x$ causes $y$ to decrease.

If:

$$
f'(a)=0
$$

the tangent line is horizontal.

The function has zero instantaneous rate of change at that point.

## 3.10 Positive, Negative, and Zero Derivative

The sign of a derivative tells us about the local direction of the function.

### Positive derivative

If:

$$
f'(a)>0
$$

the tangent line slopes upward from left to right.

Locally, the function is increasing.

---

### Negative derivative

If:

$$
f'(a)<0
$$

the tangent line slopes downward from left to right.

Locally, the function is decreasing.

---

### Zero derivative

If:

$$
f'(a)=0
$$

the tangent line is horizontal.

The function has no instantaneous increase or decrease at that point.

This often occurs at a local maximum or local minimum, although a zero derivative does not automatically mean that the point is a maximum or minimum.

Therefore:

$$
\boxed{
f'(a)>0
\Rightarrow
\text{locally increasing}
}
$$

$$
\boxed{
f'(a)<0
\Rightarrow
\text{locally decreasing}
}
$$

$$
\boxed{
f'(a)=0
\Rightarrow
\text{horizontal tangent}
}
$$

## 3.11 Example — $f(x)=x^2$

Consider:

$$
f(x)=x^2
$$

From first principles, we found:

$$
f'(x)=2x
$$

Now examine different points.

At:

$$
x=-2
$$

we have:

$$
f'(-2)=2(-2)=-4
$$

So the tangent slope is:

$$
-4
$$

The function is decreasing there.

At:

$$
x=0
$$

we have:

$$
f'(0)=0
$$

So the tangent is horizontal.

At:

$$
x=2
$$

we have:

$$
f'(2)=4
$$

So the tangent slope is:

$$
4
$$

The function is increasing there.

Therefore, the derivative function:

$$
f'(x)=2x
$$

allows us to understand the local direction of the parabola at every point.

````python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2

def derivative(x):
    return 2*x

x = np.linspace(-3, 3, 400)
y = f(x)

points = [-2, 0, 2]

plt.figure(figsize=(8, 5))

plt.plot(
    x,
    y,
    label=r"$f(x)=x^2$"
)

for a in points:
    y_a = f(a)
    slope = derivative(a)

    # Tangent line:
    # y = f(a) + f'(a)(x-a)
    tangent_y = y_a + slope * (x - a)

    plt.plot(
        x,
        tangent_y,
        linestyle="--",
        label=f"Tangent at x = {a}"
    )

    plt.scatter(
        [a],
        [y_a],
        zorder=5
    )

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Tangent Slopes at Different Points")
plt.grid(True)
plt.legend()

plt.show()
````

## 3.12 Derivative Is a Local Property

This is a subtle but extremely important idea.

The derivative at:

$$
x=a
$$

describes the behavior of the function **at and immediately around that point**.

It does not describe the entire function.

For example, for:

$$
f(x)=x^2
$$

we have:

$$
f'(x)=2x
$$

At:

$$
x=1
$$

the derivative is:

$$
f'(1)=2
$$

At:

$$
x=10
$$

the derivative is:

$$
f'(10)=20
$$

The function is the same:

$$
f(x)=x^2
$$

but its local slope is different at different locations.

Therefore:

$$
\boxed{
\text{A derivative describes local behavior}
}
$$

This local nature of derivatives is extremely important in Machine Learning.

During optimization, we often care about:

> **What happens to the loss if we make a small change to the current parameter values?**

That is a local question.

## 3.13 Derivative and Local Linear Approximation

At a differentiable point $x=a$, the tangent line is:

$$
\boxed{
y=f(a)+f'(a)(x-a)
}
$$

This equation says:

- Start at the point $(a,f(a))$.
- Move horizontally by $(x-a)$.
- Change the output approximately according to the slope $f'(a)$.

For points sufficiently close to $a$:

$$
f(x)
\approx
f(a)+f'(a)(x-a)
$$

This is called a **local linear approximation**.

The derivative therefore does more than tell us the slope.

It tells us how the function can be approximated locally by a straight line.

This idea will become very important when we study optimization and why small parameter changes produce predictable changes in a loss function.

## 3.14 A Deeper View — Why the Tangent Line Matters

Why do we care so much about replacing a curve with a straight line locally?

Because straight lines are much easier to understand and work with.

A complicated function may behave in a complicated way globally.

But near a particular point, its behavior can often be approximated by:

$$
\text{Output}
\approx
\text{Current Output}
+
\text{Slope}\times\text{Input Change}
$$

Mathematically:

$$
\boxed{
f(a+\Delta x)
\approx
f(a)+f'(a)\Delta x
}
$$

This equation is one of the most useful intuitions in Calculus.

It says:

> **For a sufficiently small input change, the derivative tells us approximately how much the output will change.**

This is one reason derivatives are so useful for Machine Learning.

If a model parameter changes by a small amount, its derivative tells us how the loss is expected to respond locally.

## 3.15 Connection to Machine Learning

Suppose a model has one parameter:

$$
w
$$

and its loss is:

$$
L(w)
$$

Imagine that we are currently at:

$$
w=w_0
$$

The derivative:

$$
\frac{dL}{dw}\bigg|_{w=w_0}
$$

is the slope of the tangent line to the loss function at the current parameter value.

This gives us local information about the loss landscape.

If:

$$
\frac{dL}{dw}>0
$$

then locally, increasing $w$ increases the loss.

If:

$$
\frac{dL}{dw}<0
$$

then locally, increasing $w$ decreases the loss.

And if:

$$
\frac{dL}{dw}=0
$$

the tangent is horizontal.

The important idea is:

> **The derivative tells the learning algorithm what the loss is doing locally around the current parameter value.**

When there are many parameters, we will need a derivative with respect to each parameter.

That leads to **partial derivatives** and eventually the **gradient**.

So the geometric picture:

$$
\text{Slope of Tangent}
$$

will eventually become the computational idea:

$$
\text{Direction of Parameter Change}
$$

in gradient-based optimization.

## 3.16 Secant vs Tangent — The Essential Difference

It is important not to confuse these two lines.

| | Secant Line | Tangent Line |
|---|---|---|
| Number of points | Two points | One point of contact |
| What it measures | Average rate of change | Instantaneous rate of change |
| Slope | Average slope | Instantaneous slope |
| Formula | $\frac{f(b)-f(a)}{b-a}$ | $f'(a)$ |
| Connection to derivative | Before taking the limit | Result after taking the limit |

The relationship is:

$$
\boxed{
\text{Secant}
=
\text{Average}
}
$$

$$
\boxed{
\text{Tangent}
=
\text{Instantaneous}
}
$$

And:

$$
\boxed{
\text{Secant Slope}
\xrightarrow{Q\rightarrow P}
\text{Tangent Slope}
}
$$

Therefore:

$$
\boxed{
\text{Derivative}
=
\text{Tangent Slope}
}
$$

## 3.17 The Complete Geometric Story

Let's put everything together.

Suppose we have:

$$
y=f(x)
$$

and want the derivative at:

$$
x=a
$$

### Step 1 — Choose the point of interest

$$
P=(a,f(a))
$$

### Step 2 — Choose a nearby point

$$
Q=(a+h,f(a+h))
$$

### Step 3 — Draw the secant line

The slope is:

$$
m_{\text{secant}}
=
\frac{f(a+h)-f(a)}
{h}
$$

This is the average rate of change.

### Step 4 — Move $Q$ closer to $P$

Make:

$$
h\rightarrow0
$$

### Step 5 — The secant line approaches the tangent line

Therefore:

$$
m_{\text{secant}}
\rightarrow
m_{\text{tangent}}
$$

### Step 6 — Define the derivative

$$
\boxed{
f'(a)
=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
}
$$

Therefore:

$$
\boxed{
f'(a)
=
m_{\text{tangent}}
}
$$

This is the geometric meaning of the derivative.

## 3.18 A Crucial Mental Picture

Whenever you see:

$$
f'(a)
$$

imagine this:

```text
                 Curve: y = f(x)
                       /
                      /
                 Q ● /
                   / 
                  / 
             P ●────────────
               /
              /

More accurately, the picture should be understood as:

**1.** $P$ is the point where we want the derivative.  

**2.** $Q$ is another point on the curve.  

**3.** The line through $P$ and $Q$ is a secant.  

**4.** Move $Q$ toward $P$.  

**5.** The secant approaches a tangent.  

**6.** The slope of that tangent is $f'(a)$.  

So whenever you see:

$$ \boxed{f'(a)} $$

think:  

> **"How steep is the curve exactly here?"**

And whenever you see:

$$ \boxed{ \lim_{h\rightarrow0} \frac{f(a+h)-f(a)} {h} } $$

think:  

> **"Take the slope between two nearby points and see what it approaches as those points come together."**


## 3.19 Important Mathematical Precision

There is a subtle point worth keeping precise.

We often say:

> "The tangent line touches the curve at exactly one point."

This is useful intuition, but it is not the mathematical definition of a tangent line.

A line can intersect a curve at more than one point and still be tangent at a particular point.

The more fundamental definition is based on the **limiting slope**:

$$
\boxed{
m_{\text{tangent}}
=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
}
$$

when this limit exists.

Therefore, the safest definition is:

> **The tangent slope at a point is the limiting value of the secant slopes as the second point approaches the point of interest.**

This is precisely why the derivative and tangent slope are connected.
