# Chapter 3 — Derivatives

## Part 2 — Derivative From First Principles

In Part 1, we asked:

> **How can we measure how quickly a function is changing at one particular point?**

We discovered that an ordinary rate of change needs two points.

For a function:

$$
y=f(x)
$$

the average rate of change between two points is:

$$
\frac{f(x_2)-f(x_1)}
{x_2-x_1}
$$

But an instantaneous rate of change concerns one particular point.

The key idea is to take a second point, move it closer and closer to the point of interest, and observe what happens to the average rate of change.

This leads to the derivative.

The conceptual journey is:

$$
\boxed{
\text{Two Points}
\rightarrow
\text{Slope}
\rightarrow
\text{Difference Quotient}
\rightarrow
\text{Smaller Interval}
\rightarrow
\text{Limit}
\rightarrow
\text{Derivative}
}
$$

In this part, we will build the derivative **from first principles**, rather than starting with differentiation rules.

## 3.2 Two-Point Slope

Let's begin with something we already know: the slope of a straight line.

Suppose we have two points:

$$
P=(x_1,y_1)
$$

and:

$$
Q=(x_2,y_2)
$$

The slope of the line joining them is:

$$
\boxed{
m=
\frac{y_2-y_1}
{x_2-x_1}
}
$$

Because:

$$
y_2-y_1=\Delta y
$$

and:

$$
x_2-x_1=\Delta x
$$

we can also write:

$$
\boxed{
m=\frac{\Delta y}{\Delta x}
}
$$

In words:

> **Slope tells us how much the output changes for a given change in the input.**

For example, if:

$$
\Delta x=2
$$

and:

$$
\Delta y=6
$$

then:

$$
m=\frac{6}{2}=3
$$

So for every $1$ unit increase in $x$, the line rises by $3$ units of $y$.

For a straight line, this slope is the same everywhere.

For a curved function, however, the situation is different.

## 3.3 Why a Curve Creates a Problem

Consider a curved function:

$$
y=f(x)
$$

Unlike a straight line, a curve does not generally have the same slope everywhere.

For example:

$$
f(x)=x^2
$$

Near $x=1$, the curve has one steepness.

Near $x=3$, it has a different steepness.

Therefore, asking:

> "What is the slope of the curve?"

is incomplete.

We need to ask:

> **"What is the slope of the curve at a particular point?"**

Suppose our point of interest is:

$$
P=(a,f(a))
$$

There is only one point.

But the ordinary slope formula requires two points.

So we need a clever way to obtain a second point without moving too far away from $P$.

We choose a nearby point:

$$
Q=(a+h,f(a+h))
$$

where $h$ represents a change in the input.

Now we have two points again.

## 3.4 The Difference Quotient

Our two points are:

$$
P=(a,f(a))
$$

and:

$$
Q=(a+h,f(a+h))
$$

The change in input is:

$$
(a+h)-a=h
$$

The change in output is:

$$
f(a+h)-f(a)
$$

Therefore, the slope between these two points is:

$$
\boxed{
m=
\frac{f(a+h)-f(a)}
{h}
}
$$

This expression is called the **difference quotient**.

It is extremely important to understand what it represents.

At this stage:

$$
\frac{f(a+h)-f(a)}
{h}
$$

is still the **average rate of change** between two points.

It is not yet the derivative.

The two points are simply chosen to be close to one another.

Therefore:

$$
\boxed{
\text{Difference Quotient}
=
\text{Average Rate of Change over a nearby interval}
}
$$

The derivative will appear when we make that interval arbitrarily small.

## 3.5 What Exactly Is $h$?

The symbol $h$ represents the change in the input.

We start at:

$$
a
$$

and move to:

$$
a+h
$$

Therefore:

$$
\boxed{
h=(a+h)-a
}
$$

So $h$ is simply:

> **The horizontal distance between the two input values.**

For example, if:

$$
a=2
$$

and:

$$
h=0.1
$$

then the second input is:

$$
a+h=2.1
$$

If:

$$
h=0.01
$$

then:

$$
a+h=2.01
$$

If:

$$
h=-0.1
$$

then:

$$
a+h=1.9
$$

This last example is important.

$h$ does not have to be positive.

- $h>0$ approaches the point from the right.
- $h<0$ approaches the point from the left.

For the ordinary derivative, we consider what happens as:

$$
h\rightarrow0
$$

from both sides.

## 3.6 Making the Interval Smaller

Suppose our point of interest is:

$$
a=2
$$

We can choose different values of $h$.

| $h$ | Second input $a+h$ |
|---:|---:|
| $1$ | $3$ |
| $0.5$ | $2.5$ |
| $0.1$ | $2.1$ |
| $0.01$ | $2.01$ |
| $0.001$ | $2.001$ |
| $-0.1$ | $1.9$ |
| $-0.01$ | $1.99$ |

As $|h|$ becomes smaller, the second point gets closer to the point of interest.

Conceptually:

$$
a+h\rightarrow a
$$

because:

$$
h\rightarrow0
$$

This is the crucial transition.

We are moving from:

> "What is the average rate of change between two nearby points?"

toward:

> **"What value does that average rate of change approach as the two points become arbitrarily close?"**

The second question is a **limit** question.

And this is exactly why limits are needed to define derivatives.

```python
a = 2

h_values = [1, 0.5, 0.1, 0.01, 0.001, -0.1, -0.01]

for h in h_values:
    print(f"h = {h:>7}   a + h = {a + h}")
```

## 3.7 Example — $f(x)=x^2$

Let's make the idea concrete.

Consider:

$$
f(x)=x^2
$$

We want to understand the rate of change at:

$$
x=a
$$

Choose the two points:

$$
a
$$

and:

$$
a+h
$$

The difference quotient is:

$$
\frac{f(a+h)-f(a)}
{h}
$$

Since:

$$
f(x)=x^2
$$

we have:

$$
f(a+h)=(a+h)^2
$$

and:

$$
f(a)=a^2
$$

Therefore:

$$
\frac{f(a+h)-f(a)}
{h}
=
\frac{(a+h)^2-a^2}
{h}
$$

Expand:

$$
(a+h)^2=a^2+2ah+h^2
$$

Therefore:

$$
\frac{a^2+2ah+h^2-a^2}
{h}
$$

Cancel $a^2$:

$$
\frac{2ah+h^2}
{h}
$$

Factor $h$:

$$
\frac{h(2a+h)}
{h}
$$

For $h\neq0$:

$$
2a+h
$$

So the difference quotient simplifies to:

$$
\boxed{
2a+h
}
$$

Now we ask what happens as:

$$
h\rightarrow0
$$

Then:

$$
2a+h\rightarrow2a
$$

Therefore:

$$
\boxed{
f'(a)=2a
}
$$

This is the derivative of $f(x)=x^2$ at the point $x=a$.

## 3.8 The Limit Definition of the Derivative

The previous example reveals the general pattern.

For a function:

$$
f(x)
$$

at a particular point:

$$
x=a
$$

the average rate of change between:

$$
a
$$

and:

$$
a+h
$$

is:

$$
\frac{f(a+h)-f(a)}
{h}
$$

To obtain the instantaneous rate of change, we make the interval arbitrarily small:

$$
h\rightarrow0
$$

Therefore:

$$
\boxed{
f'(a)
=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
}
$$

This is the **derivative from first principles**.

It is also called the **limit definition of the derivative**.

In plain English:

> **The derivative at $x=a$ is the value that the average rate of change approaches as the second point gets arbitrarily close to the first point.**

This is the fundamental definition of a derivative.

Everything else we learn about differentiation is built on this idea.

## 3.9 Derivative at a Point vs Derivative Function

There is an important notation distinction.

When we write:

$$
f'(a)
$$

we are asking for the derivative at one specific point:

$$
x=a
$$

The definition is:

$$
\boxed{
f'(a)
=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
}
$$

But we can also let the point be an arbitrary $x$.

Then:

$$
\boxed{
f'(x)
=
\lim_{h\rightarrow0}
\frac{f(x+h)-f(x)}
{h}
}
$$

This gives us the **derivative function**.

For example, if:

$$
f(x)=x^2
$$

then:

$$
f'(x)=2x
$$

This derivative function tells us the slope at every point.

For example:

$$
f'(1)=2
$$

$$
f'(2)=4
$$

$$
f'(3)=6
$$

So:

$$
\boxed{
f'(x)=\text{derivative function}
}
$$

while:

$$
\boxed{
f'(a)=\text{derivative at the specific point }x=a
}
$$

Keeping this distinction clear will become important when we later move from derivatives to partial derivatives and gradients.

## 3.10 Why $h\rightarrow0$ and Not $h=0$?

This is one of the most important conceptual points.

The derivative is:

$$
f'(a)
=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
$$

A common mistake is to think:

> "Since $h$ becomes zero, just substitute $h=0$."

But that does not work.

If we substitute:

$$
h=0
$$

we get:

$$
\frac{f(a)-f(a)}
{0}
$$

which becomes:

$$
\frac{0}{0}
$$

This expression is undefined.

So the derivative does **not** come from directly evaluating the difference quotient at $h=0$.

Instead, we study its behavior for values of $h$ that are:

$$
\text{very close to }0
$$

but not equal to zero.

For example:

$$
h=0.1,\ 0.01,\ 0.001,\ 0.0001
$$

and from the other side:

$$
h=-0.1,\ -0.01,\ -0.001,\ -0.0001
$$

Then we ask:

> **What value does the difference quotient approach?**

That value is the derivative.

Therefore:

$$
\boxed{
h\rightarrow0
\text{ means "approaches zero"}
}
$$

not:

$$
\boxed{
h=0
}
$$

This distinction is fundamental to Calculus.

## 3.11 Why Can We Cancel $h$ If $h$ Is Approaching Zero?

This can initially feel confusing.

For:

$$
f(x)=x^2
$$

we obtained:

$$
\frac{h(2a+h)}
{h}
$$

and cancelled $h$:

$$
2a+h
$$

But didn't we say:

$$
h\rightarrow0
$$

and that $h$ cannot actually be zero?

Yes.

That is precisely why the cancellation is valid.

During the algebraic simplification, we are considering values where:

$$
h\neq0
$$

The limit then asks what happens as those nonzero values get arbitrarily close to zero.

So:

$$
\frac{h(2a+h)}{h}
=
2a+h
\qquad\text{for }h\neq0
$$

Then:

$$
\lim_{h\rightarrow0}(2a+h)=2a
$$

We never needed to evaluate the original difference quotient at $h=0$.

This is a very useful pattern:

> **Simplify the expression for values where it is defined, and then take the limit.**

## 3.12 Numerical Example — $f(x)=x^2$ at $x=2$

Let's calculate the difference quotient numerically.

For:

$$
f(x)=x^2
$$

and:

$$
a=2
$$

we have:

$$
\frac{f(2+h)-f(2)}
{h}
$$

For positive $h$:

$$
h=1,\ 0.1,\ 0.01,\ 0.001,\ldots
$$

the values approach:

$$
4
$$

We can also approach from the left using negative $h$:

$$
h=-0.1,\ -0.01,\ -0.001,\ldots
$$

These values also approach:

$$
4
$$

This is important because the ordinary derivative uses a two-sided limit.

We want the behavior to approach the same value from both directions.

For this function:

$$
\boxed{
f'(2)=4
}
$$

```python
def f(x):
    return x**2

a = 2

h_values = [
    1,
    0.1,
    0.01,
    0.001,
    0.0001,
    -0.1,
    -0.01,
    -0.001,
    -0.0001
]

print("Difference quotient for f(x) = x² at x = 2\n")

for h in h_values:
    difference_quotient = (f(a + h) - f(a)) / h
    print(
        f"h = {h:>8}   "
        f"difference quotient = {difference_quotient}"
    )
```

## 3.13 Approaching From Both Sides

Because:

$$
h
$$

can be positive or negative, we can approach the point from both directions.

### From the right

If:

$$
h>0
$$

then:

$$
a+h>a
$$

So the second point lies to the right of $a$.

This corresponds to:

$$
h\rightarrow0^+
$$

### From the left

If:

$$
h<0
$$

then:

$$
a+h<a
$$

So the second point lies to the left of $a$.

This corresponds to:

$$
h\rightarrow0^-
$$

For the ordinary derivative to exist, the limiting behavior must agree from both sides.

Conceptually:

$$
\boxed{
\lim_{h\rightarrow0^-}
\frac{f(a+h)-f(a)}{h}
=
\lim_{h\rightarrow0^+}
\frac{f(a+h)-f(a)}{h}
}
$$

when the derivative exists.

This connects directly to the one-sided limits we studied in the previous chapter.

## 3.14 Difference Quotient and the Secant Line

The difference quotient:

$$
\frac{f(a+h)-f(a)}
{h}
$$

has a geometric meaning.

Consider the two points:

$$
P=(a,f(a))
$$

and:

$$
Q=(a+h,f(a+h))
$$

The straight line joining $P$ and $Q$ is a **secant line**.

Its slope is:

$$
\boxed{
m_{\text{secant}}
=
\frac{f(a+h)-f(a)}
{h}
}
$$

Therefore:

$$
\boxed{
\text{Difference Quotient}
=
\text{Slope of the Secant Line}
}
$$

As:

$$
h\rightarrow0
$$

point $Q$ moves toward point $P$.

The secant slope approaches a limiting value.

That limiting value is the slope of the tangent line.

Therefore:

$$
\boxed{
\text{Derivative}
=
\text{Limit of Secant Slopes}
}
$$

We will study this geometric interpretation in much greater detail in **Part 3 — Geometric Meaning of the Derivative**.

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2

a = 2

x = np.linspace(0.5, 3.5, 400)
y = f(x)

h_values = [1.0, 0.5, 0.1, 0.01]

for h in h_values:
    x2 = a + h

    y1 = f(a)
    y2 = f(x2)

    secant_slope = (y2 - y1) / (x2 - a)

    # Secant line passing through (a, f(a))
    secant_y = y1 + secant_slope * (x - a)

    plt.figure(figsize=(8, 5))

    plt.plot(x, y, label=r"$f(x)=x^2$")
    plt.plot(
        x,
        secant_y,
        label=f"Secant line, h = {h}"
    )

    plt.scatter(
        [a, x2],
        [y1, y2],
        zorder=5
    )

    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.title("Secant Line as h Becomes Smaller")
    plt.grid(True)
    plt.legend()

    plt.show()
```

## 3.15 First Principles — Complete Derivation of $f(x)=x^2$

Let's summarize the derivation without skipping any conceptual step.

Given:

$$
f(x)=x^2
$$

### Step 1 — Start with the definition

$$
f'(x)
=
\lim_{h\rightarrow0}
\frac{f(x+h)-f(x)}
{h}
$$

### Step 2 — Substitute the function

$$
f'(x)
=
\lim_{h\rightarrow0}
\frac{(x+h)^2-x^2}
{h}
$$

### Step 3 — Expand

$$
(x+h)^2=x^2+2xh+h^2
$$

Therefore:

$$
f'(x)
=
\lim_{h\rightarrow0}
\frac{x^2+2xh+h^2-x^2}
{h}
$$

### Step 4 — Cancel common terms

$$
f'(x)
=
\lim_{h\rightarrow0}
\frac{2xh+h^2}
{h}
$$

### Step 5 — Factor

$$
f'(x)
=
\lim_{h\rightarrow0}
\frac{h(2x+h)}
{h}
$$

### Step 6 — Cancel $h$ for $h\neq0$

$$
f'(x)
=
\lim_{h\rightarrow0}(2x+h)
$$

### Step 7 — Take the limit

As:

$$
h\rightarrow0
$$

we obtain:

$$
2x+h\rightarrow2x
$$

Therefore:

$$
\boxed{
f'(x)=2x
}
$$

This is the derivative of:

$$
f(x)=x^2
$$

We did not memorize the result.

We **derived it from the fundamental definition**.

## 3.16 First Principles Example — Linear Function

Consider:

$$
f(x)=mx+b
$$

Using the definition:

$$
f'(x)
=
\lim_{h\rightarrow0}
\frac{f(x+h)-f(x)}
{h}
$$

First:

$$
f(x+h)=m(x+h)+b
$$

Therefore:

$$
f'(x)
=
\lim_{h\rightarrow0}
\frac{m(x+h)+b-(mx+b)}
{h}
$$

Expand:

$$
=
\lim_{h\rightarrow0}
\frac{mx+mh+b-mx-b}
{h}
$$

Cancel:

$$
mx-mx=0
$$

and:

$$
b-b=0
$$

giving:

$$
f'(x)
=
\lim_{h\rightarrow0}
\frac{mh}{h}
$$

For $h\neq0$:

$$
f'(x)
=
\lim_{h\rightarrow0}m
$$

Since $m$ is constant:

$$
\boxed{
f'(x)=m
}
$$

This makes intuitive sense.

A straight line has the same slope everywhere.

Therefore:

$$
\boxed{
\text{Derivative of a linear function}
=
\text{its constant slope}
}
$$

## 3.17 First Principles Example — Constant Function

Consider:

$$
f(x)=c
$$

where $c$ is a constant.

Using the definition:

$$
f'(x)
=
\lim_{h\rightarrow0}
\frac{f(x+h)-f(x)}
{h}
$$

Since the function always equals $c$:

$$
f(x+h)=c
$$

and:

$$
f(x)=c
$$

Therefore:

$$
f'(x)
=
\lim_{h\rightarrow0}
\frac{c-c}{h}
$$

So:

$$
f'(x)
=
\lim_{h\rightarrow0}0
$$

Therefore:

$$
\boxed{
f'(x)=0
}
$$

This also makes intuitive sense.

A constant function never changes as $x$ changes.

Therefore its rate of change is zero.

## 3.18 Why Is This Called "First Principles"?

The phrase **first principles** means that we start from the most fundamental definition rather than using previously established differentiation rules.

For derivatives, the fundamental starting point is:

$$
\boxed{
f'(x)
=
\lim_{h\rightarrow0}
\frac{f(x+h)-f(x)}
{h}
}
$$

For example, instead of simply memorizing:

$$
\frac{d}{dx}(x^2)=2x
$$

we can derive it from the definition.

This matters because it answers a deeper question:

> **Where did the differentiation rules come from?**

Rules such as the:

- Power Rule
- Product Rule
- Quotient Rule
- Chain Rule

are not unrelated tricks.

They are mathematical consequences of the underlying definition of differentiation.

We will eventually use these rules because repeatedly applying first principles would be inefficient.

But first principles gives us the foundation from which those rules make sense.

## 3.19 What the Derivative Formula Is Really Saying

Consider the definition:

$$
\boxed{
f'(x)
=
\lim_{h\rightarrow0}
\frac{f(x+h)-f(x)}
{h}
}
$$

Read it from the inside out.

### $x+h$

Move slightly away from $x$.

### $f(x+h)-f(x)$

Measure how much the output changes.

### Divide by $h$

Measure output change per unit input change.

Therefore:

$$
\frac{f(x+h)-f(x)}{h}
$$

is an average rate of change over a small interval.

### Take the limit

Let the interval become arbitrarily small:

$$
h\rightarrow0
$$

This extracts the local behavior.

Therefore:

$$
\boxed{
f'(x)
=
\text{instantaneous rate of change at }x
}
$$

The formula is therefore not an arbitrary mathematical trick.

Every part of it exists because of the question we are trying to answer.

## 3.20 A Very Important Distinction — Small Change vs Infinitely Small Change

It is common to hear:

> "A derivative is the rate of change when the change becomes infinitely small."

This is useful intuition, but we should phrase it carefully.

We do not actually substitute an "infinitely small number" for $h$.

Instead, we use a limit:

$$
h\rightarrow0
$$

We consider ordinary nonzero values of $h$ that can be made as close to zero as desired, and examine what the difference quotient approaches.

So the rigorous idea is:

$$
\boxed{
\text{Derivative}
=
\text{Limit of the difference quotient as }h\rightarrow0
}
$$

This is more precise than saying that we simply "divide by an infinitely small number."

The limit is the mathematical mechanism that makes the idea precise.

## 3.21 Concept Map — From Two Points to Derivative

The entire construction can now be viewed as one logical chain.

### 1. Start with two points

$$
P=(a,f(a))
$$

$$
Q=(a+h,f(a+h))
$$

### 2. Calculate the slope

$$
\frac{f(a+h)-f(a)}
{h}
$$

This is the average rate of change.

### 3. Bring $Q$ closer to $P$

Make:

$$
h
$$

smaller.

### 4. Ask what the slope approaches

$$
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
$$

### 5. Define the derivative

$$
\boxed{
f'(a)
=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
}
$$

Therefore:

$$
\boxed{
\text{Two-point slope}
\rightarrow
\text{Difference Quotient}
\rightarrow
\text{Smaller Interval}
\rightarrow
\text{Limit}
\rightarrow
\text{Derivative}
}
$$

## 3.22 Connection to Machine Learning

The same mathematical idea appears in Machine Learning.

Suppose:

$$
L(w)
$$

is a loss function depending on a model parameter $w$.

We want to know:

> **How does the loss change when we slightly change $w$?**

The average change over a finite parameter change $\Delta w$ is:

$$
\frac{
L(w+\Delta w)-L(w)
}
{\Delta w}
$$

This tells us the average change in loss per unit change in $w$.

To understand the local behavior at the current value of $w$, we consider:

$$
\Delta w\rightarrow0
$$

Therefore:

$$
\boxed{
\frac{dL}{dw}
=
\lim_{\Delta w\rightarrow0}
\frac{
L(w+\Delta w)-L(w)
}
{\Delta w}
}
$$

This is exactly the same first-principles idea.

The only difference is that our function is now a **loss function**, and the input is a model parameter.

This derivative tells us the local sensitivity of the loss to the parameter.

Later, with many parameters, this idea becomes partial derivatives and then the gradient.

We let:

$$
h\rightarrow0
$$

The second point approaches the first point.

---

### Limit definition

The derivative is:

$$
\boxed{
f'(a)
=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
}
$$

---

### Why not $h=0$?

Because direct substitution produces:

$$
\frac00
$$

which is undefined.

We do not set $h=0$.

We study what the expression approaches as:

$$
h\rightarrow0
$$

---

### First-principles example

For:

$$
f(x)=x^2
$$

we obtain:

$$
\boxed{
f'(x)=2x
}
$$

---

### Core idea

$$
\boxed{
\text{Derivative}
=
\text{Limit of the Average Rate of Change}
}
$$

Geometrically:

$$
\boxed{
\text{Derivative}
=
\text{Limit of Secant Slopes}
}
$$

# 🧠 Mental Model — First Principles

Imagine standing at one point on a curved road.

You want to know:

> **How steep is the road exactly where I am standing?**

You cannot calculate a slope using only one point.

So you look at another point nearby.

You calculate the slope between the two points.

That gives an **average slope**.

Then you move the second point closer.

Closer.

Closer.

And ask:

> **What value does the slope approach?**

That limiting slope is the derivative.

Mathematically:

$$
\boxed{
\text{Two points}
\rightarrow
\text{Average slope}
\rightarrow
\text{Bring points closer}
\rightarrow
\text{Limit}
\rightarrow
\text{Instantaneous slope}
}
$$

The key idea is not:

> "Memorize the derivative formula."

The key idea is:

> **A derivative is created by taking the limit of the slope between two increasingly close points.**

# 🎯 Part 2 — Final Takeaways

The derivative was not introduced as a formula to memorize.

We **built it logically**.

Start with two points:

$$
(a,f(a))
$$

and:

$$
(a+h,f(a+h))
$$

Calculate their slope:

$$
\frac{f(a+h)-f(a)}
{h}
$$

This is the average rate of change over a small interval.

Now make the interval arbitrarily small:

$$
h\rightarrow0
$$

and ask what the slope approaches.

That gives:

$$
\boxed{
f'(a)
=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}
{h}
}
$$

Therefore:

$$
\boxed{
\text{Derivative}
=
\text{Instantaneous Rate of Change}
}
$$

and geometrically:

$$
\boxed{
\text{Derivative}
=
\text{Tangent Slope}
}
$$

The critical distinction is:

$$
\boxed{
h\rightarrow0
\neq
h=0
}
$$

We do not divide by zero.

We examine the behavior of the difference quotient as $h$ gets arbitrarily close to zero.

For Machine Learning, the same idea tells us how a loss changes locally when a model parameter changes:

$$
\boxed{
\frac{dL}{dw}
=
\lim_{\Delta w\rightarrow0}
\frac{
L(w+\Delta w)-L(w)
}
{\Delta w}
}
$$

This is the mathematical foundation on which the later concepts of **partial derivatives, gradients, gradient descent, and backpropagation** will be built.
