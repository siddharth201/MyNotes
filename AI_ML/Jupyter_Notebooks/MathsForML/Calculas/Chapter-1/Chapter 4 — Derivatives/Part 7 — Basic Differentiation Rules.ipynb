# Chapter 3 — Derivatives

## Part 7 — Basic Differentiation Rules

In the previous parts, we learned how a derivative is defined:

$$
f'(x)
=
\lim_{h\to0}
\frac{f(x+h)-f(x)}{h}
$$

This definition is fundamental.

However, imagine that we had to use this limit every time we wanted to differentiate a function.

For example, for:

$$
f(x)=3x^5-7x^3+4x^2-9x+12
$$

we could theoretically use the definition of the derivative, but doing that repeatedly would be slow and cumbersome.

Fortunately, mathematics gives us a collection of **differentiation rules**.

These rules allow us to find derivatives quickly while still being mathematically grounded in the fundamental definition.

In this part, we will learn five basic rules:

1. **Constant Rule**
2. **Power Rule**
3. **Constant Multiple Rule**
4. **Sum Rule**
5. **Difference Rule**

These rules form the foundation for differentiating polynomial functions and will be used repeatedly in Machine Learning.

# 7.1 The Big Picture

Before learning individual rules, remember what differentiation is doing.

Suppose:

$$
y=f(x)
$$

The derivative asks:

> **How does $y$ change when $x$ changes?**

The differentiation rules simply give us efficient ways to calculate that rate of change.

Think of them as a **toolbox**:

$$
\boxed{
\text{Function}
\longrightarrow
\text{Apply appropriate rules}
\longrightarrow
\text{Derivative}
}
$$

For example:

$$
f(x)=x^3
$$

Use the power rule:

$$
f'(x)=3x^2
$$

For:

$$
f(x)=5x^3
$$

we combine the constant multiple rule and power rule:

$$
f'(x)=5(3x^2)=15x^2
$$

For:

$$
f(x)=5x^3+2x^2
$$

we additionally use the sum rule:

$$
f'(x)=15x^2+4x
$$

So the rules can be **combined**.

# 7.2 Constant Rule

A **constant** is a quantity whose value does not change.

Examples:

$$
5,\quad 10,\quad -3,\quad 100,\quad \pi
$$

Consider the function:

$$
f(x)=5
$$

No matter what value of $x$ we choose:

$$
f(x)=5
$$

The output never changes.

Therefore, its rate of change is:

$$
\boxed{f'(x)=0}
$$

### Constant Rule

If:

$$
f(x)=c
$$

where $c$ is a constant, then:

$$
\boxed{
\frac{d}{dx}(c)=0
}
$$

## Why Is the Derivative of a Constant Zero?

Let's use the idea we already know.

The derivative is:

$$
f'(x)
=
\lim_{h\to0}
\frac{f(x+h)-f(x)}{h}
$$

For:

$$
f(x)=c
$$

we have:

$$
f(x+h)=c
$$

Therefore:

$$
f'(x)
=
\lim_{h\to0}
\frac{c-c}{h}
$$

So:

$$
f'(x)
=
\lim_{h\to0}
\frac{0}{h}
$$

which gives:

$$
\boxed{f'(x)=0}
$$

### Intuition

A horizontal line has no rise.

Its slope is:

$$
\frac{\text{rise}}{\text{run}}
=
\frac{0}{\text{run}}
=
0
$$

Therefore:

> **A constant function has zero rate of change.**

## Examples — Constant Rule

### Example 1

$$
f(x)=7
$$

Therefore:

$$
\boxed{f'(x)=0}
$$

---

### Example 2

$$
f(x)=-15
$$

Therefore:

$$
\boxed{f'(x)=0}
$$

---

### Example 3

$$
f(x)=\pi
$$

Therefore:

$$
\boxed{f'(x)=0}
$$

---

### Example 4

$$
f(x)=100000
$$

Therefore:

$$
\boxed{f'(x)=0}
$$

The size of the constant does not matter.

If the function does not change as $x$ changes, its derivative is zero.

# 7.3 Power Rule

The **power rule** is one of the most important differentiation rules.

Suppose:

$$
f(x)=x^n
$$

where $n$ is a constant.

Then:

$$
\boxed{
\frac{d}{dx}(x^n)=nx^{n-1}
}
$$

In words:

> **Bring the exponent down in front and reduce the exponent by 1.**

For example:

$$
\frac{d}{dx}(x^5)
=
5x^4
$$

because:

- The exponent $5$ comes down.
- The exponent becomes $5-1=4$.

Therefore:

$$
x^5
\rightarrow
5x^4
$$

## Why Does the Power Rule Work?

We previously derived the derivative of:

$$
x^2
$$

from first principles.

The power rule generalizes that idea.

For a positive integer $n$:

$$
f(x)=x^n
$$

Using the derivative definition:

$$
f'(x)
=
\lim_{h\to0}
\frac{(x+h)^n-x^n}{h}
$$

Using the binomial theorem:

$$
(x+h)^n
=
x^n
+
nx^{n-1}h
+
\text{terms containing }h^2,h^3,\ldots
$$

Therefore:

$$
(x+h)^n-x^n
=
nx^{n-1}h
+
\text{terms containing }h^2,h^3,\ldots
$$

Divide by $h$:

$$
\frac{(x+h)^n-x^n}{h}
=
nx^{n-1}
+
\text{terms containing }h,h^2,\ldots
$$

As:

$$
h\to0
$$

all the remaining terms containing positive powers of $h$ approach zero.

Therefore:

$$
\boxed{
\frac{d}{dx}(x^n)=nx^{n-1}
}
$$

So the power rule is not an arbitrary trick.

It comes from the fundamental definition of the derivative.

# 7.4 Understanding the Power Rule Intuitively

Consider:

$$
f(x)=x^4
$$

The power rule gives:

$$
f'(x)=4x^3
$$

What happened?

$$
x^4
\rightarrow
4x^3
$$

The exponent:

$$
4
$$

becomes the coefficient.

Then:

$$
4-1=3
$$

becomes the new exponent.

So:

$$
\boxed{
x^4
\rightarrow
4x^3
}
$$

Similarly:

$$
x^7
\rightarrow
7x^6
$$

and:

$$
x^{10}
\rightarrow
10x^9
$$

### Important

The power rule is:

$$
\boxed{
x^n\rightarrow nx^{n-1}
}
$$

not:

$$
x^n\rightarrow x^{n-1}
$$

The original exponent must become the coefficient.

$$

$$
\frac{d}{dx}(x^0)

$$
=0

$$

$$

$$
\boxed{

}
$$

This connects the power rule naturally with the constant rule.

# 7.7 Constant Multiple Rule

Now consider:

$$
f(x)=5x^3
$$

There are two parts:

- $5$ is a constant.
- $x^3$ depends on $x$.

The derivative is:

$$
f'(x)=15x^2
$$

The rule is:

$$
\boxed{
\frac{d}{dx}[c f(x)]
=
c\frac{d}{dx}[f(x)]
}
$$

where $c$ is a constant.

In words:

> **A constant multiplier can be kept outside the derivative.**

For example:

$$
\frac{d}{dx}(7x^4)
=
7\frac{d}{dx}(x^4)
$$

Then:

$$
=7(4x^3)
$$

Therefore:

$$
\boxed{
\frac{d}{dx}(7x^4)=28x^3
}
$$

## Why Does the Constant Stay?

Start from the derivative definition.

Suppose:

$$
g(x)=cf(x)
$$

Then:

$$
g'(x)
=
\lim_{h\to0}
\frac{cf(x+h)-cf(x)}{h}
$$

Factor out $c$:

$$
=
\lim_{h\to0}
c
\frac{f(x+h)-f(x)}{h}
$$

Since $c$ does not depend on $h$, it can be taken outside the limit:

$$
=
c
\lim_{h\to0}
\frac{f(x+h)-f(x)}{h}
$$

Therefore:

$$
\boxed{
g'(x)=cf'(x)
}
$$

### Intuition

If a function's output is multiplied by $5$, its rate of change is also multiplied by $5$.

For example:

$$
f(x)=x^2
$$

has derivative:

$$
f'(x)=2x
$$

Now multiply the function by $5$:

$$
g(x)=5x^2
$$

Its derivative becomes:

$$
g'(x)=10x
$$

The entire function has been vertically scaled by $5$, so its slope is also scaled by $5$.

## Examples — Constant Multiple Rule

### Example 1

$$
f(x)=3x^5
$$

Keep $3$ and differentiate $x^5$:

$$
f'(x)
=
3(5x^4)
$$

Therefore:

$$
\boxed{f'(x)=15x^4}
$$

---

### Example 2

$$
f(x)=-4x^3
$$

$$
f'(x)
=
-4(3x^2)
$$

Therefore:

$$
\boxed{f'(x)=-12x^2}
$$

---

### Example 3

$$
f(x)=\frac{1}{2}x^6
$$

$$
f'(x)
=
\frac12(6x^5)
$$

Therefore:

$$
\boxed{f'(x)=3x^5}
$$

# 7.8 Sum Rule

Now suppose a function contains multiple terms that are **added**.

For example:

$$
f(x)=x^3+x^2
$$

We can differentiate each term separately:

$$
f'(x)
=
3x^2+2x
$$

This is the **sum rule**.

If:

$$
f(x)=g(x)+h(x)
$$

then:

$$
\boxed{
f'(x)=g'(x)+h'(x)
}
$$

Or:

$$
\boxed{
\frac{d}{dx}[g(x)+h(x)]
=
\frac{dg}{dx}
+
\frac{dh}{dx}
}
$$

In words:

> **The derivative of a sum is the sum of the derivatives.**

## Why Does the Sum Rule Work?

Start from the derivative definition.

Suppose:

$$
f(x)=g(x)+h(x)
$$

Then:

$$
f'(x)
=
\lim_{t\to0}
\frac{
[g(x+t)+h(x+t)]
-
[g(x)+h(x)]
}{t}
$$

Rearrange:

$$
=
\lim_{t\to0}
\left[
\frac{g(x+t)-g(x)}{t}
+
\frac{h(x+t)-h(x)}{t}
\right]
$$

Using the linearity of limits:

$$
=
\lim_{t\to0}
\frac{g(x+t)-g(x)}{t}
+
\lim_{t\to0}
\frac{h(x+t)-h(x)}{t}
$$

Therefore:

$$
\boxed{
f'(x)=g'(x)+h'(x)
}
$$

So we can differentiate the terms independently.

## Examples — Sum Rule

### Example 1

$$
f(x)=x^3+x^2
$$

Differentiate each term:

$$
f'(x)=3x^2+2x
$$

---

### Example 2

$$
f(x)=x^5+4x^2
$$

$$
f'(x)=5x^4+8x
$$

---

### Example 3

$$
f(x)=3x^4+7x^3+x
$$

Differentiate term-by-term:

$$
f'(x)
=
12x^3+21x^2+1
$$

The derivative of each term is calculated independently.

# 7.9 Difference Rule

The same idea works when terms are **subtracted**.

Suppose:

$$
f(x)=g(x)-h(x)
$$

Then:

$$
\boxed{
f'(x)=g'(x)-h'(x)
}
$$

In words:

> **The derivative of a difference is the difference of the derivatives.**

For example:

$$
f(x)=x^4-x^2
$$

Differentiate each term:

$$
f'(x)
=
4x^3-2x
$$

The subtraction sign remains unchanged.

## Why Does the Difference Rule Work?

Subtraction can be viewed as adding a negative:

$$
g(x)-h(x)
=
g(x)+[-h(x)]
$$

The sum rule tells us:

$$
\frac{d}{dx}[g(x)-h(x)]
=
g'(x)+\frac{d}{dx}[-h(x)]
$$

Using the constant multiple rule with the constant $-1$:

$$
\frac{d}{dx}[-h(x)]
=
-h'(x)
$$

Therefore:

$$
\boxed{
\frac{d}{dx}[g(x)-h(x)]
=
g'(x)-h'(x)
}
$$

So the difference rule follows naturally from the sum rule and constant multiple rule.

## Examples — Difference Rule

### Example 1

$$
f(x)=x^5-x^2
$$

Therefore:

$$
f'(x)=5x^4-2x
$$

---

### Example 2

$$
f(x)=7x^4-3x^3
$$

Therefore:

$$
f'(x)=28x^3-9x^2
$$

---

### Example 3

$$
f(x)=10x^6-5x^2-x
$$

Therefore:

$$
f'(x)
=
60x^5-10x-1
$$

# 7.10 Combining the Basic Rules

The real power of these rules appears when we combine them.

Consider:

$$
f(x)=4x^5-3x^3+7x^2-2x+9
$$

We can identify the structure:

$$
4x^5
-
3x^3
+
7x^2
-
2x
+
9
$$

We use:

- Constant multiple rule
- Power rule
- Sum rule
- Difference rule
- Constant rule

Differentiate term-by-term.

### First term

$$
\frac{d}{dx}(4x^5)
=
4(5x^4)
=
20x^4
$$

### Second term

$$
\frac{d}{dx}(-3x^3)
=

-9x^2
$$

### Third term

$$
\frac{d}{dx}(7x^2)

# 7.11 A Reliable Differentiation Workflow

When differentiating a polynomial, use this process:

### Step 1 — Break the function into terms

For example:

$$
f(x)=5x^4-3x^2+7x-10
$$

The terms are:

$$
5x^4,\quad -3x^2,\quad 7x,\quad -10
$$

### Step 2 — Differentiate each term independently

Use the appropriate rule for each term.

### Step 3 — Keep constants attached to their terms

For example:

$$
5x^4
\rightarrow
5(4x^3)
$$

### Step 4 — Apply the power rule

$$
5(4x^3)=20x^3
$$

### Step 5 — Differentiate constants

$$
-10\rightarrow0
$$

### Step 6 — Combine the results

The final answer is:

$$
\boxed{
f'(x)=20x^3-6x+7
}
$$

### Mental algorithm

$$
\boxed{
\text{Break apart}
\rightarrow
\text{Differentiate each term}
\rightarrow
\text{Combine}
}
$$

Wrong:

$$
\frac{d}{dx}(x^5)=x^4
$$

Correct:

$$
\boxed{
\frac{d}{dx}(x^5)=5x^4
}
$$

---

## Mistake 2 — Forgetting to reduce the exponent

Wrong:

$$
\frac{d}{dx}(x^5)=5x^5
$$

Correct:

$$
\boxed{
5x^4
}
$$

---

## Mistake 3 — Differentiating a constant as itself

Wrong:

$$
\frac{d}{dx}(7)=7
$$

Correct:

$$
\boxed{
\frac{d}{dx}(7)=0
}
$$

---

## Mistake 4 — Losing the constant multiplier

Wrong:

$$
\frac{d}{dx}(4x^3)=3x^2
$$

Correct:

$$
\boxed{
\frac{d}{dx}(4x^3)=12x^2
}
$$

Wrong:

\boxed{

# 7.13 Connection to Machine Learning

These basic rules appear directly in ML mathematics.

Consider a simple linear model:

$$
\hat y=wx+b
$$

Suppose we want to understand how the prediction changes with respect to the weight $w$.

Treating $x$ and $b$ as constants with respect to $w$:

$$
\frac{\partial \hat y}{\partial w}=x
$$

Similarly:

$$
\frac{\partial \hat y}{\partial b}=1
$$

The same differentiation ideas are being used.

For example, suppose a simple loss is:

$$
L(w)=w^2+3w+5
$$

Using the basic rules:

$$
\frac{dL}{dw}
=
2w+3
$$

The derivative tells us:

> **How the loss changes when the parameter $w$ changes.**

This is the basic mathematical idea behind gradient-based optimization.

Later, when we study **partial derivatives and gradients**, these rules will become building blocks for much more complex ML derivatives.

# 7.17 Key Takeaways

The five basic differentiation rules are:

$$
\boxed{
\frac{d}{dx}(c)=0
}
$$

$$
\boxed{
\frac{d}{dx}(x^n)=nx^{n-1}
}
$$

$$
\boxed{
\frac{d}{dx}[cf(x)]
=
cf'(x)
}
$$

$$
\boxed{
\frac{d}{dx}[f(x)+g(x)]
=
f'(x)+g'(x)
}
$$

$$
\boxed{
\frac{d}{dx}[f(x)-g(x)]
=
f'(x)-g'(x)
}
$$

### The deeper idea

These are not isolated tricks.

They allow us to break complicated expressions into smaller pieces:

$$
\boxed{
\text{Complex function}
\rightarrow
\text{simple terms}
\rightarrow
\text{differentiate each term}
\rightarrow
\text{combine}
}
$$

And because these rules ultimately come from the definition of the derivative, they are mathematically consistent with everything we learned in Parts 1–6.

### One sentence to remember

> **Basic differentiation is mainly about recognizing the structure of an expression and applying the appropriate rule to each part.**

# Some Additional Rule

# 7.3 Power Rule — Complete Understanding

The power rule is one of the most frequently used rules in differentiation.

We already know:

$$
\boxed{
\frac{d}{dx}(x^n)=nx^{n-1}
}
$$

So far, we have mainly seen positive integer powers such as:

$$
x^2,\quad x^3,\quad x^5
$$

But the power rule is much more powerful.

It also works for:

- Positive integer powers
- Zero power
- Negative powers
- Fractional powers

The general rule is:

$$
\boxed{
\frac{d}{dx}(x^n)=nx^{n-1}
}
$$

where $n$ can be any real number for which the function is defined and differentiable at the point being considered.

This means that expressions such as:

$$
x^{-2}
$$

and

$$
x^{1/2}
$$

can also be differentiated using the same rule.

# 7.3.1 Positive Integer Powers

Let's begin with the simplest case.

For:

$$
f(x)=x^n
$$

where $n$ is a positive integer:

$$
\boxed{
f'(x)=nx^{n-1}
}
$$

### Examples

$$
\frac{d}{dx}(x^2)=2x
$$

$$
\frac{d}{dx}(x^3)=3x^2
$$

$$
\frac{d}{dx}(x^4)=4x^3
$$

$$
\frac{d}{dx}(x^5)=5x^4
$$

$$
\frac{d}{dx}(x^{10})=10x^9
$$

The pattern is always:

$$
\boxed{
x^n
\rightarrow
nx^{n-1}
}
$$

### Three-step process

1. **Bring the exponent down.**
2. **Multiply by the original coefficient** if one exists.
3. **Reduce the exponent by 1.**

# 7.3.2 Zero Power

Recall:

$$
x^0=1
$$

Therefore:

$$
\frac{d}{dx}(x^0)
=
\frac{d}{dx}(1)
$$

Using the constant rule:

$$
=0
$$

The power rule gives the same result:

$$
\frac{d}{dx}(x^0)
=
0x^{-1}
=
0
$$

Therefore:

$$
\boxed{
\frac{d}{dx}(x^0)=0
}
$$

This is consistent with the constant rule because:

$$
x^0=1
$$

and $1$ is a constant.

### Important insight

The power rule and constant rule are not fighting each other.

They give the same answer:

$$
\boxed{
\frac{d}{dx}(1)=0
}
$$

# 7.3.3 Negative Powers

Now consider a negative exponent.

For example:

$$
f(x)=x^{-2}
$$

The power rule still applies:

$$
\frac{d}{dx}(x^{-2})
=
-2x^{-3}
$$

Therefore:

$$
\boxed{
\frac{d}{dx}(x^{-2})=-2x^{-3}
}
$$

We can rewrite the result using fractions:

$$
-2x^{-3}
=
-\frac{2}{x^3}
$$

So:

$$
\boxed{
\frac{d}{dx}\left(\frac{1}{x^2}\right)
=
-\frac{2}{x^3}
}
$$

for $x\neq0$.

## Why Can We Use the Power Rule for Negative Powers?

Remember:

$$
x^{-n}=\frac{1}{x^n}
$$

For example:

$$
x^{-2}=\frac{1}{x^2}
$$

The negative exponent simply tells us that the power appears in the denominator.

But the differentiation rule remains:

$$
\boxed{
x^n\rightarrow nx^{n-1}
}
$$

For example:

$$
x^{-3}
\rightarrow
-3x^{-4}
$$

The exponent:

$$
-3
$$

comes down as the coefficient.

Then reduce it by $1$:

$$
-3-1=-4
$$

Therefore:

$$
\boxed{
\frac{d}{dx}(x^{-3})
=
-3x^{-4}
}
$$

### Mental model

Do **not** create a separate differentiation rule for negative powers.

Simply use the same power rule:

$$
\boxed{
n x^{n-1}
}
$$

and allow $n$ to be negative.

# 7.3.4 Fractional Powers

Now consider a fractional exponent.

For example:

$$
f(x)=x^{1/2}
$$

We can recognize:

$$
x^{1/2}=\sqrt{x}
$$

The power rule still applies.

Using:

$$
\frac{d}{dx}(x^n)=nx^{n-1}
$$

with:

$$
n=\frac12
$$

we get:

$$
\frac{d}{dx}(x^{1/2})
=

$$
\boxed{
\frac{d}{dx}(\sqrt{x})
=
\frac{1}{2\sqrt{x}}

for $x>0$.

The important point is:

> **A fractional exponent does not require a completely new differentiation rule.**

It is still the power rule.

# 7.3.5 Fractional Powers as Roots

Fractional exponents and roots are two ways of writing the same thing.

For example:

$$

$$

$$
\boxed{

$$

And:

$$
\boxed{

$$

Therefore, the power rule can differentiate functions involving roots.

$$
\boxed{

$$

where defined.

# 7.3.7 A Note About Where the Function Is Defined

There is an important mathematical detail.

The power rule gives the derivative formula, but we must still consider **where the original function and derivative are defined**.

For example:

$$
f(x)=\sqrt{x}=x^{1/2}
$$

The real-valued function is defined for:

$$
x\geq0
$$

Its derivative for $x>0$ is:

$$
f'(x)=\frac{1}{2\sqrt{x}}
$$

But at:

$$
x=0
$$

this expression is undefined because it would require division by zero.

Therefore, the function is continuous at $0$, but it is **not differentiable at $0$**.

This is consistent with what we learned in Part 6.

So always distinguish between:

> **The derivative formula**

and:

> **The points where that derivative actually exists.**

# 7.3.8 Another Important Example — $x^{2/3}$

Consider:

$$
f(x)=x^{2/3}
$$

The function is defined for all real $x$.

Using the power rule:

$$
f'(x)
=
\frac23x^{-1/3}
=
\frac{2}{3\sqrt[3]{x}}
$$

This derivative is undefined at:

$$
x=0
$$

Therefore:

$$
\boxed{
f'(0)\text{ does not exist}
}
$$

This agrees with our earlier discussion of a **cusp**.

So the power rule still gives us the correct derivative wherever the derivative exists, but it does not magically make a function differentiable at points where differentiability fails.

### Important lesson

> **A differentiation rule tells us how to differentiate where the derivative exists; we still have to respect the function's domain and differentiability conditions.**

The power rule becomes even more useful when combined with the constant multiple rule.

# 7.3.12 Common Mistakes With the Power Rule

### Mistake 1 — Forgetting the coefficient

Wrong:

$$
\frac{d}{dx}(x^5)=x^4
$$

Correct:

$$
\boxed{
5x^4
}
$$

---

### Mistake 2 — Increasing instead of decreasing the exponent

Wrong:

$$
x^5\rightarrow5x^6
$$

Correct:

$$
\boxed{
x^5\rightarrow5x^4
}
$$

---

### Mistake 3 — Mishandling a negative exponent

For:

$$
x^{-2}
$$

the exponent is $-2$.

Therefore:

$$
\frac{d}{dx}(x^{-2})
=
-2x^{-3}
$$

Do not treat $-2$ as if it were positive.

---

### Mistake 4 — Treating a fractional exponent as a special unrelated rule

For:

$$
x^{1/2}
$$

simply use:

$$
\frac{d}{dx}(x^n)=nx^{n-1}
$$

with:

$$
n=\frac12
$$

---

### Mistake 5 — Ignoring where the derivative exists

For:

$$
f(x)=\sqrt{x}
$$

we obtain:

$$
f'(x)=\frac{1}{2\sqrt{x}}
$$

but this derivative is not defined at $x=0$.

The formula and its domain must both be considered.

# 7.3.13 Big Picture of the Power Rule

We started with:

$$
\boxed{
\frac{d}{dx}(x^n)=nx^{n-1}
}
$$

Now we can apply it to:

### Positive powers

$$
x^5\rightarrow5x^4
$$

### Zero power

$$
x^0\rightarrow0
$$

### Negative powers

$$
x^{-3}\rightarrow-3x^{-4}
$$

### Fractional powers

$$
x^{1/2}\rightarrow\frac12x^{-1/2}
$$

So the same rule handles all of them.

### The deeper lesson

The power rule is not:

> "A rule only for $x^2,x^3,x^4,\ldots$."

It is much more general:

$$
\boxed{
\text{For a power }x^n,\text{ multiply by }n
\text{ and reduce the exponent by }1.
}
$$

The only additional responsibility is to check where the resulting derivative is actually defined.
