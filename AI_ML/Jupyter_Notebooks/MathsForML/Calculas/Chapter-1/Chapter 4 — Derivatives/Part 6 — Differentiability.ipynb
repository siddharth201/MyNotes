# Chapter 3 — Derivatives

## Part 6 — Differentiability

So far, we have learned that a derivative tells us the **local rate of change** of a function.

We have also seen that geometrically, the derivative represents the **slope of the tangent line**.

But there is an important question:

> **Does every function have a derivative at every point?**

The answer is:

$$
\boxed{\text{No}}
$$

A function can fail to have a derivative at certain points.

This leads us to an important concept:

# Differentiability

We will understand:

- What does **differentiable** mean?
- The relationship between **continuity and differentiability**
- Why **corners** cause problems
- Why **cusps** cause problems
- Why **discontinuities** prevent differentiability

The central idea is:

> A function is differentiable at a point if its derivative exists at that point.

# 6.1 What Does "Differentiable" Mean?

A function is **differentiable at a point** if its derivative exists at that point.

Recall the definition of a derivative:

$$
f'(a)
=
\lim_{h\to0}
\frac{f(a+h)-f(a)}{h}
$$

Therefore, $f$ is differentiable at $x=a$ if this limit exists as a **finite, well-defined value**.

In simple terms:

> **Differentiable means that the function has a well-defined instantaneous rate of change at that point.**

Geometrically:

> **Differentiable means that the function has a well-defined tangent slope at that point.**

## Example — A Smooth Function

Consider:

$$
f(x)=x^2
$$

Its derivative is:

$$
f'(x)=2x
$$

This derivative exists for every real value of $x$.

Therefore:

$$
\boxed{x^2\text{ is differentiable everywhere}}
$$

For example, at $x=2$:

$$
f'(2)=4
$$

So the graph has a well-defined tangent slope of $4$ at that point.

There is no sudden break, corner, cusp, or other obstruction preventing the local slope from being defined.

# 6.2 Differentiable at a Point vs Differentiable Everywhere

These two statements are different.

### Differentiable at a point

If the derivative exists at $x=a$:

$$
f'(a)\text{ exists}
$$

then $f$ is differentiable **at $a$**.

### Differentiable everywhere

If the derivative exists at every point in the domain:

$$
f'(x)\text{ exists for every }x\text{ in the domain}
$$

then $f$ is differentiable **everywhere on its domain**.

For example:

$$
f(x)=x^2
$$

is differentiable everywhere.

But a function can be differentiable at most points and fail to be differentiable at one or more specific points.

# 6.3 Continuity vs Differentiability

This is one of the most important relationships in calculus.

We already learned about **continuity**.

Informally, a function is continuous at a point if its graph has no break, jump, or hole at that point.

Differentiability asks a different question:

> **Does the function have a well-defined instantaneous rate of change at that point?**

So:

### Continuity

asks:

> **Is the function connected/smoothly present at the point?**

### Differentiability

asks:

> **Does a well-defined derivative exist at the point?**

These are related—but they are **not the same thing**.

# 6.4 Differentiability Implies Continuity

There is a very important theorem:

$$
\boxed{
\text{Differentiable at }a
\Rightarrow
\text{Continuous at }a
}
$$

In other words:

> If a function is differentiable at a point, it must also be continuous at that point.

But the reverse is **not** necessarily true.

$$
\boxed{
\text{Continuous at }a
\not\Rightarrow
\text{Differentiable at }a
}
$$

This distinction is extremely important.

### The relationship

Think of it as:

$$
\boxed{
\text{Differentiability}
\Longrightarrow
\text{Continuity}
}
$$

but:

$$
\boxed{
\text{Continuity}
\not\Longrightarrow
\text{Differentiability}
}
$$

# 6.5 Why Does Differentiability Guarantee Continuity?

Suppose $f$ is differentiable at $a$.

Then:

$$
f'(a)
=
\lim_{h\to0}
\frac{f(a+h)-f(a)}{h}
$$

exists and is finite.

We can rewrite the change in function value as:

$$
f(a+h)-f(a)
=
\left[
\frac{f(a+h)-f(a)}{h}
\right]h
$$

Now let:

$$
h\to0
$$

The first part approaches the finite value $f'(a)$:

$$
\frac{f(a+h)-f(a)}{h}
\to f'(a)
$$

while:

$$
h\to0
$$

Therefore:

$$
f(a+h)-f(a)\to0
$$

which means:

$$
f(a+h)\to f(a)
$$

This is exactly the condition for continuity at $a$.

Hence:

$$
\boxed{
\text{Differentiability at }a
\Rightarrow
\text{Continuity at }a
}
$$

### Intuition

If a function has a finite, well-defined instantaneous slope at a point, it cannot suddenly jump away from that point.

A sudden jump would make the local rate of change impossible to define normally.

# 6.6 But Continuity Does Not Guarantee Differentiability

Consider:

$$
f(x)=|x|
$$

This function is continuous at:

$$
x=0
$$

The graph reaches the point:

$$
(0,0)
$$

without a break, jump, or hole.

But something unusual happens at $x=0$.

The graph has a **corner**.

Therefore, the derivative at $x=0$ does not exist.

So:

$$
\boxed{
|x|\text{ is continuous at }0
}
$$

but:

$$
\boxed{
|x|\text{ is not differentiable at }0
}
$$

This gives us a concrete counterexample to:

> "Every continuous function must be differentiable."

That statement is false.

# 6.7 Corners

A **corner** occurs when the graph suddenly changes its direction of slope.

The classic example is:

$$
f(x)=|x|
$$

Written piecewise:

$$
f(x)=
\begin{cases}
-x, & x<0\\
x, & x\geq0
\end{cases}
$$

Look at the slopes on the two sides.

For $x<0$:

$$
f(x)=-x
$$

so:

$$
f'(x)=-1
$$

For $x>0$:

$$
f(x)=x
$$

so:

$$
f'(x)=1
$$

Therefore, as we approach $x=0$:

### From the left

$$
\lim_{x\to0^-}f'(x)=-1
$$

### From the right

$$
\lim_{x\to0^+}f'(x)=1
$$

The two slopes are different.

Therefore there is no single tangent slope at $x=0$.

Hence:

$$
\boxed{f'(0)\text{ does not exist}}
$$

## Why Exactly Does the Corner Break Differentiability?

Recall:

$$
f'(0)
=
\lim_{h\to0}
\frac{f(0+h)-f(0)}{h}
$$

For:

$$
f(x)=|x|
$$

we have:

$$
f(0)=0
$$

### Approach from the right

For $h>0$:

$$
\frac{|h|-0}{h}
=
\frac{h}{h}
=
1
$$

Therefore:

$$
\lim_{h\to0^+}
\frac{|h|}{h}
=
1
$$

### Approach from the left

For $h<0$:

$$
|h|=-h
$$

so:

$$
\frac{|h|}{h}
=
\frac{-h}{h}
=
-1
$$

Therefore:

$$
\lim_{h\to0^-}
\frac{|h|}{h}
=
-1
$$

Since:

$$
1\neq-1
$$

the two-sided limit does not exist.

Therefore:

$$
\boxed{
f'(0)\text{ does not exist}
}
$$

### The key lesson

A corner is not differentiable because:

> **The slope approaching the point from the left is different from the slope approaching it from the right.**

# 6.8 Cusps

A **cusp** is another type of sharp point where the derivative fails to exist.

A classic example is:

$$
f(x)=x^{2/3}
$$

At:

$$
x=0
$$

the graph has a sharp cusp.

Its derivative for $x\neq0$ is:

$$
f'(x)
=
\frac{2}{3}x^{-1/3}
$$

or:

$$
f'(x)
=
\frac{2}{3\sqrt[3]{x}}
$$

As $x$ approaches $0$:

From the right:

$$
f'(x)\to+\infty
$$

From the left:

$$
f'(x)\to-\infty
$$

So there is no finite derivative at $x=0$.

Therefore:

$$
\boxed{
f'(0)\text{ does not exist}
}
$$

### Important distinction

For a corner such as $|x|$:

- left slope = finite value
- right slope = different finite value

For a cusp such as $x^{2/3}$:

- slopes become unbounded
- there is no finite tangent slope

# 6.9 Corner vs Cusp

It is useful to distinguish them.

| Feature | Corner | Cusp |
|---|---|---|
| Graph is continuous? | Usually yes | Usually yes |
| Left/right behavior | Different finite slopes | Slopes become unbounded/oppositely directed |
| Example | $|x|$ | $x^{2/3}$ |
| Differentiable? | No | No |
| Main problem | No single slope | No finite derivative |

Both demonstrate an important principle:

> **Continuity alone is not enough for differentiability.**

# 6.10 Discontinuities

A **discontinuity** occurs when a function is not continuous at a point.

Examples include:

- a jump
- a hole
- an infinite discontinuity
- other breaks in the graph

A discontinuity automatically prevents differentiability at that point.

Why?

Because:

$$
\boxed{
\text{Differentiable}
\Rightarrow
\text{Continuous}
}
$$

Therefore, if a function is **not continuous**, it cannot be differentiable.

So:

$$
\boxed{
\text{Discontinuous at }a
\Rightarrow
\text{Not differentiable at }a
}
$$

## Example — Jump Discontinuity

Consider:

$$
f(x)=
\begin{cases}
0, & x<0\\
1, & x\geq0
\end{cases}
$$

At $x=0$:

From the left:

$$
\lim_{x\to0^-}f(x)=0
$$

From the right:

$$
\lim_{x\to0^+}f(x)=1
$$

Since:

$$
0\neq1
$$

the function is discontinuous at $0$.

Therefore it cannot be differentiable there.

We do not even need to calculate the derivative.

The discontinuity itself is enough to conclude:

$$
\boxed{
f'(0)\text{ does not exist}
}
$$

# 6.11 The Hierarchy to Remember

There is a very useful logical hierarchy:

$$
\boxed{
\text{Differentiability}
\Rightarrow
\text{Continuity}
}
$$

But:

$$
\boxed{
\text{Continuity}
\not\Rightarrow
\text{Differentiability}
}
$$

Therefore:

### If a function is differentiable:

It must be continuous.

### If a function is continuous:

It **may or may not** be differentiable.

For example:

$$
x^2
$$

is:

$$
\text{continuous}
+
\text{differentiable}
$$

while:

$$
|x|
$$

is:

$$
\text{continuous}
+
\text{not differentiable at }0
$$

and a jump function is:

$$
\text{not continuous}
+
\text{not differentiable}
$$

# 6.12 Four Important Cases

Let's organize the possibilities.

| Function behavior at a point | Continuous? | Differentiable? |
|---|---:|---:|
| Smooth curve | Yes | Yes |
| Corner | Yes | No |
| Cusp | Yes | No |
| Discontinuity | No | No |

This table gives us an extremely useful diagnostic.

If you discover that a function is **discontinuous**, you immediately know:

$$
\boxed{\text{Not differentiable}}
$$

If you discover a **corner or cusp**, the function may still be continuous, but you must check differentiability separately.

# 6.13 How Do We Check Whether a Function Is Differentiable?

At a particular point $x=a$, a practical approach is:

### Step 1 — Check continuity

Ask:

$$
\lim_{x\to a}f(x)=f(a)?
$$

If not:

$$
\boxed{\text{Not differentiable}}
$$

because differentiability requires continuity.

---

### Step 2 — If continuous, check the derivative

Calculate:

$$
f'(a)
=
\lim_{h\to0}
\frac{f(a+h)-f(a)}{h}
$$

If the limit exists and is finite:

$$
\boxed{\text{Differentiable at }a}
$$

If the limit does not exist or is not finite:

$$
\boxed{\text{Not differentiable at }a}
$$

This is particularly important at:

- corners
- cusps
- piecewise-function boundaries
- other unusual points

# 6.14 Visual Intuition

When looking at a graph, ask:

> **Can I assign one well-defined local slope to this point?**

### Smooth point

A tangent slope can be clearly defined:

$$
\boxed{\text{Differentiable}}
$$

### Corner

The direction changes abruptly:

$$
\boxed{\text{Not differentiable}}
$$

### Cusp

The slope becomes unbounded/sharply directed:

$$
\boxed{\text{Not differentiable}}
$$

### Discontinuity

The graph is broken:

$$
\boxed{\text{Not differentiable}}
$$

However, remember:

> A graph looking smooth is useful intuition, but the rigorous test is whether the derivative limit exists.

# 6.15 Why Does Differentiability Matter in Machine Learning?

Differentiability is extremely important in Machine Learning because many ML algorithms rely on **gradients**.

Suppose our model has a loss:

$$
L(w)
$$

We want to know:

$$
\frac{dL}{dw}
$$

This tells us how the loss changes when the parameter $w$ changes.

Gradient-based optimization then uses this information to decide how to update parameters.

For example, gradient descent uses the idea:

$$
w_{\text{new}}
=
w_{\text{old}}
-
\eta
\frac{dL}{dw}
$$

where $\eta$ is the learning rate.

But this requires the relevant derivative to exist.

Therefore, when studying ML models and activation functions, differentiability becomes an important consideration.

### Important ML connection

A function does **not** necessarily need to be differentiable at every point to be useful in ML.

For example, some widely used functions have points where the ordinary derivative does not exist, yet optimization methods can still work using suitable conventions or generalized/subgradient ideas.

We will study these ideas later when we discuss **activation functions and optimization**.

For now, remember:

> **Differentiability tells us whether a meaningful local gradient is available at a point.**

# 6.16 Mental Model

Think of a function as a road.

### Smooth road

You can determine the road's direction at every point.

$$
\boxed{\text{Differentiable}}
$$

### Corner

The road suddenly changes direction.

There is no single direction exactly at the corner.

$$
\boxed{\text{Continuous but not differentiable}}
$$

### Cusp

The road forms a sharp pointed shape and the slope becomes unbounded.

$$
\boxed{\text{Not differentiable}}
$$

### Broken road

The road itself is disconnected.

$$
\boxed{\text{Discontinuous and therefore not differentiable}}
$$

The key question is always:

> **Can we define one meaningful instantaneous rate of change at this point?**

# 6.18 Key Takeaways

1. **Differentiable** means that the derivative exists at the point.

2. The derivative is defined through the limit:

   $$
   f'(a)
   =
   \lim_{h\to0}
   \frac{f(a+h)-f(a)}{h}
   $$

3. Differentiability implies continuity:

   $$
   \boxed{
   \text{Differentiable}
   \Rightarrow
   \text{Continuous}
   }
   $$

4. Continuity does not imply differentiability:

   $$
   \boxed{
   \text{Continuous}
   \not\Rightarrow
   \text{Differentiable}
   }
   $$

5. A **corner** can be continuous but not differentiable.

   Example:

   $$
   f(x)=|x|
   $$

6. A **cusp** can also be continuous but not differentiable.

   Example:

   $$
   f(x)=x^{2/3}
   $$

7. A **discontinuity** automatically means the function is not differentiable at that point.

8. The most useful geometric question is:

   > **Does the function have one well-defined instantaneous slope at the point?**

9. In Machine Learning, differentiability matters because gradient-based methods depend on derivatives or related notions of gradients.

### The central relationship

$$
\boxed{
\text{Differentiability}
\Longrightarrow
\text{Continuity}
}
$$

but:

$$
\boxed{
\text{Continuity}
\not\Longrightarrow
\text{Differentiability}
}
$$
