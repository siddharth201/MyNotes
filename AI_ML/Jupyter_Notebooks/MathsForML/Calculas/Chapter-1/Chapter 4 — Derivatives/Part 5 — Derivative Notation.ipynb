# Chapter 3 — Derivatives

## Part 5 — Derivative Notation

In the previous parts, we learned **what a derivative means**:

> A derivative tells us how rapidly a quantity changes at a particular point.

We also saw that the derivative can be understood geometrically as the **slope of the tangent line**.

Now we need to learn how mathematicians **write** derivatives.

There are three important notations:

1. **Prime notation**
   $$
   f'(x)
   $$

2. **Leibniz notation**
   $$
   \frac{dy}{dx}
   $$

3. **Operator notation**
   $$
   \frac{d}{dx}
   $$

These are different ways of expressing the same underlying idea.

The notation may look complicated at first, but each notation answers a very simple question:

> **What is changing, and with respect to what?**

## 5.1 First, Recall What a Function Does

Suppose we have

$$
y=f(x)
$$

For example,

$$
y=x^2
$$

Here:

- $x$ is the **input**
- $y$ is the **output**
- $y$ depends on $x$

So when $x$ changes, $y$ can change.

For example:

$$
x=2 \Rightarrow y=4
$$

and

$$
x=3 \Rightarrow y=9
$$

The derivative asks:

> **How much does $y$ change when $x$ changes, locally around a particular value of $x$?**

That idea is what all three derivative notations represent.

# 5.2 Prime Notation — $f'(x)$

The most common notation for the derivative of a function $f(x)$ is:

$$
f'(x)
$$

It is read as:

> **"f prime of x"**

The small symbol

$$
'
$$

is called a **prime**.

Therefore:

$$
f'(x)
$$

means:

> **the derivative of $f(x)$ with respect to $x$.**

## Example

Suppose:

$$
f(x)=x^2
$$

Its derivative is:

$$
f'(x)=2x
$$

Notice the difference:

$$
f(x)=x^2
$$

is the **original function**.

Whereas:

$$
f'(x)=2x
$$

is the **derivative function**.

The derivative function tells us the slope of the original function at every value of $x$.

## Derivative Function vs Derivative at a Point

This distinction is extremely important.

Suppose:

$$
f'(x)=2x
$$

This is a **function**.

It gives us a derivative value for any $x$.

For example:

$$
f'(1)=2
$$

$$
f'(2)=4
$$

$$
f'(5)=10
$$

So:

### $f'(x)$

means:

> The derivative as a function of $x$.

### $f'(a)$

means:

> The derivative evaluated at the specific point $x=a$.

Therefore:

$$
f'(x)=2x
$$

is the general derivative function, while

$$
f'(3)=6
$$

is one particular derivative value.

### Mental Model

Think of $f'(x)$ as a **slope calculator**.

Give it an $x$-value, and it tells you the slope at that location:

$$
x \longrightarrow f'(x)
$$

For $f(x)=x^2$:

$$
3 \longrightarrow f'(3)=6
$$

So the graph has slope $6$ at $x=3$.

# 5.3 Why Do We Put a Prime?

The prime symbol is simply a notation that tells us:

> **This is the derivative of the original function.**

For example:

$$
f(x)
$$

Original function.

$$
f'(x)
$$

First derivative.

If we differentiate again:

$$
f''(x)
$$

This is called the **second derivative**.

Differentiate one more time:

$$
f'''(x)
$$

This is the **third derivative**.

So the prime notation can also indicate repeated differentiation:

$$
f(x)
\rightarrow
f'(x)
\rightarrow
f''(x)
\rightarrow
f'''(x)
$$

We will study higher-order derivatives later.

# 5.4 Leibniz Notation — $\frac{dy}{dx}$

Another extremely important way to write a derivative is:

$$
\frac{dy}{dx}
$$

It is commonly read as:

> **"dy by dx"**

or more meaningfully:

> **"the derivative of $y$ with respect to $x$."**

Suppose:

$$
y=f(x)
$$

Then we can write its derivative as:

$$
\frac{dy}{dx}
$$

So, conceptually:

$$
f'(x)
\quad\text{and}\quad
\frac{dy}{dx}
$$

can represent the same derivative.

## Example

Suppose:

$$
y=x^2
$$

Then:

$$
\frac{dy}{dx}=2x
$$

Compare:

$$
f(x)=x^2
$$

with:

$$
y=x^2
$$

These are simply two ways of naming the same function.

Therefore:

$$
f'(x)=2x
$$

and

$$
\frac{dy}{dx}=2x
$$

describe the same derivative.

### Why is the notation different?

The two notations emphasize slightly different things.

### Prime notation

$$
f'(x)
$$

emphasizes:

> **"I am taking the derivative of the function $f$."**

### Leibniz notation

$$
\frac{dy}{dx}
$$

emphasizes:

> **"I am measuring how $y$ changes with respect to $x$."**

This second interpretation becomes extremely useful in calculus and machine learning.

# 5.5 Why Does $\frac{dy}{dx}$ Look Like a Fraction?

At first glance:

$$
\frac{dy}{dx}
$$

looks like an ordinary fraction.

And in many situations, thinking of it as:

$$
\frac{\text{change in }y}{\text{change in }x}
$$

is a very useful intuition.

Recall the average rate of change:

$$
\frac{\Delta y}{\Delta x}
$$

This means:

> Change in $y$ divided by change in $x$.

The derivative comes from making the change in $x$ smaller and smaller:

$$
\frac{dy}{dx}
=
\lim_{\Delta x\to0}
\frac{\Delta y}{\Delta x}
$$

More precisely, using $h$:

$$
\frac{dy}{dx}
=
\lim_{h\to0}
\frac{f(x+h)-f(x)}{h}
$$

So the notation $\frac{dy}{dx}$ has a very natural connection with the idea of **rate of change**.

### Important Precision

Although $\frac{dy}{dx}$ is written like a fraction, it should **not always be treated as an ordinary fraction**.

It is fundamentally a notation for a derivative.

There are situations where differential notation can be manipulated in ways that resemble fraction algebra, but this works because of deeper mathematical rules—not simply because $dy$ and $dx$ are ordinary numbers sitting above and below a fraction bar.

For now, the safest mental model is:

$$
\boxed{
\frac{dy}{dx}
=
\text{rate of change of }y\text{ with respect to }x
}
$$

# 5.6 Operator Notation — $\frac{d}{dx}$

The third notation is:

$$
\frac{d}{dx}
$$

This is called the **derivative operator**.

The word **operator** means:

> Something that acts on an object and produces a result.

You have already seen operators in mathematics.

For example:

$$
+
$$

is an addition operator.

$$
\times
$$

is a multiplication operator.

Similarly,

$$
\frac{d}{dx}
$$

is an operator that means:

> **Take the derivative with respect to $x$.**

## Example

Suppose:

$$
y=x^2
$$

We can apply the derivative operator to $x^2$:

$$
\frac{d}{dx}(x^2)
$$

The result is:

$$
\frac{d}{dx}(x^2)=2x
$$

Here:

$$
\frac{d}{dx}
$$

is the **instruction**:

> Differentiate with respect to $x$.

And:

$$
x^2
$$

is the **thing being differentiated**.

## Why Is $\frac{d}{dx}$ Different From $\frac{dy}{dx}$?

Consider:

$$
y=x^2
$$

We can write:

$$
\frac{dy}{dx}=2x
$$

Here, $\frac{dy}{dx}$ is the **resulting derivative**.

But:

$$
\frac{d}{dx}(x^2)=2x
$$

Here, $\frac{d}{dx}$ is the **operation that produces the derivative**.

A useful analogy is:

$$
\text{Operator} + \text{Input} \rightarrow \text{Output}
$$

Therefore:

$$
\frac{d}{dx}
+
x^2
\rightarrow
2x
$$

Or more formally:

$$
\boxed{
\frac{d}{dx}(x^2)=2x
}
$$

### Mental Model

Think of:

$$
\frac{d}{dx}
$$

as a **"differentiate this" machine**.

You give it:

$$
x^2
$$

and it returns:

$$
2x
$$

# 5.7 The Three Notations Together

Suppose:

$$
y=f(x)=x^2
$$

The derivative can be written in three equivalent ways:

### Prime notation

$$
f'(x)=2x
$$

### Leibniz notation

$$
\frac{dy}{dx}=2x
$$

### Operator notation

$$
\frac{d}{dx}(x^2)=2x
$$

The notation changes, but the underlying mathematical idea is the same:

> **How rapidly does the output change when the input $x$ changes?**

## A Very Important Relationship

If:

$$
y=f(x)
$$

then we can write:

$$
\boxed{
f'(x)
=
\frac{dy}{dx}
=
\frac{d}{dx}[f(x)]
}
$$

These expressions represent the derivative of the same function with respect to $x$.

However, they are not literally the same *notation structure*:

- $f'(x)$ uses **prime notation**
- $\frac{dy}{dx}$ uses **Leibniz notation**
- $\frac{d}{dx}$ is the **derivative operator**

Understanding this distinction prevents a lot of confusion later.

# 5.8 What Does "With Respect To" Mean?

This phrase is one of the most important ideas in calculus.

When we say:

> **"Change in $y$ with respect to $x$"**

we mean:

> **We are changing $x$ and observing how $y$ responds.**

So:

$$
\frac{dy}{dx}
$$

means:

> How does $y$ change when $x$ changes?

The phrase **"with respect to"** tells us the variable whose change we are considering.

Here:

$$
\frac{dy}{dx}
$$

- $y$ = quantity whose change we are observing
- $x$ = quantity with respect to which we measure that change

## A Simple Real-Life Example

Suppose a car is moving.

Let:

$$
s=t^2
$$

where:

- $s$ = position
- $t$ = time

The derivative is:

$$
\frac{ds}{dt}=2t
$$

Read this as:

> **How does position change with respect to time?**

That is exactly what we call **velocity**.

So:

$$
\boxed{
\frac{ds}{dt}
=
\text{velocity}
}
$$

The phrase "with respect to time" is telling us:

> Time is the variable whose change we are using to measure the change in position.

# 5.9 "With Respect To" in Machine Learning

This idea becomes extremely important in Machine Learning.

Suppose our model has a parameter $w$ and a loss function:

$$
L(w)
$$

We can calculate:

$$
\frac{dL}{dw}
$$

Read this as:

> **How does the loss $L$ change with respect to the weight $w$?**

In other words:

> If we slightly change $w$, how does the loss respond?

Similarly, if the model has a bias $b$:

$$
\frac{dL}{db}
$$

means:

> **How does the loss change with respect to the bias $b$?**

This is the fundamental idea behind gradient-based optimization.

## The Variable After "with Respect To" Matters

Consider:

$$
L(w)=w^2
$$

Then:

$$
\frac{dL}{dw}=2w
$$

We are asking:

> How does $L$ change when $w$ changes?

Now imagine:

$$
L(w,b)=w^2+3b^2
$$

If we differentiate with respect to $w$:

$$
\frac{\partial L}{\partial w}=2w
$$

If we differentiate with respect to $b$:

$$
\frac{\partial L}{\partial b}=6b
$$

The quantity being differentiated is the same:

$$
L(w,b)
$$

But the variable we are asking about is different.

This is the key idea:

> **"With respect to" tells us which variable's change we are studying.**

We will study the $\partial$ notation in detail when we reach **partial derivatives**.

# 5.10 One Example — All Three Notations

Let:

$$
f(x)=3x^2+2x+1
$$

Its derivative is:

$$
f'(x)=6x+2
$$

If we define:

$$
y=3x^2+2x+1
$$

then:

$$
\frac{dy}{dx}=6x+2
$$

Using the derivative operator:

$$
\frac{d}{dx}
\left(3x^2+2x+1\right)
=
6x+2
$$

Therefore:

$$
\boxed{
f'(x)
=
\frac{dy}{dx}
=
\frac{d}{dx}[f(x)]
=
6x+2
}
$$

All three are describing the same derivative.

# 5.11 Quick Comparison

| Notation | Name | Main idea |
|---|---|---|
| $f'(x)$ | Prime notation | Derivative of function $f$ |
| $\frac{dy}{dx}$ | Leibniz notation | Change in $y$ with respect to $x$ |
| $\frac{d}{dx}$ | Derivative operator | Instruction to differentiate with respect to $x$ |

### Think of them like this:

$$
\boxed{
f'(x)
}
$$

**"What is the derivative of $f$?"**

---

$$
\boxed{
\frac{dy}{dx}
}
$$

**"How does $y$ change with respect to $x$?"**

---

$$
\boxed{
\frac{d}{dx}
}
$$

**"Differentiate this with respect to $x$."**

# 5.12 Mental Model

Keep this picture in your mind:

$$
\boxed{
\text{Input }x
\longrightarrow
\boxed{f}
\longrightarrow
\text{Output }y
}
$$

Now change $x$ slightly.

The question is:

> **How much does $y$ respond?**

That response is the derivative.

So:

$$
\boxed{
\frac{dy}{dx}
=
\frac{\text{change in output}}
{\text{change in input}}
}
$$

locally, in the limiting sense.

Or:

$$
\boxed{
f'(x)
=
\text{local slope of }f
}
$$

And:

$$
\boxed{
\frac{d}{dx}
=
\text{the operation used to find that slope}
}
$$

Finally:

> **"With respect to $x$" means that $x$ is the variable whose change we are using to study the response.**

# 5.14 Key Takeaways

1. A derivative can be written in different notations.

2. **Prime notation:**
   $$
   f'(x)
   $$
   means the derivative of $f$.

3. **Leibniz notation:**
   $$
   \frac{dy}{dx}
   $$
   means the rate of change of $y$ with respect to $x$.

4. **Operator notation:**
   $$
   \frac{d}{dx}
   $$
   means "differentiate with respect to $x$."

5. $f'(x)$ is the **derivative function**, while $f'(a)$ is the **derivative at a specific point**.

6. The phrase **"with respect to"** tells us which variable's change we are studying.

7. In Machine Learning:

   $$
   \frac{dL}{dw}
   $$

   asks:

   > How does the loss change when the weight changes?

8. The three notations can describe the same derivative:

   $$
   \boxed{
   f'(x)
   =
   \frac{dy}{dx}
   =
   \frac{d}{dx}[f(x)]
   }
   $$

### The single most important idea

> **A derivative measures how one quantity changes with respect to another quantity.**

Everything else is notation around this idea.
