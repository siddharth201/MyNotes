# Chapter 3 — Derivatives

## Part 8 — Derivatives of Common Functions

In Part 7, we learned the basic differentiation rules:

- Constant Rule
- Power Rule
- Constant Multiple Rule
- Sum Rule
- Difference Rule

These rules allow us to differentiate many algebraic expressions.

However, Machine Learning and applied mathematics use many other types of functions, such as:

$$
e^x,\qquad \ln x,\qquad \sin x,\qquad \cos x
$$

So we need to build a small **derivative library** for commonly occurring functions.

In this part, we will learn the derivatives of:

1. **Polynomial functions**
2. **Exponential functions**
3. **Logarithmic functions**
4. **Trigonometric functions**
5. **Important special cases**

The goal is not simply to memorize a table.

We want to understand:

> **What type of function are we looking at, and what derivative does that function naturally produce?**

# 8.1 The Big Picture

A useful way to think about this part is:

$$
\boxed{
\text{Recognize the function}
\rightarrow
\text{Know its basic derivative}
\rightarrow
\text{Apply the rules we already learned}
}
$$

For example:

### Polynomial

$$
x^4
\rightarrow
4x^3
$$

### Exponential

$$
e^x
\rightarrow
e^x
$$

### Logarithmic

$$
\ln x
\rightarrow
\frac1x
$$

### Trigonometric

$$
\sin x
\rightarrow
\cos x
$$

These become fundamental building blocks for more advanced differentiation.

Later, when we learn the **product rule, quotient rule, and chain rule**, we will combine these basic derivatives to differentiate much more complicated ML functions.

# 8.2 Polynomial Functions

A polynomial has the general form:

$$
f(x)
=
a_nx^n
+
a_{n-1}x^{n-1}
+
\cdots
+
a_2x^2
+
a_1x
+
a_0
$$

where the coefficients:

$$
a_n,a_{n-1},\ldots,a_0
$$

are constants.

For example:

$$
f(x)=3x^4-5x^3+2x^2+7x-9
$$

The derivative of a polynomial is found by differentiating **each term independently**.

This works because of the Sum Rule and Difference Rule from Part 7.

# 8.2.1 Derivative of a Polynomial

Consider:

$$
f(x)=3x^4-5x^3+2x^2+7x-9
$$

Differentiate each term.

### First term

$$
\frac{d}{dx}(3x^4)
=
3(4x^3)
=
12x^3
$$

### Second term

$$
\frac{d}{dx}(-5x^3)
=
-5(3x^2)
=
-15x^2
$$

### Third term

$$
\frac{d}{dx}(2x^2)
=
2(2x)
=
4x
$$

### Fourth term

$$
\frac{d}{dx}(7x)
=
7
$$

### Fifth term

$$
\frac{d}{dx}(-9)
=
0
$$

Therefore:

$$
\boxed{
f'(x)
=
12x^3-15x^2+4x+7
}
$$

This is simply the basic differentiation rules from Part 7 working together.

# 8.2.2 General Polynomial Derivative

For:

$$
f(x)
=
a_nx^n
+
a_{n-1}x^{n-1}
+
\cdots
+
a_1x
+
a_0
$$

the derivative is:

$$
\boxed{
f'(x)
=
na_nx^{n-1}
+
(n-1)a_{n-1}x^{n-2}
+
\cdots
+
a_1
}
$$

The constant term:

$$
a_0
$$

disappears because:

$$
\frac{d}{dx}(a_0)=0
$$

### Important observation

Differentiating a polynomial:

- reduces the degree by one, provided the polynomial is nonconstant;
- keeps the polynomial structure;
- removes the constant term.

For example:

$$
5x^4-3x^2+7x+10
$$

becomes:

$$
20x^3-6x+7
$$

# 8.3 Exponential Functions

An exponential function has the variable in the exponent.

For example:

$$
2^x,\qquad 5^x,\qquad e^x
$$

This is different from a polynomial such as:

$$
x^2
$$

because in:

$$
x^2
$$

the variable is the **base**.

But in:

$$
2^x
$$

the variable is the **exponent**.

This distinction is important.

Among all exponential functions, one base is particularly special:

$$
\boxed{e}
$$

where:

$$
e\approx2.71828
$$

The exponential function:

$$
e^x
$$

has an extraordinary property:

$$
\boxed{
\frac{d}{dx}(e^x)=e^x
}
$$

In other words:

> **The exponential function $e^x$ is its own derivative.**

# 8.3.1 Why Is $e^x$ Special?

Consider:

$$
f(x)=e^x
$$

Its derivative is:

$$
f'(x)=e^x
$$

So at every point:

$$
\text{rate of change}
=
\text{current value}
$$

For example:

At $x=0$:

$$
e^0=1
$$

and:

$$
f'(0)=1
$$

At $x=1$:

$$
f(1)=e
$$

and:

$$
f'(1)=e
$$

At $x=2$:

$$
f(2)=e^2
$$

and:

$$
f'(2)=e^2
$$

The function and its instantaneous rate of change are always equal.

### Mental Model

Think of $e^x$ as the function that grows at a rate exactly proportional to its current value.

That is why $e$ appears naturally in:

- continuous growth
- probability
- statistics
- differential equations
- optimization
- Machine Learning

# 8.3.2 Exponential Function With Another Base

Consider:

$$
f(x)=a^x
$$

where:

$$
a>0,\qquad a\neq1
$$

Its derivative is:

$$
\boxed{
\frac{d}{dx}(a^x)=a^x\ln(a)
}
$$

Notice what happens when:

$$
a=e
$$

Since:

$$
\ln(e)=1
$$

we obtain:

$$
\frac{d}{dx}(e^x)
=
e^x\ln(e)
=
e^x
$$

This is precisely why $e$ is special.

### Examples

$$
\frac{d}{dx}(2^x)
=
2^x\ln2
$$

and:

$$
\frac{d}{dx}(10^x)
=
10^x\ln10
$$

Therefore:

$$
\boxed{
e^x
\text{ is the unique exponential function whose derivative is itself.}
}
$$

# 8.3.3 Exponential Functions With a Constant Multiplier

Using the Constant Multiple Rule:

$$
f(x)=5e^x
$$

Then:

$$
f'(x)
=
5\frac{d}{dx}(e^x)
$$

Therefore:

$$
\boxed{
f'(x)=5e^x
}
$$

Similarly:

$$
\frac{d}{dx}(-3e^x)
=
-3e^x
$$

and:

$$
\frac{d}{dx}(7\cdot2^x)
=
7(2^x\ln2)
$$

So the rules from Part 7 continue to work with exponential functions.

# 8.4 Logarithmic Functions

A logarithm answers the question:

> **What exponent do I need to raise a base to in order to obtain a particular number?**

For example:

$$
\log_2(8)=3
$$

because:

$$
2^3=8
$$

The logarithm with base $e$ is called the **natural logarithm**:

$$
\boxed{\ln x=\log_e x}
$$

The natural logarithm is particularly important in calculus and Machine Learning.

Its derivative is:

$$
\boxed{
\frac{d}{dx}(\ln x)=\frac1x
}
$$

for:

$$
x>0
$$

# 8.4.1 Why Is $\ln(x)$ Special?

There is a beautiful relationship between:

$$
e^x
$$

and:

$$
\ln x
$$

They are inverse functions.

That means:

$$
\ln(e^x)=x
$$

and:

$$
e^{\ln x}=x
$$

Because of this inverse relationship, their derivatives are closely connected.

We already know:

$$
\frac{d}{dx}(e^x)=e^x
$$

and the corresponding logarithmic derivative is:

$$
\boxed{
\frac{d}{dx}(\ln x)=\frac1x
}
$$

### Important domain

For real-valued $\ln x$:

$$
\boxed{x>0}
$$

Therefore, $\ln x$ is not defined as a real-valued function for:

$$
x\leq0
$$

and its derivative is consequently considered on:

$$
x>0
$$

# 8.4.2 Derivative of a Logarithm With Another Base

For:

$$
f(x)=\log_a x
$$

where:

$$
a>0,\qquad a\neq1
$$

the derivative is:

$$
\boxed{
\frac{d}{dx}(\log_a x)
=
\frac{1}{x\ln a}
}
$$

When:

$$
a=e
$$

we have:

$$
\ln e=1
$$

so:

$$
\frac{d}{dx}(\log_e x)
=
\frac1x
$$

which gives:

$$
\boxed{
\frac{d}{dx}(\ln x)=\frac1x
}
$$

# 8.4.3 Examples — Logarithmic Functions

### Example 1

$$
f(x)=\ln x
$$

Therefore:

$$
\boxed{
f'(x)=\frac1x
}
$$

---

### Example 2

$$
f(x)=3\ln x
$$

Using the Constant Multiple Rule:

$$
f'(x)
=
3\left(\frac1x\right)
$$

Therefore:

$$
\boxed{
f'(x)=\frac3x
}
$$

---

### Example 3

$$
f(x)=\log_{10}x
$$

Using:

$$
\frac{d}{dx}(\log_a x)
=
\frac{1}{x\ln a}
$$

we get:

$$
\boxed{
f'(x)=\frac{1}{x\ln10}
}
$$

# 8.5 Trigonometric Functions

Trigonometric functions describe relationships involving angles and are fundamental in mathematics, physics, engineering, and many areas of applied mathematics.

The three most important basic trigonometric derivatives are:

$$
\boxed{
\frac{d}{dx}(\sin x)=\cos x
}
$$

$$
\boxed{
\frac{d}{dx}(\cos x)=-\sin x
}
$$

$$
\boxed{
\frac{d}{dx}(\tan x)=\sec^2x
}
$$

These should eventually become familiar enough to recognize immediately.

# 8.5.1 Derivative of $\sin x$

For:

$$
f(x)=\sin x
$$

the derivative is:

$$
\boxed{
f'(x)=\cos x
}
$$

So:

$$
\sin x
\rightarrow
\cos x
$$

### Intuition

The sine function continuously rises and falls.

Its derivative tells us the instantaneous slope of that oscillating curve.

At:

$$
x=0
$$

we have:

$$
\sin0=0
$$

but:

$$
\cos0=1
$$

Therefore:

$$
f'(0)=1
$$

meaning that $\sin x$ has slope $1$ at $x=0$.

# 8.5.2 Derivative of $\cos x$

For:

$$
f(x)=\cos x
$$

the derivative is:

$$
\boxed{
f'(x)=-\sin x
}
$$

The negative sign is important.

So:

$$
\cos x
\rightarrow
-\sin x
$$

For example:

$$
\frac{d}{dx}(5\cos x)
=
-5\sin x
$$

### Important memory connection

The sine and cosine derivatives form a cycle:

$$
\sin x
\rightarrow
\cos x
\rightarrow
-\sin x
\rightarrow
-\cos x
\rightarrow
\sin x
$$

The pattern repeats after four differentiations.

# 8.5.3 Derivative of $\tan x$

For:

$$
f(x)=\tan x
$$

the derivative is:

$$
\boxed{
f'(x)=\sec^2x
}
$$

Therefore:

$$
\boxed{
\frac{d}{dx}(\tan x)=\sec^2x
}
$$

This derivative exists wherever $\tan x$ itself is defined.

Since:

$$
\tan x=\frac{\sin x}{\cos x}
$$

we require:

$$
\cos x\neq0
$$

Therefore:

$$
x\neq
\frac{\pi}{2}+k\pi,
\qquad
k\in\mathbb Z
$$

### Important

The derivative of $\tan x$ is **not** $\tan^2x$.

It is:

$$
\boxed{\sec^2x}
$$

# 8.5.4 Reciprocal Trigonometric Functions

There are three additional trigonometric functions that are useful to know.

### Cotangent

$$
\cot x=\frac{\cos x}{\sin x}
$$

Derivative:

$$
\boxed{
\frac{d}{dx}(\cot x)
=
-\csc^2x
}
$$

---

### Secant

$$
\sec x=\frac1{\cos x}
$$

Derivative:

$$
\boxed{
\frac{d}{dx}(\sec x)
=
\sec x\tan x
}
$$

---

### Cosecant

$$
\csc x=\frac1{\sin x}
$$

Derivative:

$$
\boxed{
\frac{d}{dx}(\csc x)
=
-\csc x\cot x
}
$$

These are less frequently encountered in introductory ML mathematics than $\sin$, $\cos$, and $\tan$, but they are important members of the standard derivative library.

Some derivatives occur so frequently that they should become almost automatic.

### 1. Derivative of $x$

$$

$$
\boxed{
\frac{d}{dx}(x)=1

$$

---

### 2. Derivative of a constant

$$
\boxed{
\frac{d}{dx}(c)=0
}
$$

---

### 3. Derivative of $x^n$

$$
\boxed{
\frac{d}{dx}(x^n)=nx^{n-1}
}
$$

---

### 4. Derivative of $e^x$

$$
\boxed{
\frac{d}{dx}(e^x)=e^x
}
$$

---

### 5. Derivative of $\ln x$

$$
\boxed{
\frac{d}{dx}(\ln x)=\frac1x
}
$$

---

### 6. Derivative of $\sin x$

$$
\boxed{
\frac{d}{dx}(\sin x)=\cos x
}
$$

---

### 7. Derivative of $\cos x$

$$
\boxed{
\frac{d}{dx}(\cos x)=-\sin x
}
$$

---

### 8. Derivative of $\tan x$

$$
\boxed{
\frac{d}{dx}(\tan x)=\sec^2x
}
$$

# 8.6.1 Important Domain Restrictions

A derivative formula must always be considered together with the domain where it is valid.

For example:

### $\ln x$

$$
\frac{d}{dx}(\ln x)=\frac1x
$$

requires:

$$
x>0
$$

---

### $\frac1x$

Rewrite:

$$
\frac1x=x^{-1}
$$

Then:

$$
\frac{d}{dx}\left(\frac1x\right)
=
-\frac1{x^2}
$$

but:

$$
x\neq0
$$

---

### $\sqrt{x}$

$$
\frac{d}{dx}(\sqrt{x})
=
\frac1{2\sqrt{x}}
$$

The real-valued derivative exists for:

$$
x>0
$$

even though the original function is defined at $x=0$.

---

### $\tan x$

$$
\frac{d}{dx}(\tan x)=\sec^2x
$$

requires:

$$
\cos x\neq0
$$

### Important lesson

> **Never memorize a derivative formula while ignoring where the function and derivative are defined.**

# 8.8 Connection to Machine Learning

These derivatives appear directly in Machine Learning.

## Exponential Functions

The exponential function is fundamental to models involving probabilities and scores.

For example, the exponential term:

$$
e^z
$$

appears in the **softmax** function and other probabilistic models.

Its derivative is especially convenient:

$$
\frac{d}{dz}e^z=e^z
$$

---

## Logarithms

The natural logarithm appears throughout ML, especially in:

- log-likelihoods
- cross-entropy
- maximum likelihood estimation
- probabilistic models

The basic derivative:

$$
\frac{d}{dx}\ln x=\frac1x
$$

is therefore extremely important.

---

## Polynomial Functions

Polynomial expressions appear frequently in:

- regression
- loss functions
- regularization
- Taylor approximations

For example:

$$
L(w)=w^2
$$

gives:

$$
\frac{dL}{dw}=2w
$$

---

## Trigonometric Functions

Trigonometric functions are less central to basic supervised learning than exponentials and logarithms, but they occur in:

- signal processing
- Fourier analysis
- positional representations
- scientific ML
- neural-network architectures involving periodic representations

So these derivatives remain part of the standard calculus toolkit.

# 8.8.1 Why These Derivatives Matter for ML

Machine Learning often asks questions such as:

> If I change a parameter slightly, how does the output or loss change?

Suppose:

$$
L(w)=e^w+\ln w+w^2
$$

for:

$$
w>0
$$

Using the derivatives we have learned:

$$
\frac{d}{dw}(e^w)=e^w
$$

$$
\frac{d}{dw}(\ln w)=\frac1w
$$

and:

$$
\frac{d}{dw}(w^2)=2w
$$

Therefore:

$$
\boxed{
\frac{dL}{dw}
=
e^w+\frac1w+2w
}
$$

This is a simple example of how a complex-looking ML expression can be broken into familiar pieces.

Later, the **chain rule** will allow us to differentiate expressions where these functions are nested inside one another, such as:

$$
e^{wx+b}
$$

or:

$$
\ln(1+e^z)
$$

Those patterns are extremely important in Machine Learning.

# 8.10 Key Takeaways

We now have a basic derivative library.

### Polynomial

Use the rules from Part 7:

$$
\boxed{
\frac{d}{dx}(x^n)=nx^{n-1}
}
$$

and differentiate polynomial terms independently.

### Exponential

$$
\boxed{
\frac{d}{dx}(e^x)=e^x
}
$$

and more generally:

$$
\boxed{
\frac{d}{dx}(a^x)=a^x\ln a
}
$$

### Logarithmic

$$
\boxed{
\frac{d}{dx}(\ln x)=\frac1x
}
$$

and:

$$
\boxed{
\frac{d}{dx}(\log_a x)
=
\frac{1}{x\ln a}
}
$$

### Trigonometric

$$
\boxed{
\frac{d}{dx}(\sin x)=\cos x
}
$$

$$
\boxed{
\frac{d}{dx}(\cos x)=-\sin x
}
$$

$$
\boxed{
\frac{d}{dx}(\tan x)=\sec^2x
}
$$

and the reciprocal-function derivatives can be added to the standard toolkit.

### The most important mental model

> **These are not isolated formulas. They are basic building blocks that we will combine using differentiation rules to handle increasingly complicated functions.**

In Machine Learning, the derivatives of **polynomials, exponentials, and logarithms** are especially important because they appear repeatedly in models, probability, loss functions, and optimization.
