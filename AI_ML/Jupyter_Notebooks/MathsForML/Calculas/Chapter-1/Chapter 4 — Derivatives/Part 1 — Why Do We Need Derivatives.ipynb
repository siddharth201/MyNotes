# Chapter 3 — Derivatives

## Part 1 — Why Do We Need Derivatives?

In the previous chapters, we learned about **functions** and **limits**.

A function tells us how one quantity depends on another:

$$
y=f(x)
$$

A limit helps us understand what happens when an input gets arbitrarily close to a particular value.

Now we are ready for a new and very important question:

> **How quickly is a quantity changing at a particular moment or point?**

For example:

- How fast is a car moving **right now**?
- How quickly is the temperature increasing **at this moment**?
- How quickly is a company's revenue changing **at a particular time**?
- How sensitive is a model's prediction to a small change in an input?
- How does a machine-learning loss change when we slightly change a model parameter?

The mathematical concept that answers these questions is the **derivative**.

At a high level:

$$
\boxed{
\text{Derivative}
=
\text{Instantaneous Rate of Change}
}
$$

But rather than memorizing this definition, we will build the idea from the ground up.

The journey will be:

$$
\text{Change}
\rightarrow
\text{Rate of Change}
\rightarrow
\text{Average Rate of Change}
\rightarrow
\text{Instantaneous Rate of Change}
\rightarrow
\text{Derivative}
$$

## 3.1 What Does "Change" Mean?

The idea of a derivative begins with something very simple:

> **Change means that the value of a quantity is different at two different points.**

Suppose a quantity changes from:

$$
60
$$

to:

$$
65
$$

The amount of change is:

$$
65-60=5
$$

In mathematics, the Greek capital letter delta,

$$
\Delta
$$

is commonly used to represent a **finite change**.

Therefore:

$$
\boxed{
\Delta y = y_2-y_1
}
$$

means:

> Change in $y$ = final value of $y$ − initial value of $y$.

Similarly:

$$
\boxed{
\Delta x = x_2-x_1
}
$$

means:

> Change in $x$ = final value of $x$ − initial value of $x$.

So the fundamental idea is:

$$
\boxed{
\text{Change}
=
\text{Final Value}
-
\text{Initial Value}
}
$$

## 3.2 Change Alone Does Not Tell the Whole Story

Knowing how much something changed is useful, but often it is not enough.

Consider two cars.

Both travel:

$$
100\text{ km}
$$

But:

- Car A takes $2$ hours.
- Car B takes $5$ hours.

Both cars experienced the same change in distance:

$$
\Delta d=100\text{ km}
$$

But clearly, the two journeys were very different.

The important question is:

> **How quickly did the change happen?**

This leads to the idea of a **rate of change**.

A rate tells us how much one quantity changes relative to another quantity.

In general:

$$
\boxed{
\text{Rate of Change}
=
\frac{\text{Change in Output}}
{\text{Change in Input}}
}
$$

For a journey:

$$
\boxed{
\text{Average Speed}
=
\frac{\text{Change in Distance}}
{\text{Change in Time}}
}
$$

or:

$$
\boxed{
\text{Average Speed}
=
\frac{\Delta d}{\Delta t}
}
$$

This simple idea of "change per unit change" is the foundation on which derivatives are built.

## 3.3 Average Rate of Change

Now connect the idea of rate of change to a mathematical function.

Suppose:

$$
y=f(x)
$$

We choose two input values:

$$
x_1
\quad\text{and}\quad
x_2
$$

Their corresponding outputs are:

$$
f(x_1)
\quad\text{and}\quad
f(x_2)
$$

The change in the input is:

$$
\Delta x=x_2-x_1
$$

The change in the output is:

$$
\Delta y=f(x_2)-f(x_1)
$$

Therefore, the average rate of change is:

$$
\boxed{
\frac{\Delta y}{\Delta x}
=
\frac{f(x_2)-f(x_1)}
{x_2-x_1}
}
$$

In simple language:

> **Average rate of change tells us how much the output changes, on average, for each unit of input change over an interval.**

The word **average** is important.

We are looking at what happened across an interval, not what happened at one particular point.

## 3.4 Example — Average Rate of Change

Consider the function:

$$
f(x)=x^2
$$

Let's calculate its average rate of change between:

$$
x=1
$$

and:

$$
x=3
$$

First calculate the two output values:

$$
f(1)=1^2=1
$$

$$
f(3)=3^2=9
$$

Therefore:

$$
\Delta y=9-1=8
$$

and:

$$
\Delta x=3-1=2
$$

Hence:

$$
\frac{\Delta y}{\Delta x}
=
\frac{8}{2}
=
4
$$

Therefore:

$$
\boxed{
\text{Average Rate of Change}=4
}
$$

This means that, over the interval from $x=1$ to $x=3$, the output changed by an average of $4$ units for every $1$ unit of input change.

Notice the word **average**.

It does not mean that the function was changing at exactly $4$ units per input unit at every point between $1$ and $3$.

## 3.5 A Real-World Example — Distance and Time

Suppose the position of a car is described by:

$$
s(t)=t^2
$$

where:

- $t$ is time in seconds
- $s(t)$ is position in meters

Consider the interval:

$$
t=2
\quad\text{to}\quad
t=4
$$

The positions are:

$$
s(2)=4
$$

and:

$$
s(4)=16
$$

Therefore:

$$
\Delta s=16-4=12
$$

and:

$$
\Delta t=4-2=2
$$

The average rate of change of position is:

$$
\frac{\Delta s}{\Delta t}
=
\frac{12}{2}
=
6
$$

So:

$$
\boxed{
\text{Average velocity}=6\text{ m/s}
}
$$

The important word here is **average**.

This tells us the average velocity over the entire interval from $t=2$ to $t=4$.

It does not tell us the velocity at exactly $t=2$, $t=3$, or $t=4$.

That limitation leads us to the next question.

## 3.6 The Problem With Average Rate of Change

Suppose you are driving a car.

You know that over a particular journey your average speed was:

$$
60\text{ km/h}
$$

Can you conclude that your speed was exactly $60$ km/h at every moment?

No.

You might have:

- accelerated from $20$ km/h to $80$ km/h,
- slowed down near a traffic signal,
- stopped temporarily,
- accelerated again.

The average summarizes the entire interval.

But sometimes we need a much more precise question:

> **What is happening right now?**

For example:

> What is the car's speed at exactly $10{:}30$ AM?

This is an **instantaneous** question.

Therefore, we need to distinguish between:

$$
\boxed{
\text{Average Rate of Change}
}
$$

and:

$$
\boxed{
\text{Instantaneous Rate of Change}
}
$$

### Average rate of change

Looks at an interval:

$$
x_1\rightarrow x_2
$$

### Instantaneous rate of change

Looks at one particular point:

$$
x=a
$$

The derivative is the mathematical tool that gives us this instantaneous rate of change.

## 3.7 Why Is Instantaneous Rate of Change Difficult?

There is a subtle mathematical problem.

The average rate of change requires two points:

$$
\frac{f(x_2)-f(x_1)}
{x_2-x_1}
$$

But an instantaneous rate of change asks about one point.

For example:

> What is the rate of change of $f(x)$ exactly at $x=2$?

We only have one point:

$$
x=2
$$

But a slope or rate of change requires a comparison.

So how can we measure change at one point?

The key idea is:

> **Use another point extremely close to the point we care about.**

Suppose our point of interest is:

$$
x
$$

and choose a nearby point:

$$
x+h
$$

where $h$ represents a small change in the input.

Now we have two points:

$$
x
\quad\text{and}\quad
x+h
$$

Their corresponding outputs are:

$$
f(x)
\quad\text{and}\quad
f(x+h)
$$

The rate of change between these two nearby points is:

$$
\frac{f(x+h)-f(x)}
{(x+h)-x}
$$

Since:

$$
(x+h)-x=h
$$

we get:

$$
\boxed{
\frac{f(x+h)-f(x)}
{h}
}
$$

This expression is called the **difference quotient**.

At this stage, it is still an average rate of change.

We have simply made the interval smaller.

## 3.8 What Happens When We Make the Interval Smaller?

Suppose we want to understand the behavior of a function around:

$$
x=2
$$

We can choose another point nearby.

For example:

$$
x=2.5
$$

Then:

$$
h=0.5
$$

We can move even closer:

$$
x=2.1
$$

so:

$$
h=0.1
$$

Closer:

$$
x=2.01
$$

so:

$$
h=0.01
$$

And closer:

$$
x=2.001
$$

so:

$$
h=0.001
$$

The second point keeps approaching $x=2$.

Conceptually:

$$
\boxed{
h\rightarrow0
}
$$

We are asking:

> **What does the average rate of change approach as the two points get arbitrarily close to each other?**

This is exactly the type of question that a **limit** allows us to answer.

Therefore, the limit connects our previous chapter on Limits to our current chapter on Derivatives.

```python
def f(x):
    return x**2

x = 2

h_values = [1, 0.5, 0.1, 0.01, 0.001, 0.0001]

for h in h_values:
    average_rate = (f(x + h) - f(x)) / h
    print(
        f"h = {h:<8} "
        f"average rate of change = {average_rate}"
    )
```

## 3.9 Numerical Experiment — Getting Closer to an Instantaneous Rate

For:

$$
f(x)=x^2
$$

let us examine the average rate of change at:

$$
x=2
$$

using increasingly smaller values of $h$.

The difference quotient is:

$$
\frac{f(2+h)-f(2)}
{h}
$$

As $h$ becomes smaller, the calculated average rate approaches:

$$
4
$$

For example, we obtain values such as:

$$
5,\quad
4.5,\quad
4.1,\quad
4.01,\quad
4.001,\ldots
$$

The values are approaching:

$$
4
$$

This suggests that the instantaneous rate of change at $x=2$ is:

$$
\boxed{4}
$$

But notice something important:

We did not simply choose a tiny value of $h$ and declare that to be the exact answer.

Instead, we are interested in the value that the expression **approaches as $h$ approaches zero**.

That distinction is fundamental.

The exact mathematical definition will come from the limit.

## 3.10 From Average Rate of Change to Instantaneous Rate of Change

We started with the average rate of change:

$$
\frac{f(x_2)-f(x_1)}
{x_2-x_1}
$$

Then we chose two nearby points:

$$
x
\quad\text{and}\quad
x+h
$$

giving:

$$
\frac{f(x+h)-f(x)}
{h}
$$

This is still an average rate of change.

Now we let the distance between the two points become arbitrarily small:

$$
h\rightarrow0
$$

Using a limit:

$$
\boxed{
\lim_{h\rightarrow0}
\frac{f(x+h)-f(x)}
{h}
}
$$

This limiting value is called the **derivative** of $f$ at $x$.

Therefore:

$$
\boxed{
f'(x)
=
\lim_{h\rightarrow0}
\frac{f(x+h)-f(x)}
{h}
}
$$

This is called the **limit definition of the derivative** or the **derivative from first principles**.

The important conceptual chain is:

$$
\boxed{
\text{Average Rate of Change}
\overset{h\rightarrow0}{\longrightarrow}
\text{Instantaneous Rate of Change}
}
$$

and:

$$
\boxed{
\text{Instantaneous Rate of Change}
=
\text{Derivative}
}
$$

## 3.11 Why Don't We Simply Put $h=0$?

A natural question is:

> If we want $h$ to become zero, why don't we simply substitute $h=0$?

Consider:

$$
\frac{f(x+h)-f(x)}
{h}
$$

If we directly put:

$$
h=0
$$

we get:

$$
\frac{f(x)-f(x)}
{0}
=
\frac{0}{0}
$$

which is undefined.

So the derivative is **not** obtained by simply setting $h=0$.

Instead, we ask:

> **What value does the expression approach as $h$ gets arbitrarily close to zero?**

That is a limit.

Therefore:

$$
\boxed{
h\rightarrow0
\neq
h=0
}
$$

This is one of the most important ideas connecting **limits** and **derivatives**.

We will explore this carefully in Part 2.

## 3.12 The Geometric Preview

There is another way to understand what we have just done.

Take two points on a curve.

The straight line passing through those two points is called a **secant line**.

The slope of that line represents the average rate of change:

$$
\boxed{
\text{Average Rate of Change}
=
\text{Secant Slope}
}
$$

Now move the second point closer and closer to the first point.

As:

$$
h\rightarrow0
$$

the secant line approaches a limiting line called the **tangent line**.

The slope of this tangent line represents the instantaneous rate of change:

$$
\boxed{
\text{Instantaneous Rate of Change}
=
\text{Tangent Slope}
}
$$

Therefore:

$$
\boxed{
\text{Derivative}
=
\text{Slope of the Tangent Line}
}
$$

This geometric interpretation will be studied properly in:

**Part 3 — Geometric Meaning of the Derivative.**

## 3.13 Derivative as Local Change

The word **local** is extremely important.

A derivative does not describe the behavior of an entire function at once.

Instead, it tells us what is happening **around a particular point**.

For example, consider:

$$
f(x)=x^2
$$

Its behavior changes depending on where we look.

Near:

$$
x=1
$$

the function has one local rate of change.

Near:

$$
x=10
$$

it has a very different local rate of change.

Therefore, the derivative gives us a way to ask:

> **What is the behavior of this function right here?**

This local perspective is one of the reasons derivatives are so powerful.

In Machine Learning, we repeatedly make decisions based on exactly this kind of local information.

## 3.14 Why Derivatives Matter in Machine Learning

Now we can connect the idea directly to Machine Learning.

Suppose a model has a parameter:

$$
w
$$

and the model's loss is:

$$
L(w)
$$

During training, our goal is to find values of $w$ that make the loss small:

$$
\min_w L(w)
$$

To improve the model, we need to understand how the loss changes when we change the parameter.

The important question becomes:

> **If I change $w$ slightly, what happens to the loss?**

The derivative:

$$
\boxed{
\frac{dL}{dw}
}
$$

measures the local rate of change of the loss with respect to $w$.

In other words:

> **It tells us how sensitive the loss is to a small change in the parameter.**

For example, if:

$$
\frac{dL}{dw}>0
$$

then, locally, increasing $w$ causes the loss to increase.

If:

$$
\frac{dL}{dw}<0
$$

then, locally, increasing $w$ causes the loss to decrease.

This directional information is extremely useful for optimization.

Later, when a model has many parameters, this idea develops into:

$$
\text{Partial Derivatives}
\rightarrow
\text{Gradient}
\rightarrow
\text{Gradient Descent}
\rightarrow
\text{Backpropagation}
$$

So derivatives are not merely a topic from Calculus.

They are one of the mathematical foundations of how many Machine Learning models are trained.

## 3.15 A Simple ML Intuition

Imagine that the loss function looks like a landscape.

The horizontal direction represents a model parameter:

$$
w
$$

The vertical direction represents the loss:

$$
L(w)
$$

You are standing somewhere on this landscape.

You want to reach a low point.

A derivative gives you local information about the landscape:

> **If I take a small step in this direction, will I go uphill or downhill?**

If:

$$
\frac{dL}{dw}>0
$$

the loss is increasing as $w$ increases.

If:

$$
\frac{dL}{dw}<0
$$

the loss is decreasing as $w$ increases.

Therefore, the derivative gives us a local sense of the direction in which the loss is changing.

This is the basic intuition behind gradient-based optimization.

We will study the optimization process itself later rather than jumping into Gradient Descent here.

## 3.16 The Three Core Interpretations of a Derivative

A derivative can be understood in three closely connected ways.

### 1. Instantaneous Rate of Change

It tells us:

> How quickly is the output changing at this particular point?

$$
\boxed{
f'(x)
=
\text{Instantaneous Rate of Change}
}
$$

---

### 2. Geometric Slope

It tells us:

> What is the slope of the tangent line at this point?

$$
\boxed{
f'(x)
=
\text{Tangent Slope}
}
$$

---

### 3. Local Sensitivity

It tells us:

> How sensitive is the output to a small change in the input?

$$
\boxed{
f'(x)
=
\text{Local Sensitivity}
}
$$

These are not three different mathematical concepts.

They are three different ways of understanding the same derivative.

Therefore:

$$
\boxed{
\text{Rate of Change}
=
\text{Tangent Slope}
=
\text{Local Sensitivity}
}
$$

## 3.17 The Complete Intuition

The entire motivation for derivatives can now be summarized as a sequence.

### Step 1 — Something changes

A quantity moves from one value to another:

$$
\Delta y
$$

### Step 2 — We want to know how quickly it changes

So we compare output change with input change:

$$
\frac{\Delta y}{\Delta x}
$$

### Step 3 — This gives an average rate

We are looking across an interval:

$$
x_1\rightarrow x_2
$$

### Step 4 — But sometimes we need the rate at one particular point

For example:

> What is the speed right now?

### Step 5 — Use a second point very close to the first

$$
x
\quad\text{and}\quad
x+h
$$

giving:

$$
\frac{f(x+h)-f(x)}
{h}
$$

### Step 6 — Make the distance between the points arbitrarily small

$$
h\rightarrow0
$$

### Step 7 — Use a limit

$$
\boxed{
f'(x)
=
\lim_{h\rightarrow0}
\frac{f(x+h)-f(x)}
{h}
}
$$

And that gives us the instantaneous rate of change.

Therefore:

$$
\boxed{
\text{Change}
\rightarrow
\text{Rate}
\rightarrow
\text{Average Rate}
\rightarrow
\text{Local Rate}
\rightarrow
\text{Limit}
\rightarrow
\text{Derivative}
}
$$

## 3.18 Why This Foundation Matters for Machine Learning

The same pattern appears repeatedly in Machine Learning.

A Machine Learning model is fundamentally a mathematical function:

$$
\hat{y}=f(\mathbf{x};\mathbf{w})
$$

where:

- $\mathbf{x}$ represents the input features.
- $\mathbf{w}$ represents the model parameters.
- $\hat{y}$ represents the prediction.

The model also has a loss:

$$
L(\mathbf{w})
$$

Training asks:

> **How should we change the parameters so that the loss decreases?**

To answer that, we need to understand how the loss changes when parameters change.

For one parameter:

$$
\frac{dL}{dw}
$$

For multiple parameters, we will eventually use partial derivatives:

$$
\frac{\partial L}{\partial w_1},
\frac{\partial L}{\partial w_2},
\ldots,
\frac{\partial L}{\partial w_n}
$$

These partial derivatives are combined into the gradient:

$$
\boxed{
\nabla L
}
$$

which then becomes central to optimization algorithms such as Gradient Descent.

So our Calculus journey has a very deliberate purpose:

$$
\boxed{
\text{Derivative}
\rightarrow
\text{Partial Derivatives}
\rightarrow
\text{Gradient}
\rightarrow
\text{Optimization}
}
$$

This is why understanding the intuition behind derivatives is far more valuable than simply memorizing differentiation formulas.
