# Chapter 1 — Functions

Before we can understand **limits, derivatives, gradients, or optimization**, we need to understand the object that Calculus studies.

That object is often a **function**.

And this is especially important for Machine Learning because, at its core, an ML model is also a function.

---

# 1.1 Why Do We Need Functions?

Let's start with a simple problem.

Suppose you go to a shop and buy apples.

If one apple costs ₹20, then:

* 1 apple → ₹20
* 2 apples → ₹40
* 3 apples → ₹60
* 5 apples → ₹100

There is a relationship between:

> **Number of apples → Total cost**

We can describe it mathematically:

$$
\text{Cost}=20\times\text{Number of apples}
$$

Let:

$$
x=\text{number of apples}
$$

and:

$$
y=\text{total cost}
$$

Then:

$$
y=20x
$$

This equation describes a **relationship between two quantities**.

A function gives us a systematic way to describe such relationships.

---

# 1.2 What Is a Function?

A simple definition is:

> **A function is a rule that takes an input and produces exactly one output.**

We can visualize it as a machine:

```text
             FUNCTION
                │
                │
Input ─────────►│─────────► Output
                │
                │
              Rule
```

For example:

```text
Input
  2
  │
  ▼
┌─────────────┐
│ Multiply by │
│      3      │
└─────────────┘
  │
  ▼
Output
  6
```

So:

$$
f(x)=3x
$$

If:

$$
x=2
$$

then:

$$
f(2)=3(2)=6
$$

---

# 1.3 The Most Important Rule

There is one property that makes something a **function**:

> **Every valid input must correspond to exactly one output.**

For example:

$$
f(x)=x^2
$$

If:

$$
x=3
$$

then:

$$
f(3)=9
$$

There is only one output:

$$
3\rightarrow9
$$

That's valid.

But imagine a rule where:

$$
x=3
$$

could produce both:

$$
9
$$

and:

$$
10
$$

Then it would **not** be a function.

### Important distinction

A function **can have multiple inputs producing the same output**.

For example:

$$
f(x)=x^2
$$

gives:

$$
f(2)=4
$$

and:

$$
f(-2)=4
$$

That's perfectly valid.

So:

```text
Different inputs
      │
      ├──────► same output
      │
      └──────►
```

is allowed.

But:

```text
One input
    │
    ├──────► output A
    │
    └──────► output B
```

is not a function.

---

# 1.4 Function Notation

Instead of repeatedly saying:

> "The output produced by the function when the input is x"

we use:

$$
f(x)
$$

Read this as:

> **"f of x"**

It means:

> the output of function $f$ when the input is $x$.

For example:

$$
f(x)=2x+1
$$

If:

$$
x=3
$$

then:

$$
f(3)=2(3)+1
$$

$$
f(3)=7
$$

So:

```text
Input              Output
  3      ──f──►      7
```

---

# 1.5 A Function Is Not the Same Thing as $f(x)$

This distinction is worth understanding early.

Consider:

$$
f(x)=x^2
$$

Here:

**$f$** represents the function/rule.

**$x$** represents the input variable.

**$f(x)$** represents the output produced for input $x$.

So when we write:

$$
f(3)=9
$$

we are evaluating the function at the specific input $3$.

Think of it like:

```text
f        → the machine/rule
x        → input
f(x)     → output for that input
f(3)     → output when input = 3
```

This distinction becomes very important later when we differentiate:

$$
f'(x)
$$

because then we will be asking:

> **How does the output of the function change as its input changes?**

---

# 1.6 Function as a Machine

Let's make the idea even more intuitive.

Suppose we have:

$$
f(x)=x^2+1
$$

Think of the function as a machine:

```text
             ┌─────────────────┐
      x ────►│ Square the input │
             │       + 1        │
             └────────┬────────┘
                      │
                      ▼
                    f(x)
```

For:

$$
x=2
$$

the machine performs:

$$
2
\rightarrow
2^2
\rightarrow
4
\rightarrow
4+1
\rightarrow
5
$$

Therefore:

$$
f(2)=5
$$

For:

$$
x=5
$$

we get:

$$
5
\rightarrow
25
\rightarrow
26
$$

Therefore:

$$
f(5)=26
$$

The rule stays the same.

Only the input changes.

---

# 1.7 Function as a Mapping

There is another mathematical way to describe a function.

A function **maps** elements from one set to another.

For example:

$$
f:A\rightarrow B
$$

This means:

> Function $f$ maps elements of set $A$ to elements of set $B$.

For example:

$$
A=\{1,2,3\}
$$

and:

$$
B=\{2,4,6\}
$$

Suppose:

$$
f(x)=2x
$$

Then:

$$
1\rightarrow2
$$

$$
2\rightarrow4
$$

$$
3\rightarrow6
$$

So:

```text
A                       B

1 ───────────────────► 2

2 ───────────────────► 4

3 ───────────────────► 6
```

This mapping perspective will become very useful later when we discuss **vectors, multivariable functions, and ML models**.

---

# 1.8 Domain and Range

Two terms are extremely important.

## Domain

The **domain** is the set of inputs that are allowed.

## Range

The **range** is the set of outputs that the function actually produces.

For example:

$$
f(x)=x^2
$$

If we allow all real numbers as inputs:

$$
x\in\mathbb{R}
$$

then the domain is:

$$
\boxed{\mathbb{R}}
$$

What outputs can we get?

Since:

$$
x^2\geq0
$$

we can never get a negative output.

Therefore:

$$
\boxed{\text{Range}=[0,\infty)}
$$

---

# 1.9 Why Does Domain Matter?

Consider:

$$
f(x)=\frac{1}{x}
$$

Can we put:

$$
x=2
$$

Yes:

$$
f(2)=\frac12
$$

Can we put:

$$
x=0
$$

No.

Because:

$$
\frac{1}{0}
$$

is undefined.

Therefore:

$$
x\neq0
$$

is part of the domain restriction.

This will become important later when studying **limits and continuity**.

---

# 1.11 Functions and Graphs

For a single-variable function:

$$ y=f(x) $$

we can visualize the relationship between the input $x$ and the output $f(x)$ using a coordinate plane.

The horizontal axis represents the input:

$$ x $$

The vertical axis represents the output:

$$ y=f(x) $$

For example:

$$ f(x)=x^2 $$
**Jupyter Visualization**

````python
import numpy as np
import matplotlib.pyplot as plt

# Input values
x = np.linspace(-3, 3, 400)

# Function
f_x = x**2

# Create graph
plt.figure(figsize=(8, 5))
plt.plot(x, f_x)

# Coordinate axes
plt.axhline(0)
plt.axvline(0)

# Labels
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Graph of f(x) = x²")

# Grid
plt.grid(True)

plt.show()
````

What Does the Graph Tell Us?

The graph gives us a visual representation of the function.

For example:

$$ f(-2)=4 $$ $$ f(0)=0 $$ $$ f(2)=4 $$

So the graph shows us:

> **How the output changes as the input changes.**

This visual interpretation will become extremely important when we study Calculus.

Later, instead of merely asking:

> **"What is the value of $f(x)$?"**

we will ask:

> **"How quickly is $f(x)$ changing as $x$ changes?"**

That question leads directly to the derivative.




# 1.12 Function vs Equation

These are related but not identical concepts.

An **equation** states that two expressions are equal:

$$
y=2x+1
$$

A **function** describes a rule mapping input to output:

$$
f(x)=2x+1
$$

We can represent the same relationship using either notation, but the function notation makes the **input-output relationship explicit**.

This distinction becomes especially useful in ML.

---

# 1.13 Functions in Machine Learning

Now let's connect this to ML.

Suppose we build a simple model that predicts house prices.

The model might take:

* house size
* number of bedrooms
* location score
* age of house

and produce:

> predicted price

Conceptually:

```text
House Features
      │
      ▼
┌──────────────┐
│ ML Model     │
│   Function   │
└──────┬───────┘
       │
       ▼
Predicted Price
```

Mathematically:

$$
\hat y=f(\vec{x})
$$

where:

$$
\vec{x}
=
\begin{bmatrix}
x_1\\
x_2\\
\vdots\\
x_n
\end{bmatrix}
$$

represents the input features.

For example:

$$
x_1=\text{house size}
$$

$$
x_2=\text{number of bedrooms}
$$

and so on.

The model is a function that maps:

$$
\boxed{\text{Features}\rightarrow\text{Prediction}}
$$

---

# 1.14 ML Model as a Function

Consider Linear Regression.

A simple one-feature model is:

$$
\hat y=wx+b
$$

This is a function.

We could write:

$$
f(x)=wx+b
$$

Here:

* $x$ = input feature
* $w$ = model parameter
* $b$ = bias
* $\hat y$ = prediction

So:

$$
\boxed{\hat y=f(x)}
$$

The ML model takes an input and produces an output.

That's exactly the basic idea of a function.

---

# 1.15 But ML Functions Are Usually More Interesting

In real ML:

$$
\hat y=f(\vec{x};\vec{w})
$$

Now there are two important things:

### Input

$$
\vec{x}
$$

The data provided to the model.

### Parameters

$$
\vec{w}
$$

The values the model learns.

The semicolon is often used to distinguish them conceptually:

$$
f(\vec{x};\vec{w})
$$

meaning:

> The model takes input $\vec{x}$, and its behavior is controlled by parameters $\vec{w}$.

This distinction will become **extremely important** later.

---

# 1.16 Why Calculus Cares About Functions

Now we can connect this back to our introduction to Calculus.

We have a function:

$$
y=f(x)
$$

Calculus asks:

> **What happens to $y$ when $x$ changes?**

For example:

If $x$ changes slightly:

$$
x\rightarrow x+\Delta x
$$

what happens to:

$$
f(x)?
$$

That leads us toward:

$$
\frac{\Delta y}{\Delta x}
$$

and eventually:

$$
\boxed{\frac{dy}{dx}}
$$

—the derivative.

So our journey is beginning to connect:

```text
Function
   │
   │ "What happens when input changes?"
   ▼
Change
   │
   ▼
Rate of Change
   │
   ▼
Derivative
```

---

# 1.17 A Very Important ML Perspective

There are actually **two different things that can change** in an ML model.

Suppose:

$$
\hat y=f(x;w)
$$

We can change:

### The input

$$
x\rightarrow x+\Delta x
$$

and ask:

> How does the prediction change?

Or we can change:

### The parameter

$$
w\rightarrow w+\Delta w
$$

and ask:

> How does the prediction or loss change?

The second question becomes central to **training**.

Eventually we will study things like:

$$
\frac{\partial L}{\partial w}
$$

which asks:

> **How does the loss change when the model parameter $w$ changes?**

This is where Functions → Calculus → ML begins to come together.

---

# 1.18 Function Composition

One more idea is essential before we move on.

Suppose:

$$
g(x)=x+1
$$

and:

$$
f(x)=x^2
$$

We can put the output of $g$ into $f$:

$$
f(g(x))
$$

First:

$$
g(x)=x+1
$$

Then:

$$
f(g(x))
=
(x+1)^2
$$

This is called **composition of functions**.

Conceptually:

```text
x
│
▼
┌─────────┐
│   g     │
└────┬────┘
     │
     ▼
   g(x)
     │
     ▼
┌─────────┐
│   f     │
└────┬────┘
     │
     ▼
   f(g(x))
```

This might look like a small mathematical idea now.

But later it becomes extremely important because:

> **A neural network is essentially a composition of many functions.**

And that is why the **chain rule** becomes so important in Deep Learning.

---

# 1.19 The Function Perspective of a Neural Network

Very roughly:

$$
\vec{x}
\rightarrow
f_1
\rightarrow
f_2
\rightarrow
f_3
\rightarrow
\hat y
$$

Mathematically:

$$
\hat y
=
f_3(f_2(f_1(\vec{x})))
$$

So even a neural network can be viewed as one large function:

$$
\boxed{\hat y=F(\vec{x};\vec{w})}
$$

Later:

**Functions**

↓

**Composition**

↓

**Chain Rule**

↓

**Gradients**

↓

**Backpropagation**

This is one of the most important conceptual chains in our entire Calculus-for-ML journey.
