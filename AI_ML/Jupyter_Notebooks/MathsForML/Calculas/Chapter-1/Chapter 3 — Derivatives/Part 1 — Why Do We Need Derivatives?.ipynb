# Chapter 3 — Derivatives

## Part 1 — Why Do We Need Derivatives?

In the previous chapters, we learned:

$$
\text{Functions}
$$

and:

$$
\text{Limits}
$$

A function tells us how an output depends on an input.

A limit tells us what happens to the output when the input gets arbitrarily close to a particular value.

Now we want to answer a new question:

> **How quickly is a quantity changing at a particular point?**

This question appears everywhere:

- How fast is a car moving at this exact moment?
- How quickly is the temperature changing right now?
- How quickly is a company's revenue changing?
- How quickly does the output of a model change when an input changes?
- How quickly does a loss function change when a model parameter changes?

The mathematical tool for answering these questions is the **derivative**.

At a high level:

$$
\boxed{
\text{Derivative}
=
\text{instantaneous rate of change}
}
$$

But to understand why we need derivatives, we first need to understand the difference between:

$$
\boxed{\text{Average Rate of Change}}
$$

and:

$$
\boxed{\text{Instantaneous Rate of Change}}
$$
