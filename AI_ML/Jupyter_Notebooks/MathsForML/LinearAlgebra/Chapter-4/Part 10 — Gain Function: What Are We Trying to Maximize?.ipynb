# Chapter 4 — Part 10

# Gain Function: What Are We Trying to Maximize?

> **Core Question:**  
> If a positive class-adjusted score means correct classification, can we define a quantity that tells us how good our classifier is and then try to maximize it?

In Part 9, we arrived at the unified classification expression:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

We learned that:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
$$

means that the training example is correctly classified.

Now we want to go one step further.

A classifier should not merely classify examples correctly.

Ideally, it should classify them **confidently**.

This leads to a natural question:

> **Can we measure how strongly the classifier agrees with the true class?**

That is the motivation for introducing a gain function.

## 1. Correct Classification Is Not the Whole Story

Consider two positive examples.

Suppose both have:

$$
y_i=+1
$$

Now suppose their classifier scores are:

$$
\vec{w}^{\,T}\vec{x}_1+b=1
$$

and:

$$
\vec{w}^{\,T}\vec{x}_2+b=10
$$

Both examples are correctly classified because both scores are positive.

Their class-adjusted scores are:

$$
(+1)(1)=1
$$

and:

$$
(+1)(10)=10
$$

So we have:

$$
1>0
$$

and:

$$
10>0
$$

Both are correct.

But the second example has a much larger class-adjusted score.

This suggests that there is more information than simply:

> Correct or incorrect?

We can also ask:

> **How strongly does the classifier agree with the true class?**

## 2. From Correctness to Confidence

Recall the class-adjusted score:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

Its sign tells us whether the prediction is correct.

But its magnitude also gives us information.

For example:

$$
+1
$$

and:

$$
+10
$$

are both positive.

Therefore, both correspond to correct classification.

However:

$$
+10
$$

is a much larger positive value.

So we can think of the quantity:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

as measuring how strongly the classifier's score agrees with the true class.

A larger positive value represents stronger agreement.

This motivates the idea of a **gain**.

## 3. Define the Gain for One Example

For a training example $(\vec{x}_i,y_i)$, define its gain as:

$$
\boxed{
G_i
=
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

Here:

- $y_i$ is the true class label.
- $\vec{w}^{\,T}\vec{x}_i+b$ is the classifier score.
- $G_i$ is the resulting class-adjusted score.

The interpretation is:

$$
G_i>0
\quad\Rightarrow\quad
\text{correct classification}
$$

$$
G_i<0
\quad\Rightarrow\quad
\text{incorrect classification}
$$

and:

$$
G_i=0
\quad\Rightarrow\quad
\text{point lies on the decision boundary}
$$

A larger positive $G_i$ means stronger agreement between the classifier and the true class.

## 4. Why Call It "Gain"?

The word **gain** gives us an intuitive interpretation.

Think of:

$$
G_i
=
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

as asking:

> **How much positive value do we get after accounting for the true class?**

For a correctly classified example:

$$
G_i>0
$$

For an incorrectly classified example:

$$
G_i<0
$$

Therefore:

- positive gain → classifier agrees with the true class
- negative gain → classifier disagrees with the true class
- larger positive gain → stronger agreement

This gives us a natural quantity that we might want to maximize.

## 5. From One Example to the Whole Dataset

So far, we have defined gain for one training example:

$$
G_i
=
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

But a machine learning model is trained using an entire dataset.

Suppose our training set contains $m$ examples:

$$
(\vec{x}_1,y_1),
(\vec{x}_2,y_2),
\ldots,
(\vec{x}_m,y_m)
$$

We can therefore define the total gain across the training set as:

$$
\boxed{
G(\vec{w},b)
=
\sum_{i=1}^{m}
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

Now we have a single quantity that summarizes the class-adjusted scores of all training examples.

This leads to the natural optimization idea:

$$
\boxed{
\text{Choose } \vec{w},b
\text{ to maximize } G(\vec{w},b)
}
$$

## 6. The Intuition Behind Maximizing Gain

If:

$$
G(\vec{w},b)
=
\sum_{i=1}^{m}
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

is large and positive, we might expect the classifier to be doing well.

Why?

Because each correctly classified example contributes a positive value.

For example:

$$
+2,\quad +5,\quad +8
$$

all increase the total gain.

An incorrectly classified example contributes a negative value.

For example:

$$
-3
$$

reduces the total gain.

So, at first glance, maximizing the total gain seems like a reasonable objective.

We might therefore write:

$$
\boxed{
\max_{\vec{w},b}
\sum_{i=1}^{m}
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

But now we encounter an important problem.

## 7. 🚨 The Problem: Gain Can Be Made Arbitrarily Large

Suppose a particular classifier has parameters:

$$
\vec{w},b
$$

Now multiply both by a positive constant $c$:

$$
\vec{w}'=c\vec{w}
$$

and:

$$
b'=cb
$$

The new classifier score becomes:

$$
\vec{w}'^{\,T}\vec{x}_i+b'
$$

Substituting:

$$
(c\vec{w})^{T}\vec{x}_i+cb
$$

which gives:

$$
c\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

Therefore the new gain is:

$$
y_i
\left(
c\left(\vec{w}^{\,T}\vec{x}_i+b\right)
\right)
$$

which is:

$$
c\,y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

So the gain is multiplied by $c$.

## 8. But Did the Decision Boundary Change?

This is the surprising part.

The original decision boundary is:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

After scaling, it becomes:

$$
(c\vec{w})^{T}\vec{x}+cb=0
$$

which simplifies to:

$$
c\left(\vec{w}^{\,T}\vec{x}+b\right)=0
$$

For any positive $c$, this is equivalent to:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

Therefore:

$$
\boxed{
\text{The decision boundary has not changed.}
}
$$

The classifier is making exactly the same predictions.

Yet the gain has been multiplied by $c$.

This means we can make the gain larger simply by scaling $\vec{w}$ and $b$.

## 9. A Numerical Example of the Scaling Problem

Suppose a correctly classified example has:

$$
y_i=+1
$$

and:

$$
\vec{w}^{\,T}\vec{x}_i+b=5
$$

Therefore:

$$
G_i=(+1)(5)=5
$$

Now multiply both $\vec{w}$ and $b$ by $10$.

The new score becomes:

$$
10(5)=50
$$

Therefore:

$$
G_i'=50
$$

The gain increased from:

$$
5
$$

to:

$$
50
$$

But the decision boundary did not move.

So we have a problem:

$$
\boxed{
\text{Larger gain does not necessarily mean a better decision boundary.}
}
$$

We could keep increasing the scale:

$$
5
\rightarrow
50
\rightarrow
500
\rightarrow
5000
\rightarrow\cdots
$$

without changing the actual classifier boundary.

## 10. Why This Matters

We originally wanted to maximize gain because a larger positive gain seemed to represent stronger classification.

But we have discovered that:

$$
\vec{w},b
$$

can be scaled without changing the decision boundary.

At the same time:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

can become arbitrarily large.

Therefore, the optimization problem:

$$
\boxed{
\max_{\vec{w},b}
\sum_{i=1}^{m}
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

is not well-defined as a way to choose the best separating boundary.

There is no finite maximum.

We can always increase the objective by scaling $\vec{w}$ and $b$.

This is the key limitation we have uncovered.

## 11. What Are We Really Trying to Maximize?

This forces us to refine our original question.

We do not actually want:

> "Make the mathematical score as large as possible."

Instead, we want something more meaningful:

> **Find a decision boundary that classifies the training examples correctly and places them confidently away from the boundary.**

In other words, we care about the **geometric separation** between the classes.

Recall the geometric margin from the previous parts:

$$
\gamma_i
=
\frac{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}{
\|\vec{w}\|
}
$$

Unlike the functional margin, this quantity accounts for the scale of $\vec{w}$.

Therefore, scaling $\vec{w}$ and $b$ does not arbitrarily increase the geometric distance.

This gives us a much more meaningful quantity to optimize.

## 12. Functional Gain vs Geometric Meaning

We now need to keep two ideas separate.

### Functional quantity

$$
\boxed{
G_i
=
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}
$$

This is useful because its sign tells us whether the classification is correct.

But it depends on the arbitrary scale of $\vec{w}$ and $b$.

---

### Geometric quantity

$$
\boxed{
\gamma_i
=
\frac{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}{
\|\vec{w}\|
}
}
$$

This represents the signed perpendicular distance from the decision boundary, adjusted according to the true class.

Therefore:

$$
\boxed{
\text{Functional quantity}
\neq
\text{Geometric quantity}
}
$$

The first is a scale-dependent score.

The second has geometric meaning.

## 13. What Should a Good Classifier Do?

Let's return to the original goal.

Suppose we have two possible decision boundaries.

### Boundary A

The training examples are correctly classified, but several points are very close to the boundary.

### Boundary B

The training examples are correctly classified and there is a larger gap between the two classes.

Intuitively, Boundary B is preferable.

Why?

Because its predictions have more separation from the decision boundary.

This motivates the central idea behind maximum-margin classification:

$$
\boxed{
\text{We want to maximize the geometric margin.}
}
$$

So the question has evolved:

$$
\text{Maximize gain}
$$

becomes:

$$
\text{Maximize meaningful separation}
$$

which leads to:

$$
\boxed{
\text{Maximize geometric margin}
}
$$

## 14. The Optimization Idea

For a linearly separable dataset, we want every training example to have a positive class-adjusted score.

More strongly, we want every example to be separated from the decision boundary by at least some positive amount.

We can express this idea using:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
\geq \gamma
$$

for every training example.

Here $\gamma$ represents a required margin level under an appropriate normalization.

Our conceptual goal is therefore:

$$
\boxed{
\text{Find the classifier with the largest possible margin.}
}
$$

This is the central idea behind the **maximum-margin classifier** and ultimately the **support vector machine**.

## 15. The Big Conceptual Shift

We started with a very simple idea:

> Correct classification should produce a positive value.

That gave us:

$$
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)>0
$$

Then we asked:

> Can we make this value larger?

This led us to the idea of gain:

$$
G_i
=
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
$$

But we discovered a problem.

Scaling $\vec{w}$ and $b$ can increase the functional gain without changing the decision boundary.

Therefore:

$$
\boxed{
\text{Functional gain alone is not enough.}
}
$$

We need a quantity that reflects actual geometric separation.

That quantity is the geometric margin:

$$
\boxed{
\gamma_i
=
\frac{
y_i\left(\vec{w}^{\,T}\vec{x}_i+b\right)
}{
\|\vec{w}\|
}
}
$$

This is the key conceptual transition from **classification correctness** to **maximum-margin classification**.

## 16. Python — Seeing the Scaling Problem

````python
import numpy as np

# One correctly classified example
y = 1
score = 5

# Original gain
gain = y * score

# Scale the parameters by different positive constants
scales = np.array([1, 2, 10, 100])

scaled_gains = scales * gain

print("Original gain:", gain)
print("Scales:", scales)
print("Scaled gains:", scaled_gains)
````

### Expected Output

```text
Original gain: 5
Scales: [  1   2  10 100]
Scaled gains: [  5  10  50 500]



The gain keeps increasing as the scale increases.

But remember:

Scaling $\vec{w}$ and $b$ by the same positive constant does not change the decision boundary.

Therefore, simply maximizing the functional gain does not solve our actual problem.

One important correction to our terminology

I want to make one thing explicit because we're aiming for a best-in-class ML notebook.

"Gain function" is useful as our pedagogical term here, but it is not the standard name used in the SVM literature
 for $y_i(\vec w^T\vec x_i+b)$. The standard term is functional margin. 
 
 Stanford CS229 defines exactly that quantity as the functional margin and then develops geometric margin and maximum-margin classification. (CS229 Machine Learning)

So I have intentionally introduced:

$[G_i=y_i(\vec w^{,T}\vec x_i+b)]$

as an intuitive gain, but I do not want the notebook to leave you believing that "gain function" is the formal SVM terminology.

The formal hierarchy is:

$$\boxed{\text{Class-adjusted score}
\text{Functional margin}}$$

and then:

$$\boxed{\text{Geometric margin}
\frac{\text{Functional margin}}{\|\vec{w}\|}}$$


That distinction will become very important in the next stages of Chapter 4.
