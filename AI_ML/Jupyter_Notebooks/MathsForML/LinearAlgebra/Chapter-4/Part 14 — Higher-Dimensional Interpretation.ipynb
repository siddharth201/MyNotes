# Chapter 4 — Part 14

# Higher-Dimensional Interpretation

> **Core Question:**  
> Everything we have visualized so far is easy to understand in one or two dimensions. But what happens when a Machine Learning dataset has many features?

In Machine Learning, a data point can have many features.

For example, a house could be represented using:

$$
\vec{x}
=
\begin{bmatrix}
\text{area}\\
\text{bedrooms}\\
\text{age}\\
\text{distance from city center}
\end{bmatrix}
$$

This is a four-dimensional feature vector.

We cannot directly visualize a four-dimensional space in the same way that we visualize a two-dimensional graph.

However, the mathematics of a linear classifier does not change.

The same expression still works:

$$
\boxed{
\vec{w}^{\,T}\vec{x}+b=0
}
$$

The important insight is:

> **The geometry generalizes to higher dimensions even when our ability to visualize it does not.**

## 1. From One Feature to Multiple Features

Let's start with one feature.

If we have a single feature $x$, a linear decision boundary can be written as:

$$
wx+b=0
$$

For two features:

$$
\vec{x}
=
\begin{bmatrix}
x_1\\
x_2
\end{bmatrix}
$$

and:

$$
\vec{w}
=
\begin{bmatrix}
w_1\\
w_2
\end{bmatrix}
$$

The decision boundary becomes:

$$
w_1x_1+w_2x_2+b=0
$$

which can be written compactly as:

$$
\boxed{
\vec{w}^{\,T}\vec{x}+b=0
}
$$

Now add another feature.

For three features:

$$
\vec{x}
=
\begin{bmatrix}
x_1\\
x_2\\
x_3
\end{bmatrix}
$$

and:

$$
\vec{w}
=
\begin{bmatrix}
w_1\\
w_2\\
w_3
\end{bmatrix}
$$

The boundary becomes:

$$
w_1x_1+w_2x_2+w_3x_3+b=0
$$

Again:

$$
\boxed{
\vec{w}^{\,T}\vec{x}+b=0
}
$$

The compact vector notation works regardless of the number of features.

## 2. What Changes as the Number of Features Increases?

The mathematical expression remains the same.

What changes is the dimensionality of the space.

| Number of features | Feature space | Linear decision boundary |
|---:|---|---|
| 1 | 1D | Point |
| 2 | 2D | Line |
| 3 | 3D | Plane |
| $n$ | $n$ D | $(n-1)$ D hyperplane |

The last row is the important one.

For an $n$-dimensional feature space, the linear decision boundary is an:

$$
\boxed{
(n-1)\text{-dimensional hyperplane}
}
$$

The word **hyperplane** is simply the general term we use for a flat linear boundary in an arbitrary-dimensional space.

## 3. One-Dimensional Case

Consider:

$$
x=0
$$

The feature space is one-dimensional.

There are two regions:

$$
x<0
$$

and:

$$
x>0
$$

The decision boundary itself is the single point:

$$
x=0
$$

So in one dimension:

$$
\boxed{
\text{1D space}
\rightarrow
\text{0D boundary}
}
$$

A point is therefore the one-dimensional version of a hyperplane.

## 4. Two-Dimensional Case

Now suppose we have two features:

$$
\vec{x}
=
\begin{bmatrix}
x_1\\
x_2
\end{bmatrix}
$$

A linear decision boundary has the form:

$$
w_1x_1+w_2x_2+b=0
$$

This is a line in two-dimensional space.

So:

$$
\boxed{
\text{2D space}
\rightarrow
\text{1D boundary}
}
$$

The line divides the plane into two regions.

The classifier score:

$$
\vec{w}^{\,T}\vec{x}+b
$$

tells us which side of the line the point lies on.

## 5. Three-Dimensional Case

Now suppose each example has three features:

$$
\vec{x}
=
\begin{bmatrix}
x_1\\
x_2\\
x_3
\end{bmatrix}
$$

The linear decision boundary is:

$$
w_1x_1+w_2x_2+w_3x_3+b=0
$$

This is a plane in three-dimensional space.

Therefore:

$$
\boxed{
\text{3D space}
\rightarrow
\text{2D boundary}
}
$$

The plane divides the three-dimensional feature space into two half-spaces.

The same score:

$$
\vec{w}^{\,T}\vec{x}+b
$$

still tells us which side of the plane the point lies on.

## 6. What About 4, 10, or 100 Features?

Now consider a dataset with many features.

For example:

$$
\vec{x}
=
\begin{bmatrix}
x_1\\
x_2\\
\vdots\\
x_{100}
\end{bmatrix}
$$

This point lives in a 100-dimensional feature space.

The weight vector is:

$$
\vec{w}
=
\begin{bmatrix}
w_1\\
w_2\\
\vdots\\
w_{100}
\end{bmatrix}
$$

The decision boundary is still:

$$
\boxed{
\vec{w}^{\,T}\vec{x}+b=0
}
$$

We cannot visualize this boundary directly.

But we do not need to.

The mathematical operations remain completely well-defined.

We can still calculate:

$$
\vec{w}^{\,T}\vec{x}+b
$$

and determine the sign of the result.

Therefore:

> **Visualization becomes difficult in high dimensions, but the mathematics does not break down.**

## 7. The Weight Vector Has a Geometric Meaning

The vector $\vec{w}$ is not just a collection of model parameters.

It has an important geometric interpretation.

The weight vector is **normal (perpendicular)** to the decision hyperplane.

Therefore:

$$
\boxed{
\vec{w}
\perp
\text{decision hyperplane}
}
$$

This is why $\vec{w}$ appears in the distance formula:

$$
\frac{
\vec{w}^{\,T}\vec{x}+b
}{
\|\vec{w}\|
}
$$

The norm:

$$
\|\vec{w}\|
$$

normalizes the score according to the length of the normal vector.

This converts the linear score into a perpendicular distance.

## 8. Why Does $\vec{w}$ Determine the Orientation?

Consider a two-dimensional boundary:

$$
w_1x_1+w_2x_2+b=0
$$

The vector:

$$
\vec{w}
=
\begin{bmatrix}
w_1\\
w_2
\end{bmatrix}
$$

is perpendicular to this line.

Changing $\vec{w}$ changes the orientation of the boundary.

For example:

$$
\vec{w}
=
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

produces a vertical boundary.

While:

$$
\vec{w}
=
\begin{bmatrix}
0\\
1
\end{bmatrix}
$$

produces a horizontal boundary.

Therefore:

$$
\boxed{
\vec{w}
\text{ determines the orientation of the decision boundary}
}
$$

while $b$ controls its position.

## 9. The Bias Controls Position

Consider:

$$
w_1x_1+w_2x_2+b=0
$$

The vector $\vec{w}$ determines the orientation of the boundary.

The bias $b$ shifts the boundary.

For example:

$$
x_1=0
$$

can be written as:

$$
x_1+0=0
$$

If we change the bias:

$$
x_1-3=0
$$

the boundary becomes:

$$
x_1=3
$$

The orientation is unchanged.

Only the position has changed.

Therefore:

$$
\boxed{
\vec{w}
\rightarrow
\text{orientation}
}
$$

$$
\boxed{
b
\rightarrow
\text{position}
}
$$

## 10. The Two Sides Still Exist

A hyperplane divides the feature space into two regions.

The decision boundary is:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

The two sides are:

$$
\vec{w}^{\,T}\vec{x}+b>0
$$

and:

$$
\vec{w}^{\,T}\vec{x}+b<0
$$

This is exactly the same idea we studied in one dimension.

The only difference is that the boundary is now a hyperplane instead of a point or line.

Therefore:

$$
\boxed{
\text{Positive score}
\rightarrow
\text{one side}
}
$$

$$
\boxed{
\text{Negative score}
\rightarrow
\text{the other side}
}
$$

The sign-based interpretation does not depend on the dimensionality of the feature space.

## 11. Distance Also Generalizes

The signed perpendicular distance from a point $\vec{x}$ to the hyperplane:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

is:

$$
\boxed{
d(\vec{x})
=
\frac{
\vec{w}^{\,T}\vec{x}+b
}{
\|\vec{w}\|
}
}
$$

This formula works in:

- 2D
- 3D
- 10D
- 100D
- and any other finite-dimensional feature space.

The sign tells us the side.

The magnitude tells us the perpendicular distance.

Therefore:

$$
\boxed{
\text{Sign}
\rightarrow
\text{which side}
}
$$

$$
\boxed{
\text{Magnitude}
\rightarrow
\text{how far}
}
$$

This is the same interpretation we developed earlier.

## 12. Class-Oriented Distance in Higher Dimensions

For a labeled example $(\vec{x}_i,y_i)$, the class-oriented geometric margin is:

$$
\boxed{
\gamma_i
=
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
}
$$

The expression combines:

### True class

$$
y_i\in\{-1,+1\}
$$

### Classifier score

$$
\vec{w}^{\,T}\vec{x}_i+b
$$

### Normalization

$$
\|\vec{w}\|
$$

Therefore, the same quantity still tells us:

- whether the point is on the correct side
- how far it is from the decision hyperplane
- measured in the perpendicular direction

Nothing fundamental changes when we move to higher dimensions.

## 13. A Four-Dimensional Example

Suppose a data point has four features:

$$
\vec{x}
=
\begin{bmatrix}
2\\
1\\
3\\
4
\end{bmatrix}
$$

and the weight vector is:

$$
\vec{w}
=
\begin{bmatrix}
1\\
-2\\
1\\
2
\end{bmatrix}
$$

Let:

$$
b=-3
$$

The classifier score is:

$$
\vec{w}^{\,T}\vec{x}+b
$$

Calculate the dot product:

$$
\vec{w}^{\,T}\vec{x}
=
(1)(2)
+
(-2)(1)
+
(1)(3)
+
(2)(4)
$$

Therefore:

$$
\vec{w}^{\,T}\vec{x}
=
2-2+3+8
$$

So:

$$
\vec{w}^{\,T}\vec{x}=11
$$

Now include the bias:

$$
\vec{w}^{\,T}\vec{x}+b
=
11-3
$$

Therefore:

$$
\boxed{
\vec{w}^{\,T}\vec{x}+b=8
}
$$

The score is positive.

Therefore, the point lies on the positive side of the decision hyperplane.

## 14. Calculate Its Geometric Distance

We already calculated:

$$
\vec{w}^{\,T}\vec{x}+b=8
$$

Now calculate:

$$
\|\vec{w}\|
$$

Since:

$$
\vec{w}
=
\begin{bmatrix}
1\\
-2\\
1\\
2
\end{bmatrix}
$$

we have:

$$
\|\vec{w}\|
=
\sqrt{
1^2+(-2)^2+1^2+2^2
}
$$

Therefore:

$$
\|\vec{w}\|
=
\sqrt{1+4+1+4}
$$

So:

$$
\boxed{
\|\vec{w}\|=\sqrt{10}
}
$$

The signed perpendicular distance is therefore:

$$
d(\vec{x})
=
\frac{8}{\sqrt{10}}
$$

Approximately:

$$
\boxed{
d(\vec{x})\approx2.53
}
$$

Even though the point lives in a four-dimensional space, we can still calculate its perpendicular distance from the decision hyperplane.

## 15. Add the Class Label

Suppose the true class of the four-dimensional point is:

$$
y_i=+1
$$

The geometric margin becomes:

$$
\gamma_i
=
\frac{
(+1)(8)
}{
\sqrt{10}
}
$$

Therefore:

$$
\boxed{
\gamma_i
=
\frac{8}{\sqrt{10}}
\approx2.53
}
$$

The positive value means that the point lies on the correct side of the decision hyperplane.

The magnitude tells us the perpendicular separation from the boundary.

## 16. What If the Point Belongs to the Negative Class?

Suppose the same point has:

$$
y_i=-1
$$

The classifier score is still:

$$
\vec{w}^{\,T}\vec{x}+b=8
$$

But now the class-oriented geometric margin becomes:

$$
\gamma_i
=
\frac{
(-1)(8)
}{
\sqrt{10}
}
$$

Therefore:

$$
\boxed{
\gamma_i
=
-\frac{8}{\sqrt{10}}
\approx-2.53
}
$$

The negative sign tells us that the point lies on the wrong side relative to its true class.

This demonstrates again why multiplying by $y_i$ is so useful.

The same geometric position can be interpreted differently depending on the true class.

## 17. Why We Cannot Visualize 100 Dimensions

Suppose:

$$
\vec{x}\in\mathbb{R}^{100}
$$

Each point has 100 features.

The decision boundary is a:

$$
99\text{-dimensional hyperplane}
$$

We cannot directly draw this in an ordinary three-dimensional graph.

But the mathematical representation remains simple:

$$
\vec{w}^{\,T}\vec{x}+b=0
$$

The classifier score remains:

$$
\vec{w}^{\,T}\vec{x}+b
$$

The geometric distance remains:

$$
\frac{
\vec{w}^{\,T}\vec{x}+b
}{
\|\vec{w}\|
}
$$

Therefore:

> **High dimensionality is primarily a visualization problem, not a mathematical problem.**

## 18. Features Are Coordinates

A useful way to think about high-dimensional Machine Learning data is:

> **Each feature is one coordinate axis.**

For example:

$$
\vec{x}
=
\begin{bmatrix}
x_1\\
x_2\\
x_3
\end{bmatrix}
$$

means that the point has three coordinates.

With 100 features:

$$
\vec{x}
=
\begin{bmatrix}
x_1\\
x_2\\
\vdots\\
x_{100}
\end{bmatrix}
$$

the point has 100 coordinates.

The vector $\vec{w}$ assigns a weight to each feature:

$$
\vec{w}
=
\begin{bmatrix}
w_1\\
w_2\\
\vdots\\
w_{100}
\end{bmatrix}
$$

The dot product:

$$
\vec{w}^{\,T}\vec{x}
$$

combines these feature contributions into a single scalar score.

This scalar score is what allows a high-dimensional point to be evaluated relative to the decision hyperplane.

## 19. The Dot Product as a Projection

The expression:

$$
\vec{w}^{\,T}\vec{x}
$$

can also be understood geometrically.

It measures how strongly $\vec{x}$ aligns with the direction of $\vec{w}$.

Recall the dot-product relationship:

$$
\vec{w}^{\,T}\vec{x}
=
\|\vec{w}\|
\|\vec{x}\|
\cos\theta
$$

where $\theta$ is the angle between the vectors.

Therefore, the classifier score:

$$
\vec{w}^{\,T}\vec{x}+b
$$

can be interpreted as measuring the position of $\vec{x}$ along the direction normal to the decision hyperplane, followed by a shift determined by $b$.

This is another way to understand why $\vec{w}$ is normal to the boundary.

## 20. The Geometry Survives in Higher Dimensions

Although we cannot visualize a 100-dimensional hyperplane, the geometric relationships remain valid.

We still have:

$$
\boxed{
\vec{w}
\perp
\text{decision hyperplane}
}
$$

We still have:

$$
\boxed{
\vec{w}^{\,T}\vec{x}+b=0
}
$$

We still have:

$$
\boxed{
\text{sign of score}
\rightarrow
\text{side of hyperplane}
}
$$

We still have:

$$
\boxed{
\frac{
\vec{w}^{\,T}\vec{x}+b
}{
\|\vec{w}\|
}
\rightarrow
\text{signed perpendicular distance}
}
$$

And for a labeled example:

$$
\boxed{
\frac{
y_i
\left(
\vec{w}^{\,T}\vec{x}_i+b
\right)
}{
\|\vec{w}\|
}
\rightarrow
\text{class-oriented geometric margin}
}
$$

The dimensionality changes.

The underlying geometry does not.

## 21. Machine Learning Perspective

In real Machine Learning problems, high-dimensional feature spaces are normal.

For example, an image can contain thousands or millions of pixel-related features.

A text representation can contain thousands of dimensions.

A recommendation system can represent users and items using hundreds or thousands of features.

We usually cannot visualize these spaces directly.

Nevertheless, a linear model can still operate on them using:

$$
\vec{w}^{\,T}\vec{x}+b
$$

The model does not need us to visualize the hyperplane.

It only needs to calculate the mathematical operations correctly.

This gives us an important Machine Learning lesson:

> **Understanding the geometry helps us understand the model, even when the geometry cannot be directly visualized.**
